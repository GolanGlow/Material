# <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Die arithmetischen Operatoren</strong>

- Viele arithmetische Operatoren sehen in C++ ähnlich aus wie in Python
- Wichtig: Bei Ganzzahlen verhält sich insbesondere die Division anders als in Python
- Sobald mindestens ein Operand ein Gleitkommatyp ist, erfolgt die Berechnung im Gleitkommaformat
- Einfache Beispiele zu den arithmetischen Operatoren in C++

**Additions-Operator (+)** 
- Wird verwendet, um zwei Werte zu addieren

```cpp
int a = 2;  
int b = 3;  
int c = a + b;
```
- Hier werden die Variablen `a` und `b` initialisiert und mithilfe des Operators `+` addiert
- Das Ergebnis wird in der Variable `c` gespeichert

**Subtraktions-Operator (-)**
- Wird verwendet, um einen Wert von einem anderen zu subtrahieren

```cpp
int a = 5;  
int b = 3;  
int c = a - b;
```
- Hier werden die Variablen `a` und `b` initialisiert und mithilfe des Operators `-` subtrahiert
- Das Ergebnis wird in der Variable `c` gespeichert

**Multiplikations-Operator (*)**
- Wird verwendet, um zwei Werte zu multiplizieren
```cpp
int a = 2;  
int b = 3;  
int c = a * b; // c = 6  
```
- Hier werden die Variablen `a` und `b` initialisiert und mithilfe des Operators `*` multipliziert
- Das Ergebnis wird in der Variable `c` gespeichert

**Divisions-Operator (/)**
- Wird verwendet, um einen Wert durch einen anderen zu teilen
```cpp
int a = 6;  
int b = 3;  
int c = a / b; // c = 2  
```
- Hier werden die Variablen `a` und `b` initialisiert und mithilfe des Operators `/` dividiert
- Das Ergebnis wird in der Variable `c` gespeichert

- **Wichtig bei Ganzzahlen:** Ganzzahldivision verwirft den Nachkommanteil und rundet in Richtung `0`

```cpp
int x = 7;
int y = 2;

int ganzzahl = x / y;         // 3
double gleitkomma = 7.0 / 2;  // 3.5
int negativ = -7 / 2;         // -3
```

- **Hinweis**
  - Bei Ganzzahltypen führen Division oder Modulo mit `0` zu undefiniertem Verhalten
    - Das Programm kann abstürzen, falsche Ergebnisse liefern oder sich scheinbar unauffällig verhalten
  - Vor einer Division sollte der Nenner daher immer geprüft werden
  - **Bei Gleitkommazahlen (`float`, `double`)** liefern viele heutige Systeme nach IEEE 754 Sonderwerte statt eines Absturzes:
    - `x / 0.0` ergibt typischerweise `inf` oder `-inf`, wenn `x != 0`
    - `0.0 / 0.0` ergibt typischerweise `nan` (Not a Number)
  - Auch diese Fälle sollten bewusst behandelt und nicht als normale Ergebnisse weiterverwendet werden

````c++
#include <iostream>

int main() {
    int numerator = 5;
    int denominator = 0;

    if (denominator != 0) {
        std::cout << numerator / denominator << std::endl;
    } else {
        std::cout << "Fehler: Division durch Null" << std::endl;
    }

    return 0;
}
````

Die oben genannten arithmetischen Operatoren können auch auf Variablen anderer Datentypen wie `float` oder `double` angewendet werden.
- Bei gemischten Typen führt C++ implizite Konvertierungen durch
- Häufig wird dabei in den Typ mit größerem Wertebereich konvertiert, z. B. von `int` zu `double`

```cpp
int a = 5;
double b = 2.0;
double c = a / b;   // c = 2.5
double d = a + b;   // d = 7.0
```

- Für benutzerdefinierte Typen können viele Operatoren überladen werden, dazu später mehr

**Modulo-Operator (%)**
- Wird verwendet, um den Rest einer Ganzzahldivision zu berechnen
```cpp
int a = 7;  
int b = 3;  
int c = a % b; // c = 1  
```
- Hier werden die Variablen `a` und `b` initialisiert und mithilfe des Operators `%` der Divisionsrest berechnet
- Das Ergebnis ist der Rest der Division, der in der Variable `c` gespeichert wird

- **Hinweis**: 
  - Der `%`-Operator kann in C++ nicht direkt auf `double`-Variablen angewendet werden
  - Dies liegt daran, dass der Modulo-Operator speziell für Ganzzahltypen wie `int`, `long` usw. entwickelt wurde
  - Bei negativen Operanden kann das Ergebnis ungewohnt sein: `-7 % 3` ergibt in C++ `-1`
  - Für Gleitkommazahlen kann `std::fmod(...)` aus `<cmath>` verwendet werden
  - Das Vorzeichen des Ergebnisses von `std::fmod(a, b)` folgt dem ersten Operanden `a`

````c++
#include <iostream>  
#include <cmath>  
  
int main() {  
    double a = 7.0;  
    double b = 2.5;  
    double c = std::fmod(a, b); 
     
    std::cout << c << std::endl;
    return 0;  
} 
````

### **Übersicht der arithmetischen Operatoren in C++**
| Operator | Beschreibung | Beispiel | Ergebnis |
|----------|--------------|-----------|------------|
| `+` | Addition | `c = a + b;` | `c = 5 + 3 → 8` |
| `-` | Subtraktion | `c = a - b;` | `c = 5 - 3 → 2` |
| `*` | Multiplikation | `c = a * b;` | `c = 5 * 3 → 15` |
| `/` | Division | `c = a / b;` | `c = 6 / 3 → 2` |
| `%` | Modulo | `c = a % b;` | `c = 7 % 3 → 1` |

**Kombinierte Zuweisungsoperatoren**
| Kurzform | Entspricht |
|----------|------------|
| `x += y` | `x = x + y` |
| `x -= y` | `x = x - y` |
| `x *= y` | `x = x * y` |
| `x /= y` | `x = x / y` |
| `x %= y` | `x = x % y` |

**Operatorrangfolge**
- `*`, `/` und `%` binden stärker als `+` und `-`
- Klammern machen die Auswertungsreihenfolge explizit und verbessern die Lesbarkeit

```cpp
int a = 2 + 3 * 4;    // 14
int b = (2 + 3) * 4;  // 20
```

**Häufige Fehlerquellen**
- `7 / 2` ergibt mit `int` den Wert `3`, nicht `3.5`
- Division oder Restbildung mit `0` ist bei Ganzzahltypen nicht erlaubt
- `%` funktioniert nur für Ganzzahltypen; für `double` verwendet man `std::fmod(...)`
- Bei längeren Ausdrücken helfen Klammern, Missverständnisse zu vermeiden

# <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Inkrement/Dekrement</strong>

In C++ gibt es zwei Arten von Inkrement- und Dekrement-Operatoren
  - Prä-Inkrement/-Dekrement-Operator (`++x`, `--x`)
    - Dieser erhöht/dekrementiert den Wert einer Variablen um eins und gibt den *neuen Wert* der Variablen zurück
  - Post-Inkrement/-Dekrement-Operator (`x++`, `x--`)
    - Dieser erhöht/dekrementiert auch den Wert der Variablen um eins, gibt aber den *alten Wert* der Variablen zurück

- Diese Operatoren ändern den Wert immer um genau `1`
- Sie werden häufig bei Zählern und Iteratoren verwendet
- **Praxisregel:** `++` und `--` möglichst als eigene Anweisung verwenden und nicht mehrfach in komplexen Ausdrücken kombinieren
  - Ausdrücke wie `x++ + ++x` sind fehleranfällig und sollten vermieden werden

### Unterschied: Präfix vs. Postfix

| Ausdruck  | Bedeutung | Beispiel |
|-----------|--------------|------------|
| `++x` (Präfix) | Erhöht `x`, dann verwendet den neuen Wert | `int x = 5; int y = ++x; // x = 6, y = 6` |
| `x++` (Postfix) | Verwendet den alten Wert, dann erhöht `x` | `int x = 5; int y = x++; // x = 6, y = 5` |

````c++
#include <iostream>  
  
int main() {  
    int x = 0;  
    int y;
  
    y = ++x; // pre-increment
    std::cout << "x = " << x << ", y = " << y << std::endl; 

    x = 0;
    y = x++; // post-increment  
    std::cout << "x = " << x << ", y = " << y << std::endl;
  
    x = 0;
    y = --x; // pre-decrement
    std::cout << "x = " << x << ", y = " << y << std::endl;
  
    x = 0;
    y = x--; // post-decrement  
    std::cout << "x = " << x << ", y = " << y << std::endl;
  
    return 0;  
} 
````
