 # <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Einführung zu Selektionen</strong>

 - Bisher wurden Programme betrachtet, bei denen alle Anweisungen sequenziell, also nacheinander in der festgelegten Reihenfolge ausgeführt wurden
- In der Praxis ist es jedoch häufig erforderlich, einzelne Anweisungen zu überspringen, zwischen verschiedenen Anweisungsgruppen zu wählen oder auf unterschiedliche Bedingungen und Eingaben flexibel zu reagieren
- Dies wird durch sogenannte Kontrollstrukturen ermöglicht
  - Kontrollstrukturen lenken die Ausführung des Programms abhängig von Bedingungen und Eingabewerten
  - Sie sind ein zentrales Konzept in der Programmierung und erlauben es, komplexere und dynamisch reagierende Programme zu schreiben

**Beispiele für Kontrollstrukturen**
- Selektionen (Bedingungen, z.B. `if`, `else`, `switch`)
- Schleifen (Wiederholungen, z.B. `for`, `while`)

### **Selektionen in Python**

- In Python werden Selektionen (bedingte Anweisungen) mit der `if`-Anweisung realisiert
- Sie ermöglichen, bestimmte Code-Blöcke nur dann auszuführen, wenn eine vorgegebene Bedingung erfüllt ist
- Die Syntax der einfachen `if`-Anweisung in Python lautet:

```python
if condition:
    # do something, i.e. body of if statement
```


### **Selektionen in C/C++**

Die `if`-Anweisung funktioniert in C/C++ im Wesentlichen ähnlich wie in Python, weist jedoch einige syntaktische Unterschiede auf
  - Die Syntax der `if`-Anweisung in C/C++ lautet:
  
```cpp
if(condition) {
    // do something
}
```

- In C/C++ werden Anweisungen in Blöcken zusammengefasst
  - Ein Block wird durch geschweifte Klammern `{ ... }` definiert
  - Die Einrückung und Formatierung des Codes dient nur der Lesbarkeit und hat für den Compiler keine Bedeutung
  - Häufige Fehler entstehen durch vergessene `{}`-Klammern oder fehlende bzw. überflüssige Semikolons (`;`)
- Wenn nur eine einzelne Anweisung zum Block einer `if`-Anweisung gehört, können die geschweiften Klammern weggelassen werden:
```cpp
if (condition)
    std::cout << "True" << std::endl;
```
**Empfohlene Praxis**
- Verwenden Sie trotzdem stets geschweifte Klammern, um Fehler durch spätere Erweiterungen oder Änderungen am Code zu vermeiden:
```cpp
if (condition) {
    std::cout << "True" << std::endl;
}
```

````c++
#include <iostream>

int main() {
    if(0 < -1) {
        std::cout << "Hello ";
        std::cout << "World!" << std::endl;
    }

    return 0;
}
````

Wesentliche Unterschiede zwischen C/C++ und Python:
- Bedingungen (*condition*)
  - In C/C++ müssen Bedingungen innerhalb von runden Klammern `(...)` stehen
  - In Python sind diese Klammern nicht notwendig
- Doppelpunkt (`:`)
  - In Python muss am Ende der Bedingung ein Doppelpunkt (`:`) gesetzt werden
  - In C/C++ entfällt der Doppelpunkt
- Gruppierung von Anweisungen (Codeblöcke)
  - In C/C++ erfolgt die Gruppierung von Anweisungen mithilfe von geschweiften Klammern `{ ... }`; die Einrückung ist optional und dient nur der besseren Lesbarkeit
  - In Python werden Codeblöcke ausschließlich durch Einrückung definiert; die Einrückung ist daher zwingend erforderlich

 # <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Selektionen in Python</strong>

**Python `if...else`-Anweisung**
- Die Syntax der `if`...`else`-Anweisung in Python lautet:

```python
if bedingung:
    # Rumpf der if-Anweisung
else:
    # Rumpf der else-Anweisung
```

**Python `if...elif...else`-Anweisung**
- Wenn eine Auswahl zwischen mehr als zwei Alternativen getroffen werden muss, wird die `if`...`elif`...`else`-Anweisung verwendet

```python
if bedingung1:
    # Rumpf der if-Anweisung (Anweisungsblock)
elif bedingung2:
    # Rumpf der elif-Anweisung
else:
    # Rumpf der else-Anweisung
```

 # <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Selektionen in C/C++</strong>

Die einzelnen *if*-Anweisungen können auch beliebig verschachtelt werden
```cpp
if (<bedingung1>) {
  // etwas tun, wenn bedingung1 erfüllt ist
  if (<bedingung2>) {
    // etwas tun, wenn bedingung1 und bedingung2 erfüllt sind
  }
  else {
    // etwas tun, wenn bedingung1 erfüllt ist und bedingung2 NICHT erfüllt ist
  }
}
else {
  // etwas tun, wenn bedingung1 NICHT erfüllt ist
}
```

# <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Bedingungen</strong>
C/C++ unterstützt gleich wie Python mehrere wichtige Vergleichsoperatoren:

|Operator |Bedeutung       |Ergebnis: **`True`** | Ergebnis: **`False`**|
|:-------:|:--------------:|:--------------:|:------------:|
| `<`     | Kleiner        | `2.1 < 3.2`    | `32 < 3`     |
| `<=`    | Kleiner gleich | `6 <= 6`       | `6.9 <= 6`   |
| `>`     | Größer         | `8 > 3.87`     | `32 > 32.3`  |
| `>=`    | Größer gleich  | `3 >= 2`       | `7 >= 7.22`  |
| `==`    | Gleich         | `1.2 == 1.2`   | `2 == 1.999` |
| `!=`    | Ungleich       | `2.1 != 3.2`   | `1.2 != 1.2` |


****

## <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Logische Operatoren</strong>
- Bedingungen können wiederum mit den sogenannten logischen Operatoren zu komplexeren Bedingungsausdrücken verknüpft werden
- Diese Operatoren sind `&&` (und), `||` (oder) und `!` (nicht)

|Logischer Operator Python | Logischer Operator C/C++ |Bedeutung  |Beispiel               | Ergebnis|
|:------------------------:|:---------:|:---------------------:|:-------:|:-------:|
| `and`                    | &&        | und |`-0.4 < 1 && 3 > 0`  | `True`  |
| `or`                     | \|\|        | oder | `-0.4 > 1 \|\| 3 > 0` | `True`  |
| `not`                    | !         | nicht | `!(0.1 < 3)`         | `False` |

**WICHTIG!** Es ist zu beachten, dass die Verwendung von einfachen `&` bzw. `|` eine bitweise und bzw. oder Operation in C/C++ bedeutet und andere Ergebnisse liefert

## <strong style="color: rgb(0, 102, 153); opacity: 1.0;">`switch ... case` Anweisung</strong>
- Selektives Ausführung verschiedener Programmteile

````c++
#include <iostream>

using namespace std;

int main() {
    long monat = 3;
    
    if (monat == 1) {
        cout << "January";
    } 
    else if (monat == 2) {
        cout << "February";
    } 
    else if (monat == 3) {
        cout << "March";
    } 
    else if (monat == 4) {
        cout << "April";
    } 
    else if (monat == 5) {
        cout << "May";
    } 
    else if (monat == 6) {
        cout << "June";
    } 
    else if (monat == 7) {
        cout << "July";
    } 
    else if (monat == 8) {
        cout << "August";
    } 
    else if (monat == 9) {
        cout << "September";
    } 
    else if (monat == 10) {
        cout << "October";
    } 
    else if (monat == 11) {
        cout << "November";
    } 
    else if (monat == 12) {
        cout << "December";
    } 
    else {
        cout << "Etwas ist schief gelaufen!";
    }

    return 0;
}
````

- Obiger Code lässt sich mittels `switch ... case` Anweisung kompakter darstellen
```cpp
switch (condition)
{
    case expression: statement
    ...
}
```
  - Dabei dürfen Sprungmarken nur ganze Zahlen sein, z.B. int, long, char
  - Wurde zu einer Sprungmarke verzweigt, wird die Ausführung unabhängig von weiteren Sprungmarken fortgesetzt, bis `break` eintritt oder der `switch` Block beendet wird

````c++
#include <iostream>

using namespace std;

int main() {
    long monat = 3;
    
    switch(monat) {
    case 1:
        cout << "January";
        break;
    case 2:
        cout << "February";
        break;
    case 3:
        cout << "March";
        break;
    case 4:
        cout << "April";
        break;
    case 5:
        cout << "May";
        break;
    case 6:
        cout << "June";
        break;
    case 7:
        cout << "July";
        break;
    case 8:
        cout << "August";
        break;
    case 9:
        cout << "September";
        break;
    case 10:
        cout << "October";
        break;
    case 11:  
        cout << "November";
        break;
    case 12:
        cout << "December";
        break;
    default: 
        cout << "Etwas ist schief gelaufen!";
    }

    return 0;
}
````

**Abschließende Anmerkungen zu Selektionen in C/C++**
- Die Effizienz der Kontrollstrukturen `if-else` und `switch-case` in C++ hängt stark von der spezifischen Anwendung und der verwendeten Compiler-Optimierung ab
- Prinzipiell sind beide Varianten korrekt und vielfach austauschbar, allerdings
  - Bei einer größeren Anzahl möglicher Bedingungen ist der Einsatz von `switch-case` oft effizienter, da hier vom Compiler unter Umständen eine sogenannte *Jump Table* generiert werden kann
  - Eine *Jump Table* ermöglicht es, direkt auf die entsprechende Anweisung zu springen, ohne jede Bedingung einzeln prüfen zu müssen
- Vorteil von `switch-case`
  - Effizientere Ausführung bei vielen diskreten Fällen
  - Potenziell schnellere Laufzeit durch direkte Sprünge
- Empfehlung für die Praxis
  - Verwenden Sie bei wenigen oder komplexen Bedingungen (Bereiche, nicht diskrete Werte) besser `if-else`
  - Nutzen Sie `switch-case`, wenn Sie eine übersichtliche, klar abgegrenzte Anzahl von Fällen haben (z.B. diskrete Zahlenwerte, Enumerationen)
- Lesbarkeit und Wartbarkeit sind entscheidend
  - Verwenden Sie immer jene Struktur, die am besten zur klaren Darstellung der Programmabsicht beiträgt
  - Klarer Code reduziert Fehler und erleichtert langfristig die Wartung
