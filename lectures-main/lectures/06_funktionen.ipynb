 # <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Allgemeines zu Funktionen</strong>

**Warum Funktionen?**
- Wiederverwendbarkeit: Einmal definierte Funktionen können mehrfach genutzt werden.
- Strukturierung des Codes: Der Code wird in logische Einheiten unterteilt.
- Wartbarkeit: Änderungen müssen nur an einer Stelle vorgenommen werden.
- Abstraktion: Wenn wir wissen, was eine Funktion macht, ist es uns egal, wie sie es macht

**Funktionen helfen bei**
- der Strukturierung des Programms
- der Programmierung von in sich abgeschlossenen definierten Teilaufgaben
- der Erweiterung der Funktionalität von bestehendem Programmcode

**Funktionen in C/C++**
- Eine Funktion ist ein Code-Block, der (meistens) Eingaben erhält, Berechnungen durchführt und (meistens) Ausgaben liefert
- Die Syntax einer Funktion: `Rückgabetyp Funktionsname( Parameterliste mit ihren Typen );`
```cpp
void nothing_to_give();
long round(double x);
double multiply(double x, double y);
```

**Im Vergleich dazu obige Funktionen in Python**
```Python
import numpy as np

def nothing_to_give():
  pass

def round(x):
  return np.floor(x+0.5)

def multiply(x, y):
  return x*y
```

 # <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Erstellen einer Funktion</strong>

**Definition einer Funktion**
 - Achtung: Hier ohne `;`
```cpp
long meineFunktion(double d, long l) {
  long etwas;
  // etwas aus d und l berechnen
  return etwas;
}
```

- Angabe eines eindeutigen Funktionsnamens, des Rückgabewerts, der Parameter und des kompletten Funktionsrumpfes (Inhalt)
```cpp
Typ Funktionsname(Parameterliste mit ihren Typen) {
  // Funktionsrumpf
}
```
- Da die Funktionsdefinition ausführbaren Programmcode enthält, muss sie sich in einer `.cpp`-Datei befinden!

**Deklaration einer Funktion**
- Die Deklaration (Bekanntmachung) einer Funktion informiert den Compiler über ihre Existenz.  
- Syntax: Eine Funktionsdeklaration endet mit einem Semikolon (`;`):  
  ```cpp
  long meineFunktion(double d, long l);
  ```
- Die Deklaration einer Funktion besteht aus:
  - Dem Rückgabetyp
  - Einem eindeutigen Funktionsnamen
  - Den Parametern
- Die Funktionsdeklaration befindet sich in der Regel in einer `.hpp`-Datei (Header-Datei).
- Eine Funktion muss vor ihrem Aufruf entweder deklariert oder definiert sein.
- Beim Linken (Erstellung des ausführbaren Programms) muss die Funktionsdefinition eingebunden sein.

**Beispiel** einer Funktion welche zwei Zahlen addiert:

````c++
#include <iostream>

// Function declaration (tells the compiler that this function exists)
double addNumbers(double a, double b);

// Function definition (actual implementation)
double addNumbers(double a, double b) {
    return a + b;
}

int main() {
    double result;
    double num1 = 3.25, num2 = 4.67;

    result = addNumbers(num1, num2); // Function call
    std::cout << "Result: " << result << std::endl;

    return 0;
}
````

**Parameter einer Funktion**
- Funktionen können eine beliebige Anzahl von Parametern haben, die in der Parameterliste durch Beistriche (`,`) getrennt sind.  
- In **C** und **C++** werden Argumente standardmäßig als Kopie übergeben (*Call by Value*).  
  - Das bedeutet, dass innerhalb der Funktion eine zusätzliche Kopie des Arguments existiert.  
  - Änderungen innerhalb der Funktion beeinflussen die Originalvariable im aufrufenden Code nicht.
- In **C++** können Parameter auch per Referenz übergeben werden (*Call by Reference*).  
  - Dadurch arbeitet die Funktion direkt mit den Originaldaten, sodass Änderungen sichtbar bleiben.  
- Alternativ kann ein *Call by Reference* auch durch Zeiger (`*` bzw. `&`) realisiert werden:  
  - In C ist dies die einzige Möglichkeit, eine Funktion mit veränderbaren Originaldaten zu nutzen.  
- Referenzen und Zeiger werden in einem späteren Abschnitt genauer behandelt.  

**Rückgabetyp einer Funktion**
- Jede Funktion in **C/C++** muss einen Rückgabetyp haben.  
```cpp 
long meineFunktion(double d, long l);
```
- Das `long` vor `meineFunktion` gibt an, dass die Funktion einen Wert vom Typ `long` zurückgibt.
- Der Rückgabetyp kann ein beliebiger grundlegender oder abgeleiteter Datentyp sein.
- Eine Funktion kann auch den Rückgabetyp `void` haben, wodurch sie keinen Wert zurückgibt.
- Falls ein Rückgabetyp angegeben ist, muss die Funktion mit return einen entsprechenden Wert zurückgeben.
- In C/C++ kann eine Funktion nur einen einzigen Rückgabewert liefern.
  - Mehrere Werte können jedoch in einer Struktur (`struct` oder `class`) zusammengefasst und zurückgegeben werden.

````c++
#include <iostream>

const double PI = 3.141592653589793;

// Function that calculates the area of a circle
double circleArea(double radius) {
    return radius * radius * PI;
}

// Function that prints the area (demonstrating a void function)
void printArea(double area) {
    std::cout << "Area = " << area << std::endl;
}

int main() {
    double radius = 1.0;
    double area = circleArea(radius);
    
    printArea(area); // Output result
    
    return 0;
}
````

 # <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Funktionen und Sichtbarkeit von Variablen</strong>

**Globale Variablen**
- Variablen, die außerhalb von Funktionen definiert sind, sind global und gelten in allen Funktionen.  
- Diese Variablen werden beim Programmstart erzeugt und initialisiert.  
- Beim Programmende werden sie automatisch zerstört.  
- Globale Variablen werden standardmäßig mit `0` initialisiert, sollten aber dennoch immer explizit initialisiert werden.  

**Lokale Variablen**
- Innerhalb einer Funktion definierte Variablen werden als lokale Variablen bezeichnet.  
- Auf lokale Variablen kann außerhalb der Funktion nicht zugegriffen werden.  
- Lokale Variablen werden erst erzeugt, wenn ihre Definition ausgeführt wird (sie sind standardmäßig uninitialisiert).  
- Sobald der Codeblock verlassen wird, endet die Lebensdauer der Variable.  
- Der Speicherbereich wird automatisch freigegeben, und ein späterer Zugriff darauf ist ungültig und nicht definiert.  

**Statische Variablen**
- Lokale Variablen können als `static` definiert werden, wodurch sie ihren Wert zwischen Funktionsaufrufen behalten.  
- Statische Variablen werden nur einmal erzeugt und initialisiert, wenn ihre Definition das erste Mal durchlaufen wird.  
- Die Variable und ihr Inhalt bleiben bis zum Programmende erhalten.  
- Um eine Variable als statische Variable zu definieren, wird das Schlüsselwort `static` vorangestellt:  
  ```cpp
  void exampleFunction() {
      static int counter = 0; 
      // Wird beim ersten Funktionsaufruf initialisiert und behält den Wert zwischen weiteren Aufrufen
      counter++;
      std::cout << "Counter: " << counter << std::endl;
  }
  ```

***
### **Ein Beispiel zur Gültigkeit von Variablen**

````c++
#include <iostream>

int globalVar = 1; // Global variable

void myfunc() {
    static int staticVar = 1; // Retains value between function calls
    int localVar = 1;         // Reset to 1 every function call

    std::cout << "Static: " << staticVar;
    std::cout << " Global: " << globalVar;
    std::cout << " Local: " << localVar << std::endl;

    staticVar++; // Increments and keeps its value for the next call
    globalVar++; // Also increments globally
    localVar++;  // Resets to 1 each time
}

int main() {
    myfunc();
    myfunc();
    myfunc();

    // localVar  = 8; // Compiler error: localVar only exists inside `myfunc`
    // staticVar = 8; // Compiler error: staticVar is function-local
    globalVar = 8; // OK: Global variable can be changed anywhere

    myfunc(); // Call again to observe static behavior

    return 0;
}
````
