# <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Was ist `std::function`?</strong>

- `std::function` ist ein `class`-Template (Klassenvorlage) die seit C++11 bereitgestellt wurde
- Eine Instanz von `std::function` kann ein aufrufbares Funktionsobjekt (engl. *Callable*) speichern, kopieren und aufrufen
    - Aufrufbare Objekte können Funktionen, Lambdas (anonyme Funktionen, Funktionsliterale) oder andere Funktionsobjekte, sowie Zeiger auf Member-Funktionen und Zeiger auf Datenmember sein (Mehr zum Thema Zeiger folgt später)
    - Für unsere Anwendungsfälle werden aufrufbare Objekte auf Funktionen und Lambdas beschränkt
    - Wir werden `std::function` verwenden, um diese aufzurufen

## <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Deklaration von `std::function`</strong>
Die Syntax zur Einrichtung einer `std::function` lautet:<br /> `std::function<returnType(parameterTypes...)> myFunction`
- Der Inhalt der spitzen Klammern `<>` entspricht genau einer Funktionssignatur:<br /> `returnType myFunctionName(parameterTypes...)`
- `myFunction` wird daher verwendet, um ein aufrufbares Funktionsobjekt (engl. *Callable*) mit dieser Funktionssignatur aufzurufen:
    - Der Rückgabetyp der Funktion muss `returnType` sein.
    - Die aufzurufende Funktion sollte die Anzahl der Argumente und deren Typen gemäß `parameterTypes` haben
- Um `std::function` zu verwenden, muss der `functional`-Header mit `#include <functional>` eingefügt werden

## <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Beispiel für den Aufruf einer Funktion mit `std::function`</strong>

```c++
// Very simple use case of calling functions using std::function
#include <iostream>    // std::cout
#include <functional>  // std::function

// Define a function to print a value
void printv(int v) {
    std::cout << v << std::endl;
}

int main() {
    // Define a std::function called func which can refer to
    // any callable with the signature void(int)
    std::function<void(int)> func;

    // Set func to refer to printv()
    func = printv;

    // Call func while passing the value 5
    func(5);

    return 0;
}
```

## <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Vorteile</strong>
- Bietet Flexibilität, indem es heterogene (verschiedene Datentypen) Sammlungen von aufrufbare Objekte (engl. *Callables*) erlaubt
- Hilft bei der generischen Programmierung und Code-Entkopplung
- Verbessert die Lesbarkeit und Wartbarkeit von Code
- Ermöglicht Laufzeit-Polymorphismus mit Typ-Erasure (Mehr zum Thema Zeiger folgt später)

## <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Ein Beispiel für den Aufruf mehrerer Funktionen mit `std::function`</strong>

```c++
// Simple use case of calling functions using std::function
#include <iostream>    // std::cout
#include <functional>  // std::function

// Define a function to return the sum of two integers
auto add(int a, int b) {
    int sum = a + b;
    std::cout << a << " + " << b << " = " << sum << std::endl;
    return sum;
}

// Define a function to return the difference between two integers
auto subtract(int a, int b) {
    int difference = a - b;
    std::cout << a << " - " << b << " = " << difference << std::endl;
    return difference;
}

int main() {
    // Define a std::function called calc which can refer to
    // any callable with the signature int(int, int)
    std::function<int(int, int)> calc;

    calc = add;                          // set calc to refer to the add function
    auto result = calc(5, 3);            // execute calc, which will call add
    std::cout << "result = " << result << std::endl;

    calc = subtract;                     // set calc to reference the subtract function
    result = calc(5, 3);                 // execute calc, which will call subtract
    std::cout << "result = " << result << std::endl;

    return 0;
}
```

# <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Lambdas</strong>
- `std::function` wird oft in Kombination mit Lambdas verwendet
- Lambda-Funktionen gibt es auch in Python, aber sie wurden im Wintersemester nicht behandelt
- Schauen wir uns zuerst ein Beispiel an wofür Lambda-Funktionen verwendet werden können

```c++
#include <iostream>   // std::cout, std::endl
#include <vector>     // std::vector
#include <functional> // std::function

// std::function<void(int)> can be used to refer to any function or lambda
// which takes one int argument and does not return a value (void)
void doSomething(const std::vector<int>& values, std::function<void(int)> func) {
    for (int value : values){
        func(value);          // what do we actually do here?
    }
}

int main() {
    std::vector<int> values = { 1, 2, 3, 4 };
    doSomething(values, [](int value){ std::cout << "Value = " << value << std::endl; });
    // What just happened?
    
    auto myLambda = [](int value){ std::cout << "Value = " << value << std::endl; };
    doSomething(values, myLambda); // Here we pass the function as if it was a variable
    
    return 0;
}
```

- Im obigen Beispiel verschiebt die bereitgestellte Funktion `doSomething` die auszuführende Aktion auf `func`
- Wir können dann inline bei Bedarf entscheiden, was `func` tatsächlich bewirken soll, indem wir eine Lambda-Funktion verwenden
- Die Lambda-Funktion kann als Argument an `doSomething` übergeben werden
- In dem obigen Beispiel hat die Lambda-Funktion keinen Rückgabetyp

## <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Was sind Lambdas?</strong>
- Lambdas sind kleine Inline-Funktionen
- Lambdas bieten die Möglichkeit, bei Bedarf schnell und einfach eine Funktion zu definieren, ohne eine tatsächliche Funktion separat definieren zu müssen
- Sie ermöglichen es uns unter anderem, eine Funktion als Variable zu behandeln, so dass wir sie als Parameter an eine andere Funktion übergeben können
- Das bedeutet, dass wir sie direkt als Argument einer Funktion übergeben können
- Die Funktion fängt dann die Lambda-Funktion, z.B. mit einem `std::function`-Datentyp ab
- Wichtig dabei ist, dass die Signatur der Lambda-Funktion gleich der Signatur des `std::function`-Elements in der Deklaration ist

```c++
#include <iostream>  // std::cout, std::endl;

int main() {
    // Define a lambda function which returns the sum of two integers
    auto add = [](int a, int b) -> int { return a + b; };
    // -> int notes the return type of the lambda functional
    // If this should be deduced automatically "auto" is not required

    auto addxy = [](int x, int y){ return x + y; };

    // Call the lambda function and store the result
    int result = add(3, 5);
    int resultxy = addxy(3, 5);

    // Print the result
    std::cout << "Result: " << result << std::endl;
    std::cout << "Result x, y: " << resultxy << std::endl;

    return 0;
}
```

## <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Vorteile</strong>
- **Übersichtlichkeit**: Mit Lambdas können kleine Inline-Funktionen direkt am Verwendungsort definiert werden, wodurch die Notwendigkeit separater Funktionsdeklarationen und -definitionen entfällt
- **Lokale Kontexterfassung**: Lambdas können Variablen aus ihrem umschließenden Bereich per Referenz oder Wert erfassen
- **Flexibilität**: Lambdas können überall dort verwendet werden, wo ein Funktionsobjekt erwartet wird, was funktionale Programmierparadigmen ermöglicht und generische Programmierpraktiken fördert
- **Closure**: Lambdas erstellen Closures, die sowohl den Code als auch seinen Kontext (d. h. die erfassten Variablen) kapseln. Dies erleichtert die Erstellung und Weitergabe von Funktionen, die von einem bestimmten Zustand abhängen, ohne dass dieser Zustand für den globalen Bereich offengelegt wird

```c++
// Example of a lambda function which returns a multiple by a previously-defined factor
#include <iostream>  // std::cout, std::endl

int main() {
    // Define a factor
    int factor = 5;

    // Define a lambda function that captures 'factor' by value
    auto multiplyByFactor = [factor](int x){ return x * factor; };

    // Use the closure to multiply a number
    std::cout << "Multiplying 10 by " << factor << " gives: " << multiplyByFactor(10) << std::endl;

    return 0;
}
```

- Im obigen Beispiel kann `multiplyByFactor` auf die Variable `factor` zugreifen, obwohl `factor` außerhalb der Funktion definiert ist
- Die Variable `faktor` wird in der Aufnahmeliste der Lambda-Funktion `[]` erfasst

- **Geringerer Aufwand bei der Namensgebung**: Da Lambdas keinen eigenen Namen benötigen, können sie direkt verwendet werden, ohne zusätzliche Funktionsnamen im jeweiligen Namensraum zu belegen
- **Inline-Optimierung**: Compiler können Lambdas so optimieren, dass sie inline sind, wodurch der Overhead von Funktionsaufrufen reduziert wird, insbesondere bei kleinen, häufig verwendeten Funktionen

## <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Verwendung von `std::function` und Lambdas mit einem Rückgabewert</strong>

```c++
// Example: Using std::function with a return value and additional logic
#include <iostream>     // std::cout, std::endl
#include <functional>   // std::function

// This function only applies the operation if both inputs are positive
// -> The function controls WHEN the operation is executed
// -> The lambda controls WHAT operation is executed
int applyIfPositive(int a, int b, std::function<int(int, int)> op) {
    if (a > 0 && b > 0) {
        return op(a, b);  // call the passed lambda/function
    }
    return 0; // fallback if condition is not met
}

int main() {
    // Lambda that multiplies two numbers
    auto multiply = [](int a, int b) {
        return a * b;
    };

    // Lambda that adds two numbers
    auto add = [](int a, int b) {
        return a + b;
    };

    // Both values are positive -> operation is executed
    std::cout << applyIfPositive(5, 3, multiply) << std::endl;

    // One value is negative -> operation is NOT executed
    std::cout << applyIfPositive(-5, 3, add) << std::endl;

    return 0;
}
```

## <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Vektor von `std::function`-Elementen</strong>
- Mit `std::function` können wir verschiedene Lambda-Ausdrücke in einer Liste speichern
- Zu diesem Zweck können wir einen Vektor aus `std::function`-Elementen erstellen
- Wir können dann durch den Vektor iterieren und bei jedem Index die gewünschte Lambda-Funktion ausführen

```c++
#include <iostream>
#include <vector>
#include <functional>

int main() {
    // vector storing different operations
    std::vector<std::function<void(int)>> operations = {
        [](int x) { std::cout << "value: " << x << std::endl; },
        [](int x) { std::cout << "square: " << x * x << std::endl; }
    };

    // add another operation using push_back
    operations.push_back([](int x) {
        std::cout << "double: " << 2 * x << std::endl;
    });

    // iterate over functions and apply them
    for (const auto& op : operations) {
        op(5);
    }

    return 0;
}
```

## <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Zurück zum Sortierbeispiel</strong>

```c++
#include <iostream>    // std::cout, std::endl
#include <vector>      // std::vector
#include <functional>  // std::function
#include <algorithm>   // std::sort

// Define a struct representing a person
struct Student {
    int id;
    char grade;
};

// Existing sorting function by ID
void sortByID(std::vector<Student>& s) {
    std::sort(s.begin(), s.end(), [](Student a, Student b) {
        return a.id < b.id;
    });
}

int main() {
    std::vector<Student> students = {
        {12345, 'C'},
        {32135, 'B'},
        {21575, 'F'},
        {78725, 'A'}
    };

    // std::function to hold "any sorting strategy"
    std::function<void(std::vector<Student>&)> sort;

    // Use existing function
    sort = sortByID;
    sort(students);

    std::cout << "Sorted by ID:\n";
    for (const auto& student : students) {
        std::cout << student.id << ", " << student.grade << std::endl;
    }

    std::cout << std::endl;

    // Now define a NEW behavior locally using a lambda
    sort = [](std::vector<Student>& s) {
        std::sort(s.begin(), s.end(), [](Student a, Student b) {
            return a.grade < b.grade;
        });
    };

    // Apply new behavior
    sort(students);

    std::cout << "Sorted by grade (lambda):\n";
    for (const auto& student : students) {
        std::cout << student.id << ", " << student.grade << std::endl;
    }

    return 0;
}
```
