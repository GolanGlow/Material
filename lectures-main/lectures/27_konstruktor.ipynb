## Konstruktor (*engl. constructor*)

Im vorigen Beispiel hat man schon gesehen, dass es oftmals mühsam wird, alle Membervariablen einer Klasse einzeln zu initialisieren. Hierzu werden Konstruktoren angeboten.

Der Konstruktor

* ist eine Funktion und wird einmalig automatisch nach anlegen einer Klasseninstanz aufgerufen und
* wird zur Initialisierung der Elemente einer Klasse verwendet.

Automatisch erzeugt C++ einen Standardkonstruktor, der nacheinander die 
Standardkonstruktoren aller Membervariablen aufruft, sofern diese existieren (die Standard Ganzzahl- und Gleitpunkttypen werden nicht initialisiert und besitzen keinen Konstruktor).

Konstruktoren können aber auch selbst definiert werden:

```C++
class classname
{
    // member variables
public:
    classname(parameter_list) : initializer_list
    {
        // constructor_body
    }
};
```
* Der Funktionsname des Konstruktors ist gleich dem Klassennamen
* Der Konstruktor
    * besitzt keinen Rückgabetyp oder -wert,
    * kann, wie in C++ üblich, mit beliebigen Parametern definiert sein,
    * Defaultargumente enthalten und
    * auch überladen werden.
* Zumeist ist es sinnvoll, Konstruktoren `public:` zu definieren.

Nach dem `:` könenn in der `initializer_list` Membervariablen getrennt durch Kommas auf vereinfachte Weise initialisiert werden:\
`memberA(initial_valueA), memeberB(initital_valueB)` und so weiter\
Nicht verwendete Member werden hierbei mit ihren Defaultkonstruktoren (sofern vorhanden) initialisiert. Die Member werden dabei in der Reihenfolge ihres Auftretens in der Klassendefinition initialisiert.\
Die Initialisierungsliste (samt `:`) kann gegebenenfalls entfallen.
Der eigentlichen Funktionsblock kann ebenfalls leer bleiben oder zusätzlichen Code enthalten. 
Die geschwungenen Klammern sind jedenfalls erforderlich. 

Konstruktoren können, wie alle Funktionen, sowohl innerhalb als auch außerhalb der Klassen definiert werden, wobei eine Funktionsdeklaration innerhalb der Klassendefinition erforderlich ist.

Es gibt drei Arten von Konstruktoren: 
* Standardkonstruktor (*engl. default constructor*) bekommt keine Parameter übergeben und initialisiert die Klasse mit Defaultwerten.
* Parameterkonstruktor (*engl. parameterized constructor*) kann beliebige Parameter enthalten
* Kopierkonstruktor (*engl. copy constructor*) wird zur Initialisierung mit derselben Klasse verwendet

Nicht alle Konstruktorarten müssen implementiert werden. Wird ein Konstruktor implementiert, so wird kein automatischer Standardkonstruktor generiert. 
Sind Membervariablen `const` definiert, so ist der Konstruktor die einzige Möglichkeit, diese Variablen zu initialisieren.

Der Konstruktor wird direkt beim Anlegen einer Variablen aufgerufen, beispielsweise:

```c++
classname variablename1;                     // default constuctor
classname variablename1();                   // default constuctor
classname variablename2(parameters);         // parameters for the constructor
classname variablename3 = variablename;      // this is also an initialisation and calls the (copy) constructor
```


````c++
class Vec2d
{
    double x,y;                                        // we have private members, so we need a constructor                   
public:
    Vec2d() : x(0), y(0) {}                            // default constructor initializes the members x and y with 0.
    Vec2d(double x_, double y_ = 0) : x(x_), y(y_) {}  // parameterized constructor initializes x and y with user values,
                                                       // the constructor Vec2d(x_) is also implemented with y_ = 0
};

int main()
{
    Vec2d v0;       // calls default constructor
    Vec2d v1(1,1);  // call overloaded constructor with x_ and y_
    Vec2d v2(2);    // call overloaded constructor with x_ and y_ = 0
}
````

## Direkte Initialisierung der Klassenmember

In modernen C++ Kompilern ist es möglich, ähnlich einer Variablendeklaration, Member direkt bei der Memberdeklaration mit Defaultwerten zu initialisieren (*engl. in-class memember initialization*). Eine etwaige Initialisierung in einem Konstruktor hat jedenfalls vorrang.

````c++
class A 
{
    int i = 12;
    int j{14};
    std::string s{"hallo"};
    std::vector<int> v{1,2,3};
};
````

## Destruktor (*engl. destructor*)

Verliert eine Variable ihren Gültigkeitsbereich, so wird zuvor der Destruktor aufgerufen um zusammenzuräumen. 
Auch dessen Funktionalität kann eigens definiert werden. 
Der Destruktor ist eine Parameterlose Funktion, deren Name dem Klassennamen entspricht, dem ein `~` vorangestellt ist.
Standardmäßig wird ein leerer Standarddestruktor generiert.
Nach Verlassen des Funktionsrumpfs werden nacheinander die Destruktoren aller Member in umgekehrter Reihenfolge aufgerufen.

```C++
class classname
{
    // ...
    ~classname() 
    {
        // destructor code comes here
    }
};
```

## Copy Konstruktor (*engl. copy constructor*) und Zuweisungsoperator (*engl. copy assignment*)

Bei Initialisierung einer Klasse durch ein Objekt derselben Klasse wird der Copy Konstruktor aufgerufen. 
Hingegen wird bei der Zuweisung eines Objekts zu einem anderen Objekt der Zuweisungsoperator `=` aufgerufen. Standardmäßig werden auch diese Funktionalitäten generiert oder können neu definiert werden.

```C++
class classname
{
    // ...
    classname(const classname &source) : initializer_list {
        // this is the copy constructor. It initializes the new instance *this with source
        // ...
    }
    
    classname &operator=(const classname &source) {
        // this is the assignment operator. It copies the members of source to an existing *this
        // ...
    }
};
```

Zu beachten ist:
* Beim Copy Konstruktor muss der Parameter als Referenz übergeben werden. Ansonsten müsste eine Kopie von `source` generiert werden, was ja genau der Copy Konstruktor bewerkstelligen soll.
* Ebenso ist es meistens sinnvoll, auch bei der Zuweisung die Ursprungsklasse als Referenz zu übergeben und eine Referenz auf das Objekt selbst zurückzugeben. Hierdurch wird beispielsweise `c = b = a` ermöglicht.
* Des Weiteren können meist auch `const` Objekte als Quelle verwendet werden.
* Der Copy Konstruktor initialisiert eine neue Instanz mit Werten einer bestehenden Variablen, 
* während der Zuweisungsoperator einer bereits bestehenden Instanz auf der linken Seite des `=`-Zeichens die Werte der Variablen auf der rechten Seite zuweist.

Im folgenden wird ein Beispiel einer einfachen Klasse mit Konstruktoren, Destruktor und Zuweisungsoperator gezeigt

````c++
class Vec2d
{
    double x, y;
public:
    Vec2d(double x_ = 0, double y_ = 0) : x(x_), y(y_) {} // initializes x and y, default values are 0 for x and y
                                                          // this also implements the default constructor!
                                                          
    Vec2d(const Vec2d &source) : x(source.x), y(source.y) // copy constructor
    {}
    
    Vec2d &operator=(const Vec2d &source) {               // assignment operator
        x = source.x;
        y = source.y;
        return *this;
    }
    
    ~Vec2d() {                                            // destructor
        //...
    } 
};

int main()
{
    Vec2d v0;              // default constructor is called
    Vec2d v1(1,1);         // constructor
    
    Vec2d v2(v0);          // copy constructor, initializes v2 with v0
    Vec2d v3 = v1;         // !!!here we also have the copy constructor because it is the initialisation
    
    v3 = v2;               // assignment operator

    // at the end of the function all local variables loose their scope, but before the destructors of all of them are called
}
````
