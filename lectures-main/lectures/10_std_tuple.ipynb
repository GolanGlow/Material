 # <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Einleitung zu `std::tuple`</strong>

- Um große Datenmengen zu speichern werden sehr oft Listen verwendet
  - In C++ gibt es dazu seit C++98 den `std::vector`
- Für Listen und Vektoren müssen alle Elemente einer Liste oder eines Vektors jeweils denselben Typ aufweisen
- `std::tuple` ermöglicht seit C++11 das Speichern von mehreren Werten unterschiedlichen Typs in einer einzigen Datenstruktur
  - Dabei kann `std::tuple` als Container betrachtet werden, der durch eine bestimmte Reihenfolge von Indizes identifiziert wird
- Bei `std::tuple` handelt es sich um eine C++-Standardbibliotheksklasse ab C++11
- Die Klasse ist in der Header-Datei `<tuple>` definiert
- Hier ist ein Beispiel für die Verwendung von `std::tuple` in C++

````c++
#include <iostream>
#include <tuple>
#include <string>
  
int main() {
    std::tuple<int, double, std::string> myTuple(10, 3.14, "Hello");

    return 0;
}
````

- Im Vergleich dazu haben wir in Python das Konzept der Tupel bereits kennengelernt

```python
myTuple = (10, 3.14, "Hello")
print(myTuple)

# print single element
print(myTuple[1])
```

**Wie kann man ein `std::tuple` in C++ initialisieren?**

Direkte Initialisierung: Hierbei werden die Werte des `std::tuple` direkt bei der Erstellung zugewiesen
- Die Syntax hierfür sieht folgendermaßen aus

```cpp
std::tuple<int, double, std::string, ...> myTuple(42, 3.14, "Hello", ...);  
```

Verwendung von `std::make_tuple`: 
- Die `std::make_tuple`-Funktion kann verwendet werden, um ein `std::tuple` zu erstellen
- Die `std::make_tuple`-Funktion erstellt ein `std::tuple`-Objekt, indem sie die Argumente kopiert
- Das bedeutet, dass die Argumente in eine temporäre Variable kopiert werden und diese Kopien dann an das `std::tuple`-Objekt übergeben werden
- Bei Verwendung mit `auto` müssen die Typen der einzelnen Komponenten nicht explizit angegeben werden
- Die Syntax hierfür sieht folgendermaßen aus

```cpp
auto myTuple = std::make_tuple(42, 3.14, "Hello", ...);
```

Verwendung von `std::forward_as_tuple`: Die `std::forward_as_tuple`-Funktion erstellt ein Tupel aus Referenzen (wichtig für Perfect Forwarding).
- Wenn das `std::tuple`-Objekt an eine Funktion weitergegeben wird, kann dadurch die Anzahl der Kopien von Objekten reduziert werden
- Die Syntax hierfür sieht folgendermaßen aus

```cpp
auto myTuple = std::forward_as_tuple(42, 3.14,"Hello", ...);
```

Es ist auch möglich, ein `std::tuple` mit leeren Werten zu initialisieren, indem man die Standardkonstruktoren der jeweiligen Datentypen verwendet
- Zum Beispiel sieht die Initialisierung eines `std::tuple` mit 3 leeren Werten folgendermaßen aus
```cpp
std::tuple<int, double, std::string> myTuple(0, 0.0, "");
```

**Hinweis**
- Im Gegensatz zu Python-Tupeln ist ein `std::tuple` in C++ veränderbar, d.h. die gespeicherten Werte können geändert werden, sofern die Elementtypen dies erlauben
- Die Anzahl der Elemente in einem `std::tuple` ist jedoch fest: es können keine neuen Elemente nach der Initialisierung hinzugefügt oder entfernt werden
- Falls eine Erweiterung erforderlich ist, muss stattdessen ein neues `std::tuple` mit `std::tuple_cat` erstellt werden

 # <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Möglichkeiten um auf Elemente von `std::tuple` zuzugreifen</strong>

Hier sind einige Möglichkeiten, um auf die Elemente von `std::tuple` in C++ zuzugreifen

**Verwendung von `std::get`**
- Die `std::get`-Funktion kann verwendet werden, um auf die Elemente von `std::tuple` zuzugreifen
- Die Syntax sieht folgendermaßen aus

````c++
#include <iostream>
#include <tuple>

int main() {
    // std::tuple<int, double, std::string> myTuple = std::make_tuple(42, 3.14, "Hello");
    auto myTuple = std::make_tuple(42, 3.14, "Hello"); // dynamic data type

    std::cout << std::get<0>(myTuple) << std::endl;
    std::cout << std::get<1>(myTuple) << std::endl;
    std::cout << std::get<2>(myTuple) << std::endl;

    // what happens here?
    std::get<0>(myTuple) = 17;
    std::cout << std::get<0>(myTuple) << std::endl;

    return 0;
}
````

- Die Funktion `std::get` gibt eine Referenz auf das angeforderte Element zurück
  - In den Spitzen Klammern `<...>` steht der Index des Elements, auf das zugegriffen werden soll

**Verwendung von `std::tie`**
- Die `std::tie`-Funktion kann verwendet werden, um die Elemente von `std::tuple` an Variablen zu binden
  - Die Syntax sieht folgendermaßen aus

````c++
#include <iostream>
#include <tuple>
#include <string>

int main() {
    auto myTuple = std::make_tuple(42, "Hello", 3.14);

    int x;
    std::string y;
    double z;

    std::tie(x, y, z) = myTuple;

    std::cout << x << std::endl;
    std::cout << y << std::endl;
    std::cout << z << std::endl;

    // what happens here?
    x = 17;
    std::cout << std::get<0>(myTuple) << std::endl;
    
    // if you only want to know one specific value of a tuple
    std::tie(std::ignore, std::ignore, z) = myTuple;
    return 0;
}
````

- In diesem Beispiel werden die Werte des Tupels `myTuple` mithilfe von `std::tie` den Variablen `x`, `y` und `z` zugewiesen.
- Dabei erstellt `std::tie` temporäre Referenzen auf die Variablen, sodass die Tupelwerte diesen zugewiesen werden.
- Nach der Zuweisung sind die Variablen unabhängig vom Tupel.

**Verwendung von Structured Binding**
- Structured Binding (ab C++17) ermöglicht es, die Elemente von `std::tuple` direkt an Variablen zu binden
- Die Syntax sieht folgendermaßen aus

````c++
#include <iostream>
#include <tuple>
#include <string>

int main() {
    auto myTuple = std::make_tuple(42, "Hello", 3.14);

    auto [x, y, z] = myTuple;

    std::cout << x << std::endl;
    std::cout << y << std::endl;
    std::cout << z << std::endl;

    return 0;
}
````

# <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Arbeiten mit `std::tuple`</strong>
- Ein `std::tuple` ist eine Klasse aus der C++-Standardbibliothek und erfordert daher eine `#include`-Direktive, um verwendet werden zu können
- Die Elemente einer `std::tuple` werden über einen Index zugegriffen, der bei 0 beginnt

**Wenn ein `std::tuple` als Rückgabetyp von einer Funktion verwendet wird**
- Kann die `auto`-Anweisung  verwendet werden, um die einzelnen Elemente der `std::tuple` auf Variablen zuzuweisen
- Kann die `std::tie`-Funktion verwendet werden, um die einzelnen Elemente der `std::tuple` auf Variablen zuzuweisen

````c++
#include <iostream>  
#include <tuple>  
#include <string>  
  
std::tuple<int, double, std::string> myFunction() {  
  int x = 42;  
  double y = 3.14;  
  std::string z = "Hello";  
  return std::make_tuple(x, y, z);  
}  
  
int main() {
  // usage of std::tie
  int x;  
  double y;  
  std::string z;  
  std::tie(x, y, z) = myFunction();  
  
  std::cout << x << std::endl; // Ausgabe: 42  
  std::cout << y << std::endl; // Ausgabe: 3.14  
  std::cout << z << std::endl; // Ausgabe: Hello 

  // auto directive
  auto myTuple = myFunction();

  std::cout << "Size of myTuple: " << std::tuple_size_v<decltype(myTuple)> << std::endl;

  std::cout << std::get<0>(myTuple) << std::endl; // Ausgabe: 42  
  std::cout << std::get<1>(myTuple) << std::endl; // Ausgabe: 3.14  
  std::cout << std::get<2>(myTuple) << std::endl; // Ausgabe: Hello
  
  return 0;  
}
````

**Wenn ein `std::tuple` als Parameter an eine Funktion übergeben wird**
- Kann die `std::make_tuple`-Funktion verwendet werden, um ein `std::tuple` zu erstellen
- Kann die `std::tie`-Funktion verwendet werden, um die einzelnen Elemente der `std::tuple` auf Variablen zuzuweisen
- Dabei sollte die Reihenfolge der Elemente in der `std::tuple` klar dokumentiert werden, um Verwirrung zu vermeiden

````c++
#include <iostream>  
#include <tuple>  
  
void myFunction(std::tuple<int, double, std::string> myTuple) {  
  int x;  
  double y;  
  std::string z;  

  std::tie(x, y, z) = myTuple;  

  std::cout << x << std::endl; // Ausgabe: 42  
  std::cout << y << std::endl; // Ausgabe: 3.14  
  std::cout << z << std::endl; // Ausgabe: Hello  
}  
  
int main() {  
  int x = 42;  
  double y = 3.14;  
  std::string z = "Hello";  

  std::tuple<int, double, std::string> myTuple = std::make_tuple(x, y, z);  

  myFunction(myTuple);  
  
  return 0;  
}
````

**Verwenden der `using`-Anweisung**

- Mit der `using`-Anweisung kann ein eigener `Typ (Name)` für ein Tupel definiert werden
```cpp
using MyTuple = std::tuple<...>;
```
- Durch die Verwendung der `using`-Anweisung und der `[]`-Syntax kann der Code leichter lesbar und verständlicher gestaltet werden

````c++
#include <iostream>  
#include <tuple>  
#include <string>

using MyTuple = std::tuple<int, double, std::string>;  
  
MyTuple myFunction() {  
  int x = 42;  
  double y = 3.14;  
  std::string z = "Hello";  
  return std::make_tuple(x, y, z);  
}  
  
int main() {  
  auto myTuple = myFunction();  
  auto [x, y, z] = myTuple;  
  std::cout << x << std::endl; // Ausgabe: 42  
  std::cout << y << std::endl; // Ausgabe: 3.14  
  std::cout << z << std::endl; // Ausgabe: Hello  
  return 0;  
}
````

- Die Verwendung von `std::tuple` in C++ kann die Lesbarkeit des Codes verbessern
  - Insbesondere da mehrere Werte unterschiedlicher Typen zusammengefasst werden können

# <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Operationen mit `std::tuple`</strong>

- Tupel können mit den Vergleichsoperatoren `==`, `!=`, `<`, `>`, `<=`, `>=` verglichen werden
  - Der Vergleich erfolgt lexikographisch, d.h. Element für Element
- Mit `std::tuple_size<decltype(myTuple)>::value` kann die Anzahl der Elemente eines Tupels ermittelt werden
- Mit `std::tuple_element<Index, decltype(myTuple)>::type` kann der Typ eines Elements abgefragt werden

# <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Anwendungsfälle für `std::tuple`</strong>

- Rückgabe mehrerer Werte aus einer Funktion
- Speicherung heterogener Daten (z.B. Datensätze aus einer Datenbank)
- Verwendung als Schlüssel in assoziativen Containern wie `std::map`
- Parameterpakete für variadische Templates
- Implementation von Algorithmen, die mehrere Werte gleichzeitig verarbeiten müssen

````c++
#include <iostream>
#include <tuple>
#include <string>

int main() {
    auto myTuple = std::make_tuple(42, "Hello", 3.14);
    
    // Modify the first element
    std::get<0>(myTuple) = 100;
    
    // Modify the second element
    std::get<1>(myTuple) = "World";

    // Output of the modified values
    std::cout << std::get<0>(myTuple) << std::endl;
    std::cout << std::get<1>(myTuple) << std::endl;
    std::cout << std::get<2>(myTuple) << std::endl;
    
    return 0;
}
````
