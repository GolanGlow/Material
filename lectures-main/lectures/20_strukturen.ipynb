# <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Grundlagen der Strukturen</strong>
- Strukturen ermöglichen es, zusammengehörige Daten unter einem einzigen Namen zu gruppieren und sie dadurch effektiver zu organisieren und zu verwalten, insbesondere wenn es sich um komplexe Datentypen handelt
    - Strukturen gelten als zusammengesetzte Datentypen, da sie mehrere Variablen unterschiedlichen Typs enthalten können
    - Strukturen erleichtern das modulare Programmieren, indem sie komplexe Systeme in kleinere, besser handhabbare Komponenten zerlegen, da jede Struktur ein Modul oder eine Komponente des Systems darstellen kann
    - Strukturen bieten eine Abstraktionsebene, indem sie es ermöglichen, Daten und Operationen zu kapseln, die miteinander in Beziehung stehen
- Das Speicherlayout einer Struktur ist in der Regel zusammenhängend, d. h. ihre Member werden an aufeinanderfolgenden Speicherplätzen gespeichert
- Das Konzept der Strukturen gibt es zwar in vielen Programmiersprachen, aber die Syntax und die Eigenschaften sind unterschiedlich
    - C, C++, Java und Python verfügen alle über Strukturen oder ähnliche Konstrukte, jedoch mit Unterschieden in der Implementierung und Verwendung

## <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Strukturen in Python</strong>

- In Python bezieht sich der Begriff *Struktur* generell auf Datenstrukturen, die durch eingebaute Datentypen wie Listen, Tupel, Dictionaries und Sets dargestellt werden
- Die eingebauten Datenstrukturen von Python bieten eine große Flexibilität und werden häufig für verschiedene Zwecke verwendet
- Ein Beispiel einer `Person`-Struktur, implementiert als Tupel in Python:

<img align="left" width="250" src=images/student_tuple.png>

```c++
# Define a tuple representing a person: (name, age, height, is_student)
Person = ("Alice", 30, 1.65, True)

# Accessing elements of the tuple
name = Person[0]
age = Person[1]
height = Person[2]
is_student = Person[3]

# Print the person's information
print("Name:", name)
print("Age:", age)
print("Height:", height, "m")
print("Is a student?", "Yes" if is_student else "No")
```

## <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Strukturen in C++</strong>
- In C++ ist eine Struktur ein benutzerdefinierter Datentyp, der gemeinsam verwandte Variablen unter einem Namen zusammenfasst
- Eine Struktur wird mit dem Schlüsselwort `struct` erstellt
- Ein Beispiel einer `Person`-Struktur, implementiert als `struct` in C++:

```c++
#include <iostream> // std::cout, std::endl
#include <string>   // std::string

// Define a struct representing a person: (name, age, height, is_student)
struct Person {
    std::string name;
    int age;
    double height;
    bool is_student;
};

int main() {
    // Declare and initialize a person
    Person myPerson = {"Alice", 30, 1.65, true};

    // Collect myPerson's information to local variables
    auto name = myPerson.name;
    auto age = myPerson.age;
    auto height = myPerson.height;
    auto is_student = myPerson.is_student;

    // Print the person's information
    std::cout << "Name: " << name << std::endl;
    std::cout << "Age: " << age << std::endl;
    std::cout << "Height: " << height << "m" << std::endl;
    std::cout << "Is a student? ";
    if (is_student) {
        std::cout << "Yes";
    } else {
        std::cout << "No";
    }

    return 0;
}
```

# <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Initialisierung einer Strukturinstanz</strong>

- Standardmäßig werden grundlegende Datentypen wie `int`, `char` oder `double` nicht initialisiert
- Wenn einige dieser Elemente nicht explizit initialisiert werden, sind sie uninitialisiert
    - Nicht initialisierten Variablen können später im Code Werte zugewiesen werden
- In C++ gibt es mehrere Möglichkeiten, die Elemente einer Struktur zu initialisieren:
- **Elementweise Initialisierung**:
    - Jedes Element der Struktur wird bei der Deklaration oder später im Code einzeln initialisiert

```c++
Person myPerson;
myPerson.name = "Alice";
myPerson.age = 30;
myPerson.height = 1.65;
myPerson.is_student = true;
```

- **Aggregat-Initialisierung**:
    - Seit C++11 können alle `struct`-Elemente mit geschwungenen Klammern direkt initialisiert werden
    - Die Elemente müssen in der gleichen Reihenfolge initialisiert werden, in der sie in der Strukturdefinition erscheinen

```c++
// Initialization during declaration
Person myPerson = {"Alice", 30, 1.65, true};

// Values assigned after declaration
Person myPerson;
myPerson = {"Alice", 30, 1.65, true};
```

- **Designierte Initialisierer**:
    - Seit C++20 können designierte Initialisierer (*engl. designated initialization*) verwendet werden, um anzugeben, welche Elemente der Struktur initialisiert werden sollen
    - Es handelt sich im Wesentlichen um eine Erweiterung der Aggregatinitialisierung
    - Nicht explizit initialisierte Variablen werden auf ihren Standardwert initialisiert (z. B. `0` oder `false`)
    - Die Elemente müssen in der Reihenfolge initialisiert werden, in der sie in der Struktur definiert sind

```c++
// initialize only name and age
Person myPerson = {.name = "Alice", .age = 30};
// height is initialized to 0.0 and is_student to false

// ERROR: initialization must be in declaration order in C++
Person myPerson2 = {.age = 30, .height = 1.65, .is_student = true, .name = "Alice"};
```

- **Konstruktor-Initialisierung**:
    - Ein Konstruktor kann definiert werden, um die Initialisierung von Elementen während der Objekterstellung zu ermöglichen
    - Mehr dazu in einer späteren Vorlesung

# <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Zugriff und Änderung von Elementen einer Struktur</strong>
In C++ kann man auf die Elemente einer Struktur mit dem Punktoperator (`.`) zugreifen, sie auslesen bzw. gegebenfalls auch entsprechend ändern *(Der Pfeil-Operator `->` kann auch im Falle von Zeigern verwendet werden - mehr dazu später)*

```c++
#include <iostream> // std::cout, std::endl
#include <string>   // std::string

// Define a struct representing a person: (name, age, height, is_student)
struct Person {
    std::string name;
    int age;
    double height;
    bool is_student;
};

int main() {
    // Declare and initialize a person
    Person myPerson = {"Alice", 30, 1.65, true};

    // Modify myPerson's age:
    myPerson.age = 31;

    // Print the person's information
    std::cout << "Name: " << myPerson.name << std::endl;
    std::cout << "Age: " << myPerson.age << std::endl;
    std::cout << "Height: " << myPerson.height << "m" << std::endl;
    std::cout << "Is a student? ";
    if (myPerson.is_student) {
        std::cout << "Yes" << std::endl;
    } else {
        std::cout << "No" << std::endl;
    }

    return 0;
}
```
