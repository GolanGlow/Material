 # <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Einleitung zu Namensräume in C++</strong>

- Ein Namensraum (engl. *namespace*) in C++ definiert einen Bereich, innerhalb dessen Programmelemente wie Variablen, Funktionen und Klassen zusammengefasst werden können

**Zweck von Namensräumen**
- Vermeidung von Namenskonflikten durch eindeutige Zuordnung von Elementen zu einem Namensraum
- Bessere Strukturierung und Lesbarkeit des Programmcodes durch logische Gruppierung zusammengehöriger Elemente (z.B. Klassen und Funktionen, die ähnliche Aufgaben erfüllen)

- Ein bekanntes Beispiel ist der Namensraum `std`, in dem die Standardbibliothek von C++ implementiert ist:

````c++
#include <iostream>

int main() {

  std::cout << "Hello World!" << std::endl;

  return 0;
}
````

- Eigene Namensräume können mit dem Schlüsselwort `namespace` definiert werden.
- Der Zugriff auf Elemente eines Namensraums erfolgt über die Syntax:
```cpp
namespace::element
```

````c++
#include <iostream>

namespace my_namespace {
  // Declarations and definitions
  int glob_x = 42;
  void my_function() { }
}

namespace other_namespace {
  // Declarations and definitions
  int glob_x = 12;
  void my_function() { /* does other thing */ }
}


int main() {

  std::cout << "Hello World!" << std::endl;

  std::cout << my_namespace::glob_x << std::endl; // access variable from namespace
  std::cout << other_namespace::glob_x << std::endl; // access variable from other namespace

  my_namespace::my_function();    // call function from namespace
  other_namespace::my_function(); // call function from other namespace

  return 0;
}
````

 # <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Namensräume in Python</strong>

- Ein Namensraum in Python ist eine Sammlung von aktuell definierten symbolischen Namen sowie Informationen zu dem Objekt, die jeweils auf ein konkretes Objekt verweisen

- In einem Python-Programm gibt es vier Arten von Namespaces

**Globaler Namensraum**
- Der globale Namesraum (*global namespace*) enthält alle Namen, die auf der obersten Ebene eines Python-Moduls oder Skripts definiert sind
  - Python erstellt den globalen Namensraum automatisch beim Start des Programms und er bleibt aktiv, bis der Python-Interpreter beendet wird.
    - Dies ist streng genommen möglicherweise nicht der einzige globale Namensraum, der existiert
    - Der Interpreter erstellt auch einen globalen Namensraum für jedes Modul, welches mit dem `import` Statement geladen wird
  - Der globale Namensraum kann von überall im Programm aus aufgerufen werden, aber es wird empfohlen, seine Verwendung auf ein Minimum zu beschränken, um Namenskonflikte und unerwartete Seiteneffekte zu vermeiden
  - Um eine globale Variable innerhalb einer Funktion zu verändern, muss diese Variable mit dem Schlüsselwort `global` explizit als global gekennzeichnet werden

```python
x = 10  # globale Variable

def foo():
    global x  # Zugriff auf die globale Variable 'x'
    x += 5

foo()
print(x)  # Ausgabe: 15
```

- Um ein Wörterbuch aller Objekte des globalen Namensraums anzuzeigen, kann die Funktion `globals()` verwendet werden:

```python
print(globals())
```

**Eingeschlossener Namensraum**
- Ein eingeschlossener Namensraum (*enclosing namespace*) entsteht, wenn eine Funktion innerhalb einer anderen Funktion (verschachtelte Funktion) definiert wird und Zugriff auf Namen aus der umgebenden Funktion benötigt
  - Der Interpreter erstellt einen neuen Namensraum, wenn eine Funktion ausgeführt wird
  - Dieser Namensraum ist lokal für die Funktion und bleibt bestehen, bis die Funktion beendet wird
    - Jede Funktion hat einen eigenen Namensraum, der nur während der Ausführung der Funktion existiert
    - Wenn eine Funktion eine andere Funktion aufruft, wird ein neuer Namensraum für die aufgerufene Funktion erstellt
  - Das heißt, Funktionen existieren nicht nur auf der obersten Ebene des Hauptprogramms, sondern können auch innerhalb anderer Funktionen definiert werden
  - Sie können auch eine Funktion in einer anderen Funktion definieren
    - Eine in einer Funktion definierte Funktion wird als verschachtelte Funktion bezeichnet
```python
def f():
  print('Start f()')

  def g():
    print('Start g()')
    print('End g()')
    return

  g()
  print('End f()')
  return

f()

# Ausgabe im Terminal
# Start f()
# Start g()
# End g()
# End f()
```

- Wenn das Hauptprogramm die Funktion `f()` aufruft, erstellt Python einen neuen lokalen Namensraum für `f()`
- Ruft die Funktion `f()` eine innere Funktion `g()` auf, wird wiederum ein eigener lokaler Namensraum für `g()` erzeugt
- Aus Sicht von `g()` ist der Namensraum von `f()` dann der eingeschlossene (umfassende) Namensraum

**Lokaler Namensraum** 
- Der lokale Namensraum (*local namespace*) wird erstellt, wenn eine Funktion oder Methode ausgeführt wird
- Er umfasst alle Namen (z.B. Variablen, Funktionen), die innerhalb dieser Funktion definiert sind
    - Der lokale Namensraum existiert ausschließlich während der Ausführung der Funktion oder Methode
    - Um ein Wörterbuch mit allen lokalen Namen und zugehörigen Objekten abzurufen, wird die Funktion locals() verwendet

```python
def beispiel():
    x = 42
    y = "Hallo"
    print("beispiel:", locals())
    def inner():
        z = 23
        print("inner:", locals())
    inner()

beispiel()
```

 # <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Verwenden von Namensräumen in C++</strong>

- Zur Ausgabe von Text im Terminal haben wir bisher den Ausdruck `std::cout` verwendet
- Man kann dem Compiler aber mitteilen, dass man sich automatisch auf einen Namensraum bezieht
- Das geschieht mit dem Befehl `using namespace [name of namespace];`, z.B. `using namespace std;`
- Nach der `using namespace std;`-Anweisung kann das Objekt `cout` direkt angesprochen werden

````c++
#include <iostream>

using namespace std;

int main() {

  cout << "Hello World!" << endl;

  return 0;
}
````

- Das Verwenden von `using namespace...` ist aber mit Vorsicht zu genießen

````c++
#include <iostream>

using namespace std;

namespace mynamespaceA {
  int glob = 0;
}

namespace mynamespaceB {
  int glob = 1;
  int other_glob = 2;
}

// using namespace mynamespaceA;
// using namespace mynamespaceB;
// If both lines above are uncommented, a naming conflict occurs,
// because the compiler cannot determine which 'glob' you are referring to.

int main() {
  cout << "Hello World!" << endl;

  // Accessing global variables using fully qualified names:
  cout << "mynamespaceA::glob = " << mynamespaceA::glob << endl;
  cout << "mynamespaceB::glob = " << mynamespaceB::glob << endl;

  // Direct access to 'glob' without specifying the namespace is not possible,
  // unless the namespace is explicitly indicated:
  // cout << glob << endl;  // This would result in an error.

  return 0;
}
````

- Hier noch ein komplexeres Beispiel für eine Funktion, welche den Inhalt zweier Variable vertauscht

````c++
#include <iostream>

// using namespace std;
// If the line above is uncommented, the following call to 'swap' 
// becomes ambiguous because the compiler cannot choose between 
// 'std::swap' and our custom-defined 'swap'.

// Custom function template for swapping two numbers
template <class T>
void swap(T& a, T& b) {
  T temp = a;
  a = b;
  b = temp;
}

int main() {
  int a = 1, b = 2;

  std::cout << "Before swap: " << a << ", " << b << std::endl;

  swap(a, b);  // calls our custom swap function

  std::cout << "After swap: " << a << ", " << b << std::endl;

  return 0;
}
````

**Warum wird `using namespace std;` nicht empfohlen?**
- Diese Anweisung importiert alle Elemente aus dem `std`-Namensraum in den globalen Namensraum 
- Wenn später ein eigenes Element (Variable, Funktion, Klasse etc.) denselben Namen wie ein Element aus dem `std`-Namensraum trägt, entstehen Namenskonflikte oder Mehrdeutigkeiten
    - Dies kann zu Compiler-Fehlern oder unerwartetem Verhalten führen

**Empfohlene Vorgehensweise**
- Verwenden Sie immer explizite Namensangaben (z.B. `std::cout`), oder importieren Sie selektiv einzelne Elemente, um die Übersichtlichkeit zu wahren und Konflikte zu vermeiden

````c++
#include <iostream>

// import only 'cout' from namespace std
using std::cout;
using std::endl;

int main() {
    cout << "Hello World!" << endl;  // 'cout' and 'endl' can be used directly
    return 0;
}
````

 ## <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Namespace-Handling bei Funktionsdefinitionen in einer .hpp-Datei</strong>
- Wird eine Funktion in einer `.hpp`-Datei innerhalb eines Namespace deklariert, muss die Definition in der `.cpp`-Datei den Namespace beibehalten.

````c++
// math_functions.hpp file
#ifndef MATH_FUNCTIONS_HPP
#define MATH_FUNCTIONS_HPP

namespace Math {
    int add(int a, int b);  // Function declaration
    int multiply(int a, int b);
}

#endif // MATH_FUNCTIONS_HPP
````

- Zwei Möglichkeiten zur Implementierung in der .cpp-Datei:

1. Direkt im namespace-Block:

````c++
// math_functions.cpp file
#include "math_functions.hpp"

// Define functions inside the same namespace
namespace Math {
    int add(int a, int b) {
        return a + b;
    }

    int multiply(int a, int b) {
        return a * b;
    }
}

int main() {
    int sum = Math::add(5, 3);  // Function call
    int product = Math::multiply(5, 3);
    return 0;
}
````

2. Mit expliziter Namensraum-Qualifizierung:

````c++
// math_functions.cpp file
#include "math_functions.hpp"

// Define functions with explicit namespace qualification
int Math::add(int a, int b) {
    return a + b;
}

int Math::multiply(int a, int b) {
    return a * b;
}
````
