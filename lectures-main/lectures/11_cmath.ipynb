 # <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Kurze Einführung von `cmath`</strong>

- `cmath` ist eine Standardbibliothek in C++, die eine Sammlung von Funktionen für grundlegende mathematische Operationen bereitstellt.
- Während die meisten Funktionen für Gleitkommazahlen (`float`, `double`) ausgelegt sind, sind viele davon auch für `std::complex` (aus dem `<complex>`-Header) überladen.
- Im folgenden Beispiel werden einige Funktionen vorgestellt

```c++
#include <iostream>
#include <cmath>

int main() {
    double x = 9.0;  
    
    // Square root
    std::cout << "sqrt(9.0) = " << std::sqrt(x) << std::endl;

    // Logarithms
    std::cout << "log(9.0) = " << std::log(x) << std::endl;      // Natural log
    std::cout << "log10(9.0) = " << std::log10(x) << std::endl;  // Base-10 log

    // Trigonometric functions (input is in radians)
    std::cout << "sin(9.0) = " << std::sin(x) << std::endl;
    std::cout << "cos(9.0) = " << std::cos(x) << std::endl;

    return 0;
}
```

**Einige Besonderheiten von `cmath` in C++**
- Viele `cmath`-Funktionen (wie `sqrt`, `abs`, `log`) sind für komplexe Zahlen überladen, die im `<complex>`-Header definiert sind.

```c++
#include <iostream>  
#include <cmath>  
#include <complex>  
  
int main() {  
    // initialise complex number
    std::complex<double> z(1.0, 1.0);  
  
    // absolute value of a complex number
    std::cout << std::abs(z);

    return 0;  
}  
```

- Die meisten Funktionen in `cmath` sind für `float`, `double` und `long double` überladen und geben den entsprechenden Typ zurück.
- Die trigonometrischen Funktionen in `cmath` arbeiten normalerweise mit Winkeln in Bogenmaß, nicht in Grad

```c++
// On Windows with MSVC, you might need this define before including cmath
#define _USE_MATH_DEFINES
#include <iostream>  
#include <cmath>  
  
int main() {  
    double angle = 45.0; // in degrees
    // M_PI is a non-standard, but widely available constant.
    // In C++20, it's better to use std::numbers::pi
    double radians = angle * M_PI / 180.0; // Convert to radians
    double sin_val = std::sin(radians);

    std::cout << "sin(45°) = " << sin_val << std::endl;

    return 0;  
}  
```


- `cmath` enthält auch spezielle Funktionen für den Umgang mit NaN (Not-a-Number) und Inf (Unendlichkeit) Werten

```c++
#include <iostream>  
#include <cmath>
#include <limits> // Required for std::numeric_limits
  
int main() {  
    double x = std::nan(""); // Create a NaN value. The string is an optional implementation-defined tag.
    
    // check if x is NaN
    if (std::isnan(x)) {  
        std::cout << "x is NaN" << std::endl;  
    } else {  
        std::cout << "x is not NaN" << std::endl;  
    }  
  
    //x = 1;

    // check if x is finite (NaN is not finite)
    if (std::isfinite(x)) {  
        std::cout << "x is finite" << std::endl;  
    } else {  
        std::cout << "x is infinite" << std::endl;  
    }  
  
    return 0;  
}  
```

- Einige der Funktionen in `cmath` können undefiniert sein, wenn sie auf bestimmte Eingabewerte angewendet werden
  - Zum Beispiel ist der natürliche Logarithmus von 0 undefiniert

```c++
#include <iostream>  
#include <cmath>  
  
int main() {  
    double y = 0.0;
    double res = std::log(y); // log(0) → Results in -inf
    std::cout << "log(0) = " << res << std::endl;
      
    return 0;  
}  
```


- Die Funktionen in `cmath` sind normalerweise performanter als eine eigene Implementierung, da sie oft auf Hardware-Instruktionen abgebildet werden.
- Die Genauigkeit unterliegt den Grenzen der Fließkomma-Arithmetik (z.B. bei sehr kleinen oder sehr großen Zahlen).
- `<cmath>` ist die C++-Version des C-Headers `<math.h>`. Die C++-Version platziert die Namen im `std`-Namespace.
- Obwohl sie Teil der Standardbibliothek sind, erfordern manche Compiler (insbesondere auf Linux) das explizite Linken der Mathe-Bibliothek mit dem Flag `-lm`.
- `cmath` ist portabel und funktioniert auf verschiedenen Betriebssystemen und Architekturen
