## Zugriffskontrolle -- `public`, `private` und `protected` Member

Bis jetzt hatten wir Strukturen mit Membervariablen und Memberfunktionen kennengelernt.
Von außerhalb der Klasse kann mittels des `.`-Operators auf die Variablen und Funktionen zugegriffen werden. 

Es ist nicht immer erwünscht oder sinnvoll, alle Variablen und Funktionen von außen sichtbar zu halten. Das Objekt soll nach außen hin gekapselt werden. Hierzu bietet C++ feiner gegliederte Zugriffsmechanismen. Mit dem Schlüsselwort

* **`public:`**
werden alle folgenden Member der Struktur öffentlich sichtbar. Das bedeutet, von außen sind diese Member ansprechbar.

* **`private:`**
Ab hier sind die Member von außerhalb der Klasse nicht sichtbar. Die Member können nur mehr von den Memberfunktionen selbst angesprochen werden.

* **`protected:`**
Auch hier können die folgenden Member nur mehr innerhalb der Klasse angesprochen werden. *Auf `protected` Member kann auch von geerbten Klassen zugegriffen werden -- mehr dazu später*

* Alle Elemente einer Struktur sind standardmäßig öffentlich (`public`)
* In einer Struktur können mehrere dieser Blöcke folgen. Der jeweilige Zugriffsmechanismus bleibt so lange bestehen, bis ein anderer Zugriffsmechanismus definiert wird oder die Struktur beendet wird.

## Klassen in C++

Mit dem Konstrukt

```c++
class Classname
{
    // Class definitions
};
```

wird analog zu einer Struktur eine Klasse definiert. Einziger Unterschied ist, 
* dass standardmäßig alle Member `private` sind. 
* Erst durch das Schlüsselwort `public` werden Member offengelegt.

Im weiteren wird allgemein von Klassen gesprochen, unabhängig ob es sich tatsächlich um Strukturen oder Klassen handelt.

````c++
class MyClass               // all following members are private in a class!
{
    int var1, var2;    
public:                     // we need public member functions to access the members
    void set_var1(int v) {  // a function for setting var1 is required
        var1 = v;           // inside the class, var1 is accessible
    }
    void set_var2(int v) {  // a function for setting var2 is required
        var2 = v;
    }
    int get_var1() const {  // a function for reading var1 is required
        return this->var1;  // we also can access var1 via this
    }
    int get_var2() const {  // a function for reading var2 is required
        return var2;
    }
    int &access_var1() {    // here we have access to var1 because we return a reference
        return var1;
    }
private:                    // private members follow again
    void intern_function(); // do some private stuff
};

int main() {
    MyClass my;
    
    // my.var1 = 17         // this is not possible, because var1 is private
    
    my.set_var1(17);
    my.set_var2(18);
    
    std::cout << my.get_var1() << " " << my.get_var2() << std::endl;
    
    my.access_var1() = 18;
}
````

* Normalerweise werden Variablen auf `private` gesetzt und 
* über `public` Interfacefunktionen angesprochen
* Funktionen für den Zugriff auf die Member werden als Setter/Getter-Methoden bezeichnet (`set_...`, `get_...`)


### Definition von Freunden

Um Klassen den Zugriff auf geschützte Membervariablen bzw. -funktionen zu gewähren, können innerhalb einer Klassendefinition Freundklassen bzw. Freundfunktionen angegeben werden:

```c++
class MyClass 
{
    friend class classname;
    friend function_declaration;
/...
};
```

Hierbei hat die angegebene Klasse bzw. Funktion ebenfalls Zugriff auf private Member der definierten Klasse.

### Die `using` und `typedef`-Anweisung

Mit der Typedef-Anweisung wird in C/C++ ein neuer Name für einen bestehenden Klassennamen vergeben.

```c++
typedef class_name new_class_name;
```

Der elegantere Weg auf neueren C++ Compilern ist die `using`-Anweisung.

```c++
using new_class_name = class_name;
```

Zum Beispiel kann für einen kompliziereteren Klassennamen ein kürzerer vergeben werden:

````c++
typedef std::vector<unsigned int> uivec;

uivec a = {1,2,3};

using ivec = std::vector<int>;

ivec b = {4,5};
````
