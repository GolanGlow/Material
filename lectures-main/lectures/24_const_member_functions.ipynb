# <strong style="color: rgb(0, 102, 153); opacity: 1.0;">`const`-Memberfunktionen</strong>
- In der vorangegangenen Vorlesung wurde gezeigt, dass `struct`-Objekte zugehörige Funktionen enthalten können
- Diese Funktionen haben vollen Zugriff auf die Elemente der Struktur
- Was ist zu tun, wenn wir nicht wollen, dass bestimmte Memberfunktionen die Elemente verändern können?
    - Verwendung von `const` Memberfunktionen

```c++
#include <iostream> // std::cout, std::endl

// Define a struct called Point
struct Point {
    int x;
    int y;

    // Member function to set the coordinates of the Point
    void setCoordinates(int x_val, int y_val) {
        x = x_val;
        y = y_val;
    }

    // Member function to return the sum of the coordinates
    int sum() {   // do we plan to modify x or y in this function?
        return x + y;
    }

    // Member function to print the coordinates of the Point
    void printCoordinates() {   // do we plan to modify x or y in this function?
        x = 25; // NO!
        std::cout << "x = " << x << ", y = " << y;
    }

};

int main() {
    Point myPoint = {1, 2};
    myPoint.printCoordinates();

    return 0;
}
```

- Im obigen Beispiel wird deutlich, dass die Funktion `printCoordinates()` in der Lage ist, `x` und `y` zu ändern, was unerwünscht ist
- Um dieses Verhalten zu verhindern, sollten wir die Funktion als `const` kennzeichen: `Rückgabetyp FunktionsName(Funktionsargumente) const`

```c++
#include <iostream> // std::cout, std::endl

// Define a struct called Point
struct Point {
    int x;
    int y;

    // Member function to set the coordinates of the Point
    void setCoordinates(int x_val, int y_val) {
        x = x_val;
        y = y_val;
    }

    // Member function to return the sum of the coordinates
    int sum() const {  // adding const after () prevents the function from modifying x or y!
        return x + y;
    }

    // Member function to print the coordinates of the Point
    void printCoordinates() const {   // adding const after () prevents the function from modifying x or y!
        // x = 25;
        std::cout << "x = " << x << ", y = " << y;
    }

};

int main() {
    Point myPoint = {1, 2};
    myPoint.printCoordinates();

    return 0;
}
```

Daher muss sichergestellt werden, dass eine Memberfunktion nicht versehentlich Werte ändert:
- Sie sollte mit dem Schlüsselwort `const` definiert werden.
- Argumente sollten als Referenz (`&`) mit dem Schlüsselwort `const` übergeben werden

## <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Beispiel</strong>
- Beispiel für die Verwendung einer `const`-Memberfunktion und die Übergabe von Argumenten per `const`-Referenz

```c++
#include <iostream>
#include <cmath> // std::sqrt, std::pow

// Define a struct called Point
struct Point {
    double x;
    double y;

    // Member function to set the point
    void setPoint(double x_val, double y_val) {
        x = x_val;
        y = y_val;
    }

    // Member function to calculate the distance between two points
    double calculateDistance(const Point& otherPoint) const {
        // Calculate the distance using Euclidean distance formula
        return std::sqrt(std::pow(otherPoint.x - x, 2) + std::pow(otherPoint.y - y, 2));
    }

    // Member function to print the current point
    void printPoint() const {
        std::cout << "(" << x << ", " << y << ")" << std::endl;
    }
};

int main() {
    Point point1;
    Point point2;

    point1.setPoint(1.0, 2.0);
    point2.setPoint(4.0, 6.0);

    // Call member function calculateDistance() with const reference argument
    double distance = point1.calculateDistance(point2);

    point1.printPoint();
    point2.printPoint();
    std::cout << "Distance between point1 and point2 is " << distance << std::endl;

    return 0;
}
```

# <strong style="color: rgb(0, 102, 153); opacity: 1.0;">`const` als Rückgabetyp</strong>
- Wie bei der Übergabe von Argumenten können auch Rückgabewerte einer Funktion entweder als Wert oder als Referenz (Zeiger) erfolgen
- Gibt eine Funktion eine Referenz zurück, kann der Aufrufer die ursprünglichen Daten verändern
- Um dies zu verhindern, kann der Rückgabewert mit `const` geschützt werden:
  - Dadurch wird sichergestellt, dass die zurückgegebene Referenz nicht verändert werden kann
- Dies erhöht die Sicherheit und macht die Funktion explizit nur lesend

**Hinweis: Verwenden Sie `const` bei Rückgabereferenzen, wenn Sie nur Lesezugriff erlauben, aber keine Veränderung der Daten zulassen möchten.**

```c++
#include <iostream>

struct MyClass {
    int data = 10;

    // Getter returns a reference and allows modification
    int& get_data() {
        return data;
    }

    // Alternative: getter that returns a const reference and is read-only
    const int& get_data_const() const {
        return data;
    }
};

int main() {
    MyClass obj;

    std::cout << "Initial value: " << obj.get_data() << std::endl;

    // Modify internal data via non-const reference
    obj.get_data() = 20;
    std::cout << "After modification: " << obj.get_data() << std::endl;

    // Use const reference
    const MyClass const_obj = obj;
    std::cout << "Access via const getter: " << const_obj.get_data_const() << std::endl;

    //const_obj.get_data_const() = 30; // Error: cannot assign to const reference

    return 0;
}
```
