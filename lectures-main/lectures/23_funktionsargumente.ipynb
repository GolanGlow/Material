# <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Übergabe von Argumenten in Python</strong>
Wenn in Python ein Objekt an eine Funktion übergeben wird, wird es standardmäßig als Referenz übergeben:
- Dies bedeutet, dass die Funktion das Originalobjekt (seine Speicheradresse) und nicht eine Kopie enthält
    - Allerdings darf die Funktion den Wert des ursprünglichen Objekts nicht verändern, wenn es sich um einen unveränderlichen (*engl. immutable*) Typ handelt (Zahlen, String, Tupel)
    - Veränderliche (*engl. mutable*) Objekte wie Listen, Dictionaries und Sets können geändert werden, nachdem sie an eine Funktion (als Referenz) übergeben wurden

```python
# Function to double a number passed by reference (immutable object)
def double_number(num):
    num = num * 2
    return num

# Function to append an element to a list passed by reference (mutable object)
def append_element(lst, element):
    lst.append(element)  # This will modify the original list

# Example with immutable object (int)
x = 5
print("=== Immutable Example (int) ===")
print("Original value of x:", x)
print("ID of x before:", id(x))
result = double_number(x)
print("Returned (doubled) value:", result)
print("Original x after function call:", x)  # Still 5
print("ID of x after:", id(x))  # Same as before

# Example with mutable object (list)
my_list = [1, 2, 3]
print("\n=== Mutable Example (list) ===")
print("Original list:", my_list)
append_element(my_list, 4)
print("List after function call:", my_list)  # Modified in place
```

# <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Übergabe von Argumenten in C++</strong>
## <strong style="color: rgb(0, 102, 153); opacity: 1.0;">*Pass by Value*</strong>
In C++ werden die Argumente einer Funktion standardmäßig als Wert (*engl. pass by value*) übergeben
- Es wird eine Kopie des Objekts im Speicher angelegt, die der Funktion übergeben wird
    - Die Funktion kann nur auf diese Kopie zugreifen
    - Veränderungen der Parameter wirken sich nicht auf die Originalobjekte aus<br>
<img src="images/pass_by_value.png" style="width: 100%; max-width: 700px; height: auto;" />

## <strong style="color: rgb(0, 102, 153); opacity: 1.0;">*Pass by Reference*</strong>
- Um das Kopieren großer Datenmengen zu vermeiden, ist es oft erwünscht, Objekte als Referenz (*engl. pass by reference*) zu übergeben
- Besonders bei großen Objekten (z.B. `std::vector`, `std::string`) vermeidet man durch pass by reference teure Kopien und verbessert die Performance
- Die Übergabe eines Objekts als Referenz erfolgt mit dem Operator `&`<br>
<img src="images/pass_by_reference.png" style="width: 100%; max-width: 700px; height: auto;" />

**Achtung: Änderungen an Referenzparametern wirken sich auf das Originalobjekt aus, was unerwartete Seiteneffekte verursachen kann!**

````c++
#include <iostream>

// Function to double a number passed by reference
// Modifies the original variable
void doubleNumber(int &num) {
    num = num * 2;
    std::cout << "[doubleNumber] Doubled number: " << num << std::endl;
}

// Function to halve a number passed by value
// Does NOT modify the original variable
void halfNumber(double num) {
    num = num / 2.0;
    std::cout << "[halfNumber] Halved number: " << num << std::endl;
}

int main() {
    int x = 5;
    double y = 5.0;

    std::cout << "=== Pass-by-Reference Example (int) ===" << std::endl;
    std::cout << "Original x: " << x << std::endl;
    doubleNumber(x);
    std::cout << "x after doubleNumber(): " << x << std::endl;

    std::cout << "\n=== Pass-by-Value Example (double) ===" << std::endl;
    std::cout << "Original y: " << y << std::endl;
    halfNumber(y);
    std::cout << "y after halfNumber(): " << y << std::endl;

    return 0;
}
````

Was aber, wenn wir das Kopieren großer Datenmengen vermeiden wollen, aber auch nicht zulassen wollen, dass die Funktion den Wert ändert?
- In diesem Fall können wir ein Objekt als Referenz übergeben (mit dem Operator `&`) und es mit dem Schlüsselwort `const` als unveränderlich kennzeichnen
- Hierdurch kann die Variable auch in der Funktion nicht verändert werden, es kann nur lesend zugegriffen werden

````c++
#include <iostream>

// Function to double a number passed by const reference
// Efficient: avoids copy
// Safe: original cannot be modified
void doubleNumber(const int &num) {
    // num = num * 2;   // Error: cannot modify because of const
    std::cout << "[doubleNumber] Doubled value: " << num * 2 << std::endl;
}

int main() {
    int x = 5;

    std::cout << "=== Pass-by-Const-Reference Example ===" << std::endl;
    std::cout << "Original x: " << x << std::endl;
    
    // Call the function with x passed by reference to const
    doubleNumber(x);
    
    std::cout << "x after doubleNumber(): " << x << std::endl;

    return 0;
}
````

Die Verwendung von `const &` in Funktionsargumenten ist besonders nützlich, um die Leistung zu verbessern (es wird keine Kopie erstellt) und um sicherzustellen, dass Funktionen die Objekte, die sie erhalten, nicht unbeabsichtigt verändern

## <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Beispiel mit `std::vector`</strong>
Bei größeren Objekten wie `std::vector` oder `std::string` wird der Vorteil von `const &` besonders deutlich, da eine Kopie des gesamten Vektors teuer wäre:

````c++
#include <iostream>
#include <vector>   // std::vector

// Efficient: no copy of the vector is made
// Safe: the function cannot modify the vector
double average(const std::vector<double>& values) {
    double sum = 0.0;
    for (const auto& v : values) {
        sum += v;
    }
    return sum / values.size();
}

int main() {
    std::vector<double> grades = {1.0, 2.0, 1.5, 3.0, 2.5};
    std::cout << "Average grade: " << average(grades) << std::endl;
    return 0;
}
````
