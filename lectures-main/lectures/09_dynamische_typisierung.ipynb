# <strong style="color: rgb(0, 102, 153); opacity: 1.0;">`auto` für Typendeklaration</strong>

**In Python wurde der Datentyp einer Variablen bei der Zuweisung eines Wertes bestimmt**
- Man spricht hier von dynamischer Typisierung
```python
myval = 1       # integer
myval = 1.      # double
myval = "Hello" # String
```

**In C++ kann man seit C++11 mit `auto` die Typdeklaration dem Compiler überlassen**

Bisher haben wir Variablen wie folgt deklariert
```cpp
typ variable = initializer;
```
  - Die Variable hat für ihren ganzen Geltungsbereich den angegebenen Typ 
  - Der Initializer kann weggelassen werden, die Variable wird nicht initialisiert.

**Wichtig:** 

Obwohl `auto` der dynamischen Typisierung in Python ähnelt, bleibt C++ statisch typisiert:
- Im Gegensatz zu Python wird der Typ zur Kompilierzeit festgelegt, nicht zur Laufzeit
- `auto` weist den Compiler lediglich an, den Typ automatisch abzuleiten
- Sobald abgeleitet, kann sich der Typ während der Ausführung nicht mehr ändern

````c++
#include <iostream>

int main() {
    //auto myVariable;       // error: declaration of ‘auto myVariable’ has no initializer
    //auto v1 = 3, v2 = 2.5; // error: inconsistent deduction for ‘auto’: ‘int’ and then ‘double’

    auto age = 40;                    // age is int
    auto weight = 82.3;               // weight is double
    auto name = "Firstname Lastname"; // name is const char*

    std::cout << name << " is " << age << " years old and weighs " << weight << "kg" << std::endl;

    return 0;
}
````

### <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Wann sollte man `auto` verwenden?</strong>

- Verwenden Sie `auto` für komplexe Typen, die umständlich zu schreiben sind:
```cpp
std::vector<std::string>::iterator it = v.begin(); // Ohne auto
auto it = v.begin();                               // Mit auto
```
- Verwenden Sie `auto`, wenn der genaue Typ nicht wichtig ist, sondern dessen Verhalten
- Verwenden Sie `auto` für bessere Wartbarkeit, wenn sich Implementierungsdetails ändern
- Ziehen Sie explizite Typen in Betracht, wenn die Typklarheit die Lesbarkeit des Codes verbessert

# <strong style="color: rgb(0, 102, 153); opacity: 1.0;">`auto` mit Literalen</strong>

- Literale bezeichnen den Datentyp, mit welchem ein fixer Wert dargestellt werden soll, der normalerweise zur Initialisierung verwendet wird
- Literale können dem C++-Compiler bei der Bestimmung der Datentypen helfen
```cpp
auto d = 7.9d;         // double
auto f = 6.7f;         // float
auto i = 87;           // int
auto l = 12l;          // long
auto u = 66ul;         // unsigned long
auto o = 012;          // int (octal value)
auto c = "char array"; // const char*
auto s = "string"s;    // std::string (requires "using namespace std")
auto b = true;         // bool
```

# <strong style="color: rgb(0, 102, 153); opacity: 1.0;">`auto` als Rückgabewert einer Funktion</strong>

- `auto` kann auch als Rückgabetyp einer Funktion verwendet werden
- Der Compiler erkennt den Rückgabetyp der Funktion, indem er den Ausdruck der Rückgabe-Anweisung analysiert

````c++
#include<iostream>

auto add(int x, int y) {
    auto sum = x + y;
    return sum;
    // What is the return type?
}

int main() {
    auto x = 32, y = 21; // What types are x and y?
    auto sum = add(x, y); // What type is sum?
    std::cout << x << "+" << y << "=" << sum << std::endl;
    return 0;
}
````

- Mit Hilfe der Funktion `typeid(var).name()` kann der Datentyp der Variable `var` ausgegeben werden
  - Im Vergleich dazu haben wir in Python die Funktion `type` verwendet

````c++
#include<iostream>
#include <typeinfo>

auto divide(double x, double y) {
    auto quotient = x / y; // quotient is type double
    return quotient;
    // return type is double
}

int main() {
    auto quotient = divide (12.0, 32.0); // quotient is type double
    std::cout << "quotient is " << quotient << " of type " << typeid(quotient).name() << std::endl;
    return 0;
}
````

# <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Häufige Fallstricke mit `auto`</strong>

- `auto` bewahrt Referenzen oder Top-Level-`const`-Qualifizierer nicht:
```cpp
int x = 10;
const int& rx = x;
auto a = rx;      // a ist int, nicht const int&
auto& b = rx;     // b ist const int&
```

- **Vorsicht bei Proxy-Objekten (z.B. `std::vector<bool>`):**
```cpp
std::vector<int> numbers = {1, 2};
auto n = numbers[0]; // n ist int (Kopie), Referenz wird entfernt
n = 5;               // Ändert NICHT den Vektor

std::vector<bool> flags = {true, false};
auto f = flags[0]; // f ist nicht bool, sondern std::vector<bool>::reference!
f = false;         // Ändert auch den Wert im Vektor!
```
- **Kein `+` Operator für `std::vector` (anders als in Python):**
```cpp
std::vector<int> v1 = {1, 2}, v2 = {3, 4};
// auto v3 = v1 + v2; // Compiler-Fehler! std::vector unterstützt kein +
```

- Seien Sie vorsichtig mit `auto` für numerische Typen, wenn präzise Kontrolle erforderlich ist:
```cpp
auto a = 5 / 2;    // a ist int (2), nicht double (2.5)
auto b = 5.0 / 2;  // b ist double (2.5)
```

# <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Praktische Anwendung</strong>

````c++
#include <iostream>
#include <vector>
#include <map>
#include <string>

int main() {
    // Complex container types where auto is particularly useful
    std::vector<int> numbers = {1, 2, 3, 4, 5};
    std::map<std::string, std::vector<int>> data = {{"Grades", {1, 2, 3}}};
    
    // Without auto (explicit declaration)
    for (std::vector<int>::iterator it = numbers.begin(); it != numbers.end(); ++it) {
        std::cout << *it << " ";
    }
    std::cout << std::endl;
    
    // With auto (more concise)
    for (auto it = numbers.begin(); it != numbers.end(); ++it) {
        std::cout << *it << " ";
    }
    std::cout << std::endl;
    
    // Even more concise with range-based for loop
    for (auto number : numbers) {
        std::cout << number << " ";
    }
    std::cout << std::endl;
    
    // Working with complex container types
    for (const auto& [key, values] : data) {
        std::cout << key << ": ";
        for (auto value : values) {
            std::cout << value << " ";
        }
        std::cout << std::endl;
    }
    
    return 0;
}
````
