 # <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Grundsätzlicher Aufbau eines C/C++ Programms</strong>


- Ein Programm besteht aus einer Abfolge von Befehlen, die vom Computer ausgeführt werden
- Dabei hat ein Programm unter anderem folgende Bestandteile
  - Eine `main` Funktion
  - Kommentare (Zeilenweise oder Blöcke)
  - Anweisungen
    - Darunter versteht man die Befehle, aus denen ein Programm besteht
  - Blöcke
    - Dienen zur Gliederung des Programmes
- Um einzelne Anweisungen aufzurufen sowie die Blöcke zu erstellen, werden Klammern benötigt
****

 # <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Welche Arten von Klammern gibt es in C/C++?</strong>


**Runde Klammern `( ... )`**
- Zur Parameterübergabe an Funktionen
- Zur Gruppierung von Ausdrücken

**Geschwungene Klammern `{ ... }`**
- Zur Definition von Strukturen
- Zur Gruppierung von Anweisungen in Blöcken
- Definieren des Geltungsbereichs (Funktionen, Namensräume)
- Initialisieren von Variablen/Listen

**Spitze Klammern `< ... >`**
- Verwendung in `#include` – Anweisung zur Angabe des Dateinamens

**Kommentar – Klammern `/* ... */`**
- Dienen zur Begrenzung von Kommentartext

**Eckige Klammern `[ ... ]`**
- Zur Indizierung von Feldern
- Deklaration von Arrays (z.B. Zeichenketten in C)

````cpp
/* file: text.cpp
   writes a text to the terminal and demonstrates bracketing
*/

#include <iostream>

// Function to print a text
void print_text() {
    std::cout << "This is an example text." << std::endl;
}

int main() {
    print_text();

    // Using index brackets with an array
    int numbers[3] = {10, 20, 30};  // Array with three elements

    // Printing array elements using index brackets
    std::cout << "First number: " << numbers[0] << std::endl;
    std::cout << "Second number: " << numbers[1] << std::endl;
    std::cout << "Third number: " << numbers[2] << std::endl;

    return 0;
} /* end main */
````

# <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Wesentliche Unterschiede zu Python</strong>

- Im Gegensatz zu Python werden Blöcke in C/C++ durch `{` begonnen und durch `}` beendet und nicht durch Einrückung gekennzeichnet
```python
# Python
if value > 10:
  print(">10")
```

```cpp
// C++
if(value > 10){
  std::cout << ">10\n";
}
```
  - **Empfehlung**: Obwohl Einrückungen in C und C++ nicht notwendig sind, wird die Verwendung eines konsistenten Einrückungsschemas, ähnlich dem in Python, jedoch dringend empfohlen
    - Dies ist sehr hilfreich, um den Code übersichtlicher zu gestalten.
  - Besteht ein Block nur aus einem einzelnen Befehl, so kann die Klammerung mit `{` und `}` entfallen
```cpp
// C++
if(value > 10)
  std::cout << ">10\n";
```
    Hinweis: Auch wenn Klammern bei einzeiligen Blöcken weggelassen werden können, wird oft empfohlen, sie trotzdem zu setzen. Das verhindert Fehler, wenn später weitere Zeilen hinzugefügt werden.

- Einzelne Befehle werden durch ein Semikolon `;` abgeschlossen
  - Auch ein alleiniges `;` ist ein Befehl
  - Achtung: C und C++ sind sehr pingelig in Bezug auf fehlende `;`
- Auch Zeilenumbrüche sind im Wesentlichen nicht notwendig
  - Beispielsweise ist folgender Code möglich, trägt jedoch nicht immer zur Verständlichkeit bei
```cpp
// C++
if(value > 10) { std::cout << ">10\n"; } else { std::cout << "<=10\n"; }
```
