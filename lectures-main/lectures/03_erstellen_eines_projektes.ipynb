# <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Was ist ein Software-Projekt?</strong>

Ein Software-Projekt ist eine strukturierte Vorgehensweise zur Entwicklung einer Softwarelösung. Es folgt einem klaren Plan und durchläuft verschiedene Entwicklungsphasen. Typische Merkmale eines Software-Projekts sind:

- **Klare Anforderungen**: Spezifikationen und Funktionen werden vorab definiert.
- **Phasenbasierte Entwicklung**: Planung, Implementierung, Testing und Deployment.
- **Iterative Verbesserungen**: Oft werden agile Methoden wie Scrum oder Kanban genutzt.
- **Technologieabhängigkeit**: Verwendung von Programmiersprachen, Frameworks und Tools.

# <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Phasen eines Software-Projekts</strong>

Ein Software-Projekt besteht aus mehreren Entwicklungsphasen:

1. **Planungsphase**  
   - Anforderungsanalyse (z. B. Lasten- und Pflichtenheft)
   - Auswahl der Technologie und Architektur
   - Erstellung eines Projektplans mit Meilensteinen

2. **Entwicklungsphase**  
   - Implementierung der Software nach definierten Anforderungen
   - Versionskontrolle und kontinuierliche Integration (CI/CD)
   - Regelmäßige Tests zur Sicherstellung der Softwarequalität

3. **Test- und Deployment-Phase**  
   - Durchführung von Unit-, Integrations- und Systemtests
   - Deployment auf eine Test- oder Produktivumgebung
   - Fehlerbehebungen und Optimierungen

## <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Beispiel:</strong>
Ein Software-Projekt zur Entwicklung einer Anwendung mit C++ Backend und einer Python-basierten Benutzeroberfläche:
- **Backend**: C++ mit Boost für effiziente Datenverarbeitung
- **Frontend**: Python mit Tkinter oder PyQt für die Benutzeroberfläche
- **Ziel**: Eine performante und benutzerfreundliche Anwendung mit responsivem UI
- **Dauer**: 6 Monate mit iterativer Entwicklung

**C++ Backend (Datenverarbeitung & Logik)**
```cpp
// Backend in C++
#include <iostream>
#include <string>

extern "C" {
    const char* processData() {
        return "Data processed successfully!";
    }
}
```

**Python Frontend (GUI mit PyQt)**
```python
# Frontend in Python mit PyQt
import sys
from PyQt5.QtWidgets import QApplication, QWidget, QPushButton, QLabel
from ctypes import CDLL, c_char_p

# Laden der C++ Bibliothek
backend = CDLL("./backend.so")
backend.processData.restype = c_char_p

def on_button_click():
    result = backend.processData().decode("utf-8")
    label.setText(result)

app = QApplication(sys.argv)
window = QWidget()
window.setWindowTitle("Python GUI mit C++ Backend")

button = QPushButton("Daten verarbeiten", window)
button.clicked.connect(on_button_click)
button.move(50, 50)

label = QLabel("", window)
label.move(50, 100)

window.show()
sys.exit(app.exec_())
```

 # <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Der Kompilierungsvorgang</strong>

- Der gesamte Kompilierungsvorgang gliedert sich in zwei Teile
  - Zuerst wird das Programm übersetzt
  - und anschließend mit den benötigten Bibliotheken verlinkt
  
**Übersetzen von Programmen**
- Das gesamte Programm wird vor der Ausführung in Maschinensprache übersetzt
- Das erfolgt in mehreren Schritten und im Rahmen dieses Vorganges wird eine Objektdatei (`.o`) erzeugt
- Dieser Vorgang wird für jede Quelldatei (`.c` bzw. `.cpp`)  ausgeführt
  - D.h. für jede Quelldatei wird vom Kompiler eine Objektdatei erzeugt


<img align="left" width="400" src="images/prog_uebersetzer.png">

- Der Präprozessor
  - Dieser verarbeitet Präprozessoranweisungen im Quelltext
  - Solche Anweisungen beginnen mit einem `#` Symbol
  - Ein Beispiele dafür ist die `#include <...>` Anweisung
- Kontrolle
  - Hier folgt Syntaxüberprüfung
- Kompiler
  - Übersetzung in Assemblersprache
- Optimierung
  - Laufzeitoptimierung
  - Minimale Codegröße
- Assemblierer
  - Übersetzung in Maschinensprache

**Linker**
- Linker (dt. *Binder*) verbindet Objektdateien und Funktionen aus Bibliotheken zu einem ausführbaren Programm

<img align="left" width="500" src="images/prog_linker.png">

- Oft benötigte Funktionen werden in externen Programmteilen (Bibliotheken) zur Verfügung gestellt
  - Mathematische Funktionen
  - Grafische Ausgaben
  - etc.

**Durchführung eines Kompilierungsvorgangs**

- Standard C++ - Compiler `g++`
- Aufruf des Kompilers erfolgt z.B. durch Eingabe von ``` g++ -o helloworld helloworld.cpp ``` im Terminal
  - Das Beispiel befindet sich im Ordner *example_helloworld* im Vorlesungsverzeichnis

```bash
g++ -c helloworld.cpp -o helloworld.o  # Übersetzen
g++ helloworld.o -o helloworld         # Verlinken
```

## <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Programmtext- und Header-Dateien</strong>

- Ein Programm besteht mitunter aus mehreren Programm- und Headerdateien
- Das können zum Beispiel verschiedene Funktionalitäten sein, welche in einem Projekt separat entwickelt werden

**Programmtext**
- Dieser besteht aus mehreren Dateien mit der Endung `.cpp` (C++) bzw. `.c` (C)
- Diese Dateien enthalten den ausführbaren Code
  - Definition von Funktionen
  - Jene Teile, welche bei der Ausführung letztendlich Speicher und Ressourcen konsumieren

**Header-Dateien**
- Enthalten Informationen, welche in mehreren Quelldateien benötigt werden können
- Tragen die Endung `.hpp` (C++) bzw. `.h` (C)
  - Deklaration (Bekanntmachung) von Funktionen
  - Definition von Datentypen
  - Keine Definition von Funktionen (Programmcode, ausprogrammierte Funktionalität)
  - Keine Variablendefinitionen (Anlegen von Variablen) und KEINE Speicherallokation für Variablen

**Beispiel für das Einbinden von Bibliotheken und eigenen .hpp Dateien**
- Die Dateien zu diesem Beispiel finden Sie im Ordner *header_example* im Vorlesungsverzeichnis
  - Zum Compilieren des Beispiels kann folgender Befehl im Terminal verwendet werden `g++ -o main main.cpp myheader.cpp`
  - Zum Ausführen dient folgender Aufruf `./main`

- Rufen wir uns hier folgenden Python-Syntax in Erinnerung
  - Einbinden einer Bibliothek in Python erfolgt über die `import` Anweisung 
```python
import numpy
import scipy
...
```
  - Kommentarzeilen beginnen in Python mit einem `#` Symbol
```python
# this is a comment
...
```
- Einbinden einer Bibliothek in C/C++ über die `include` Anweisung 

```cpp
//+++++++++++++++++++++++++++++++
// myheader.hpp
#pragma once

int add(int a, int b);
// end myheader.hpp
```

```cpp
//+++++++++++++++++++++++++++++++
// myheader.cpp
#include "myheader.hpp"

int add(int a, int b)
{
  return a+b;
}
// end myheader.cpp
```

```cpp
//+++++++++++++++++++++++++++++++
// main.cpp
#include <iostream>
#include "myheader.hpp"

int main() {
  std::cout << "2 + 3 = " << add(2, 3) << std::endl; 
  return 0;
}
// end main.cpp
```

- **Verwendung von Funktionen aus Bibliotheken**
  - Einbindung zugehöriger Header-Dateien notwendig
  - Die `include`-Anweisung unterscheidet zwei Schreibweisen:
    - `#include <datei>`: Wird für System-Header-Dateien verwendet. Der Compiler sucht in den Standard-Systemverzeichnissen.
    - `#include "datei"`: Wird für eigene Header-Dateien verwendet. Der Compiler sucht zuerst im aktuellen Verzeichnis und dann in den Systemverzeichnissen.
  - **Unterschied zwischen C++ und C Header-Dateien:**
    - **C++:** Die Standard-Header-Dateien besitzen keine Dateiendung (z.B. `#include <iostream>`).
    - **C:** Die Standard-Header-Dateien besitzen die Endung `.h` (z.B. `#include <stdio.h>` oder `#include <math.h>`).
    - **Nutzung von C-Bibliotheken in C++:**
      - Hier wird der C-Header-Datei ein `c` vorangestellt und die Endung `.h` weggelassen (z.B. `#include <cmath>` statt `#include <math.h>`).
      - Vorteil: Die Funktionen befinden sich dadurch im `std` Namensraum.

- **Verwendung der `#pragma once` Anweisung in C++**
  - Die `#pragma once` Anweisung stellt sicher, dass eine Header-Datei nur einmal vom Präprozessor eingebunden wird, auch wenn sie mehrfach via `#include` angefordert wird.
    - Doppelte `#include` Anweisungen für dieselbe Datei werden dadurch effektiv ignoriert.
  - Hinweis: Obwohl `#pragma once` von den meisten modernen Compilern unterstützt wird, ist es kein offizieller Teil des C++-Standards. Die traditionelle Methode mit Include Guards ist portabler.

- **Verwendung von Präprozessoranweisungen (Include Guards) um doppelte Einbindung von Header-Dateien zu verhindern**
  - In C und C++ ist die Standardlösung die Verwendung von sogenannten *Include Guards*:

```c
// myheader.h
#ifndef MYHEADER_H
#define MYHEADER_H

int add(int a, int b);

#endif /* !MYHEADER_H */
// end myheader.h
```

- **Der Präprozessor – Vorverarbeitung des Quellcodes**
  - Ein Beispiel für die `#define`-Anweisung, mit der Konstanten definiert werden können. Sehen wir uns folgenden Code an:

```cpp
//+++++++++++++++++++++++++++++++
// main.cpp
#define PI 3.14
#include <iostream>

int main() {
    std::cout << "PI ist: " << PI << std::endl;
    return 0;
}
// end main.cpp
//+++++++++++++++++++++++++++++++
```

- **Was passiert während der Präprozessor-Phase?**
    - Der Präprozessor ersetzt alle Vorkommen von PI durch 3.14, sodass der Compiler folgenden Code erhält:

```cpp
//+++++++++++++++++++++++++++++++
// main.cpp
#include <iostream>

int main() {
    std::cout << "PI ist: " << 3.14 << std::endl;
    return 0;
}
// end main.cpp
//+++++++++++++++++++++++++++++++
```

> **Achtung bei `#define` für Konstanten:**
> Obwohl `#define` funktioniert, ist es in C++ eine veraltete Methode zur Definition von Konstanten. Sie birgt einige Gefahren:
> - **Keine Typinformation:** `PI` ist für den Compiler nur der Text `3.14`. Dies kann zu subtilen Fehlern führen und erschwert das Debugging.
> - **Ignoriert Namensräume und Geltungsbereiche (Scopes):** Ein `#define` gilt ab seiner Definition bis zum Ende der Datei, was zu Namenskonflikten führen kann.
> 
> **Bessere Alternative in C++:**
> Verwenden Sie `const` oder `constexpr` für Konstanten. Diese sind typsicher und respektieren Geltungsbereiche.
> ```cpp
> #include <iostream>
> 
> int main() {
>     constexpr double PI = 3.14159;
>     std::cout << "PI ist: " << PI << '\n';
>     return 0;
> }
> ```
