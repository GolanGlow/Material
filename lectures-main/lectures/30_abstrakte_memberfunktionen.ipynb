# <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Abstrakte Memberfunktionen</strong>
- Eine abstrakte Memberfunktion ist eine Funktion, welche innerhalb einer Klasse deklariert, aber nicht definiert ist
- Sie dient als Platzhalter für Funktionen, welche von jeder Klasse, die von der abstrakten Klasse erbt, implementiert werden müssen
- Abstrakte Memberfunktionen (rein virtuelle Funktionen) werden mit dem Schlüsselwort `virtual` deklariert und mit `0` initialisiert
- Abstrakte Memberfunktionen werden in einer abstrakten Klasse deklariert, die oft auch als Schnittstelle (*engl. interface*) bezeichnet wird
- Klassen, die von dieser abstrakten Klasse abgeleitet sind, **müssen** Implementierungen für alle abstrakten Memberfunktionen bereitstellen

```c++
class Interface {
public:
    // Declare a pure virtual function
    virtual returnType functionName(parameters) = 0; // Pure virtual function
    // Other methods and members may also be declared
};
```

```c++
#include <iostream>  // std::cout, std::endl
#include <vector>    // std::vector

// Abstract class (interface)
class Shape {
public:
    virtual void draw() const = 0; // Abstract member function
};

// Concrete classes implementing the Shape interface
class Circle : public Shape {
protected:
    std::vector<double> center;
    double radius;
public:
    Circle(std::vector<double> c, double r) : center(c), radius(r) {}

    void draw() const override {
        std::cout << "Drawing Circle" << std::endl;
    }
};

class Rectangle : public Shape {
protected:
    std::vector<double> corner1;
    std::vector<double> corner2;
public:
    Rectangle(std::vector<double> c1, std::vector<double> c2) : corner1(c1), corner2(c2) {}

    void draw() const override {
        std::cout << "Drawing Rectangle" << std::endl;
    }
};

int main() {
    Circle myCircle({10., 10.}, 5.);
    Rectangle myRectangle({5., 5.}, {15., 15.});

    myCircle.draw();
    myRectangle.draw();

    return 0;
}
```

# <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Virtuelle Funktion</strong>

- Eine virtuelle Funktion:
    - muss nicht unbedingt eine abstrakte Funktion sein
    - wird mit dem Schlüsselwort `virtual` deklariert, aber nicht mit `0` initialisiert
    - wird in der Basisklasse definiert, aber es wird erwartet, dass sie in der abgeleiteten Klasse überschrieben wird

```c++
class Base {
public:
    virtual void virtualFunction() {
        // Base class implementation
    }
};

class Derived : public Base {
public:
    void virtualFunction() override {
        // Derived class implementation
    }
};
```

- Das Schlüsselwort `override` wird verwendet, um ausdrücklich anzugeben, dass eine Funktion in einer abgeleiteten Klasse eine virtuelle Funktion in der Basisklasse außer Kraft setzen soll
- Wenn eine Funktion mit `override` gekennzeichnet ist, wird der Compiler darüber informiert, dass diese Funktion eine virtuelle Funktion einer Basisklasse überschreibt
- Wenn die Funktion nicht tatsächlich eine virtuelle Funktion einer Basisklasse überschreibt, erzeugt der Compiler einen Fehler und hilft somit, mögliche Fehler im Code zu erkennen
- Obwohl das Schlüsselwort `override` nicht zwingend erforderlich ist, kann es dabei helfen, Fehler frühzeitig zu erkennen und die Lesbarkeit und Wartbarkeit des Codes zu verbessern

## <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Beispiel: Dynamische Bindung</strong>

- Im folgenden Beispiel hat die Basisklasse `Shape` eine virtuelle Funktion `calculateArea()` mit einer Standardimplementierung.
    - Die abgeleiteten Klassen überschreiben diese mit ihrer eigenen Version.
    - Dank `virtual` wird die richtige Funktion zur Laufzeit aufgerufen, auch wenn man auf das Objekt über eine Referenz auf die Basisklasse zugreift. Welche Funktion ausgeführt wird, hängt vom tatsächlichen Typ des Objekts ab, nicht vom Typ der Variable.

```c++
#include <iostream>  // std::cout, std::endl

class Shape {
public:
    virtual double calculateArea() const {
        return 0.0;
    }
};

class Circle : public Shape {
private:
    double radius;

public:
    Circle(double r) : radius(r) {}

    double calculateArea() const override {
        return 3.14159 * radius * radius;
    }
};

class Rectangle : public Shape {
private:
    double width;
    double height;

public:
    Rectangle(double w, double h) : width(w), height(h) {}

    double calculateArea() const override {
        return width * height;
    }
};

int main() {
    Circle myCircle(5.0);
    Rectangle myRectangle(10.0, 5.0);

    // direct calls on concrete types, nothing special yet
    std::cout << "myCircle area:    " << myCircle.calculateArea()    << std::endl;
    std::cout << "myRectangle area: " << myRectangle.calculateArea() << std::endl;

    // virtual dispatch: accessed through a base class reference,
    // but the correct overridden function is still called at runtime
    Shape& s1 = myCircle;
    Shape& s2 = myRectangle;
    std::cout << "Shape s1 area: " << s1.calculateArea() << std::endl;  // Circle
    std::cout << "Shape s2 area: " << s2.calculateArea() << std::endl;  // Rectangle

    return 0;
}
```
