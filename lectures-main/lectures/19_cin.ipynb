# <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Terminaleingabe in Python</strong>

- In Python haben wir die Funktion `input()` verwendet, um Benutzereingaben aufzulesen
    - Die Funktion `input()` fordert den Benutzer auf, etwas einzugeben
    - Die Benutzereingabe wird von der Funktion als Zeichenkette zurückgegeben
    - Um die Eingabe als einen anderen Typ (z. B. `int` oder `float`) zu interpretieren, werden Typumwandlungsfunktionen wie `int()` oder `float()` verwendet

```c++
name = input("Enter your name: ")
print("Your name is", name)

age = int(input("Enter your age: "))
print("You are", age, "years old.")
```

# <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Eingabefunktionalität in C++</strong>
- In C++ wird die entsprechende Funktionalität von Pythons `input()` typischerweise mit `std::cin` erreicht
    - Genau wie `std::cout` ist auch `std::cin` Teil der Standard Input/Output Library `iostream`
    - Bei `std::cin` handelt es sich dabei um ein Objekt der Klasse `std::istream`
    - Das bedeutet, dass die Anweisung `#include <iostream>` vor der Verwendung von `std::cin` angegeben sein muss
    - `std::cin` verwendet den Operator `>>`, um formatierte Werte aus der Eingabequelle einzulesen
- Syntax: `std::cin >> variable;`
    - `variable`: Die Variable, in der die Eingabedaten gespeichert werden sollen
- Ein Beispiel, das den Benutzer auffordert, eine Zahl einzugeben, diese mit `std::cin` einliest und dann wieder ausgibt:

```c++
#include <iostream>  // std::cout, std::endl

int main() {
    int number;
    std::cout << "Enter a number: ";
    std::cin >> number;
    std::cout << "You entered: " << number << std::endl;
    return 0;
}
```

- Beachten Sie, dass wir den eingelesenen Input nicht explizit in einen `int`-Wert konvertieren müssen
    - Aufgrund der direkten Festlegung von Datentypen in C++ ist dem Compiler bekannt, dass der zu lesende Wert ein `int` ist
    - Der Extraktionsoperator `>>` versucht, die eingegebenen Zeichen passend zum Zieltyp zu interpretieren
    - Passt die Eingabe nicht zum erwarteten Typ, wird der Datenstrom in einen Fehlerzustand versetzt
    - Der Datentyp der gelesenen Variablen muss also mit der erwarteten Eingabe übereinstimmen

```c++
#include <iostream>  // std::cout, std::endl

int main() {
    int intValue;
    double doubleValue;
    char charValue;

    std::cout << "Enter an integer: ";
    std::cin >> intValue;

    std::cout << "Enter a double: ";
    std::cin >> doubleValue;

    std::cout << "Enter a character: ";
    std::cin >> charValue;

    std::cout << "Integer: " << intValue << std::endl;
    std::cout << "Double: " << doubleValue << std::endl;
    std::cout << "Character: " << charValue << std::endl;

    return 0;
}
```

- `std::cin >>` liest bei Zeichenketten nur bis zum nächsten Leerzeichen oder Zeilenende
- Für ganze Zeilen mit Leerzeichen verwendet man stattdessen `std::getline(...)`

## <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Datenströme</strong>
- Sowohl `std::cin` als auch `std::cout` verwenden gepufferte Datenströme
    - Bei Terminaleingaben wird die Eingabe dem Programm in der Regel erst nach dem Drücken der Eingabetaste zur Verfügung gestellt
    - Ein einfaches Newline-Zeichen `"\n"` leert den Ausgabepuffer im Allgemeinen nicht garantiert
    - `std::endl` schreibt ein Zeilenende und leert anschließend den Ausgabepuffer
    - Alternativ kann der Puffer auch mit `std::flush` oder durch direkten Aufruf von `std::cout.flush()` geleert werden
    - Vor einer Eingabe über `std::cin` wird `std::cout` oft automatisch geleert, da beide Datenströme standardmäßig gekoppelt sind

## <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Eingabefehlerbehandlung</strong>
- Im Gegensatz zu Python werden in C++ unerwartete Eingaben nicht automatisch behandelt
- Mögliche Eingabefehler müssen explizit über den Zustand des Datenstroms geprüft und behandelt werden
    - Häufig verwendet man dazu `fail()`, `clear()` und `ignore()`

```c++
#include <iostream>  // std::cout, std::endl
#include <limits>    // std::numeric_limits

int main() {
    int number;

    std::cout << "Enter an integer: ";
    std::cin >> number;

    if (std::cin.fail()) {
        std::cout << "Invalid input!" << std::endl;
        std::cin.clear();
        std::cin.ignore(std::numeric_limits<std::streamsize>::max(), '\n');
        // clear(): entfernt den Fehlerzustand des Streams
        // ignore(...): verwirft den Rest der aktuellen Eingabezeile
    } else {
        std::cout << "You entered: " << number << std::endl;
    }

    return 0;
}
```

- Soll die Eingabe wiederholt werden können, bis gültige Daten eingegeben wurden, wird der Code normalerweise in eine Schleife eingebaut

```c++
#include <iostream>  // std::cout, std::endl
#include <limits>    // std::numeric_limits

int main() {
    int number;
    bool valid_input = false;

    // Keep prompting the user until valid input is provided
    while (!valid_input) {
        std::cout << "Enter an integer: ";
        if (std::cin >> number) {
            // Input is valid
            valid_input = true;
        } else {
            std::cin.clear();
            std::cin.ignore(std::numeric_limits<std::streamsize>::max(), '\n');
            // clear(): entfernt den Fehlerzustand des Streams
            // ignore(...): verwirft den Rest der aktuellen Eingabezeile
            std::cout << "Invalid input! Try again..." << std::endl;
        }
    }
    
    std::cout << "Your integer is " << number << std::endl;

    return 0;
}
```
