## <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Freundschaftsfunktionen (*engl. friend functions*)</strong>
- Eine `friend`-Funktion einer Klasse ist eine Funktion, die kein Member der Klasse ist, aber Zugriff auf die privaten und geschützten Member der Klasse hat

- **Deklaration**: `friend`-Funktionen werden mit dem Schlüsselwort `friend` innerhalb der Klasse deklariert, die Zugriff auf ihre privaten oder geschützten Member gewähren will

- **Definition**: Die `friend`-Funktion kann innerhalb der Klasse, zusammen mit ihrer Definition, oder außerhalb der Klasse definiert werden

```c++
#include <iostream>  // std::cout, std::endl

class MyClass {
private:
    int data;

public:
    MyClass(int d) : data(d) {}

    // Declaration of friend function
    friend void displayData(const MyClass& obj);
};

// Definition of the friend function displayData
void displayData(const MyClass& obj) {
    // Accessing private member of MyClass is permitted
    std::cout << "Data: " << obj.data << std::endl;
}

int main() {
    MyClass obj(360);
    displayData(obj); // Call to friend function
    return 0;
}
```

## <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Freundschaftsklassen (*engl. friend classes*)</strong>
- Ähnlich wie Freundschaftsfunktionen können Freundschaftsklassen auf die privaten und geschützten Member einer anderen Klasse zugreifen

- **Deklaration**: Die Klasse kann innerhalb einer anderen Klasse mit dem Schlüsselwort `friend` als Freundesklasse deklariert werden

- **Definition**: Die Definition der `friend`-Klasse liegt außerhalb der Klasse

```c++
#include <iostream>

class FriendClass;

class MyClass {
private:
    int data;

    friend class FriendClass; // Declaration of friend class inside MyClass
public:
    MyClass(int d) : data(d) {}
};

class FriendClass {
public:
    void displayData(const MyClass& obj) {
        std::cout << "Data: " << obj.data << std::endl; // Accessing private member of MyClass
    }
};

int main() {
    MyClass obj(360);
    FriendClass fc;
    fc.displayData(obj); // Call to member function of friend class
    return 0;
}
```

## <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Vorteile</strong>
- Freundschaftsklassen und -funktionen bieten Flexibilität beim Zugriff auf private und geschützte Member einer Klasse
    - Die direkte Verbindung mit Klassenvariablen (ohne Getter- und Setter-Funktionen) vereinfacht den Code, da unnötige Funktionsaufrufe vermieden werden
    - Manchmal benötigt ein externes Objekt für bestimmte Operationen Zugriff auf interne Klassendetails. Mit Hilfe von Freundschaftsklassen und -funktionen kann dies erreicht werden, ohne dass alle Variablen `public` gemacht werden müssen
    - Durch den direkten Zugriff auf `private` Member kann die Leistung verbessert werden, da weniger Funktionsaufrufe und Datenkopien erforderlich sind

```c++
#include <iostream>   // std::cout, std::endl

class Vector {
private:
    double x;
    double y;
    double z;

public:
    Vector(double x, double y, double z) : x(x), y(y), z(z) {}

    // Getters for non-friend callers that only have public access
    double get_x() const {
        return x;
    }
    double get_y() const {
        return y;
    }
    double get_z() const {
        return z;
    }

    // dotProduct is declared as friend, so it can access x, y, z directly
    // without going through the getters — this is the advantage of friend
    friend double dotProduct(const Vector& v1, const Vector& v2);
};

double dotProduct(const Vector& v1, const Vector& v2) {
    return v1.x * v2.x + v1.y * v2.y + v1.z * v2.z;
}

int main() {
    Vector v1(1.0, 2.0, 3.0);
    Vector v2(4.0, 5.0, 6.0);

    std::cout << "Dot product of v1 and v2: " << dotProduct(v1, v2) << std::endl;

    return 0;
}
```

## <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Nachteile</strong>
- Freundschaftsklassen und -funktionen stellen jedoch auch einige Gefahren dar:
    - Der Programmcode wird schwieriger zu verwalten und zu verstehen, da es unübersichtlich wird, welche Teile einer Klasse wirklich privat sind
    - Der Programmcode kann fehleranfälliger werden, da unerwartete Änderungen an den privaten Membern einer Klasse möglicherweise nicht richtig kontrolliert oder berücksichtigt werden
    - Es kann schwieriger sein, das Verhalten einzelner Klassen zu isolieren, zu testen und zu debuggen, da die Interaktionen zwischen Klassen nicht so klar sind
    - Kann zu Sicherheitslücken führen, insbesondere in Systemen, in denen Datenintegrität und Datenschutz kritisch sind:

```c++
#include <iostream>   // std::cout, std::endl
#include <string>     // std::string

class BankAccount {
private:
    std::string IBAN;
    double balance;
    bool frozen = false;

public:
    BankAccount(const std::string& IBAN_, double initBalance)
        : IBAN(IBAN_), balance(initBalance) {}

    double get_balance() const { return balance; }

    friend class BankManager;
};

class BankManager {
public:
    void freezeAccount(BankAccount& account);
};

void BankManager::freezeAccount(BankAccount& account) {
    account.frozen = true;
    std::cout << "Account " << account.IBAN << " frozen." << std::endl;
    account.balance = 0;  // This should not be allowed!
}

int main() {
    BankAccount account("123456789", 1000.0);

    std::cout << "Balance before freeze: " << account.get_balance() << std::endl;

    BankManager manager;
    manager.freezeAccount(account);

    std::cout << "Balance after freeze: " << account.get_balance() << std::endl;

    return 0;
}
```
