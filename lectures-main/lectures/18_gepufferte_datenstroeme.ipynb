# <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Datenströme</strong>

- Viele Ein- und Ausgabedatenströme in Python und C++ arbeiten gepuffert
- Das bedeutet, dass die Daten nicht bei jeder einzelnen IO-Operation sofort an das Zielgerät geschrieben oder von diesem gelesen werden
- Stattdessen werden Ein- oder Ausgabedaten zunächst in einem Zwischenspeicher (Puffer) gesammelt
- Die eigentliche Übertragung erfolgt typischerweise blockweise, beim Schließen des Datenstroms oder beim expliziten Leeren des Puffers

<img align="left" width="800" src=images/gepufferter_strom.png>

# <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Vorteile der Datenpufferung</strong>
- Das Puffern reduziert die Anzahl der Systemaufrufe, welche für IO-Operationen erforderlich sind
    - Durch die Gruppierung mehrerer Lese- oder Schreibvorgänge in größeren Blöcken wird die Effizienz verbessert
- Durch die Pufferung können Daten vorübergehend im Arbeitsspeicher gespeichert werden, bevor sie auf die Festplatte geschrieben oder von ihr gelesen werden
    - Dadurch wird die Anzahl der Festplattenzugriffe minimiert
    - Festplatten-IO ist in der Regel langsamer als Arbeitsspeicheroperationen
- Bei der Netzwerkprogrammierung ermöglicht die Pufferung das Sammeln und Senden von Daten in größeren Datenblöcken, wodurch der mit der Übertragung kleiner Datenmengen über das Netzwerk verbundene Overhead reduziert wird

# <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Mögliche Nachteile</strong>
- Puffer verbrauchen Arbeitsspeicher, daher ist es wichtig, die Vorteile der Pufferung gegen die verfügbaren Speicherressourcen abzuwägen
- Gepufferte Daten können bei Systemfehlern oder -abstürzen verloren gehen, daher sind geeignete Fehlerbehandlungs- und Wiederherstellungsmechanismen unerlässlich
- Ein zu kleiner Puffer kann zu häufigem Leeren (engl. *flush*) oder zu einer schlechten Ausnutzung führen, während ein zu großer Puffer eine erhöhte Latenzzeit oder einen höheren Speicherverbrauch verursachen kann
- In C++ kann das Leeren des Ausgabepuffers vom Programmierer explizit ausgelöst werden

# <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Puffer explizit leeren</strong>
- `std::flush` leert den Ausgabepuffer, ohne ein Zeilenende auszugeben
- `std::endl` schreibt ein Zeilenende und leert anschließend ebenfalls den Puffer
- Ein einfaches Newline-Zeichen wie `"\n"` oder `'\n'` erzeugt nur einen Zeilenumbruch, aber keinen garantierten `flush`
- Das explizite Leeren ist z. B. für Statusmeldungen, Fortschrittsanzeigen oder Prompts ohne Zeilenumbruch nützlich

```cpp
#include <iostream>  // std::cout, std::flush

int main() {
    std::cout << "Test output...";
    std::cout << std::flush;  // force buffered output to become visible immediately

    std::cout << "\nOutput has been flushed!\n";
    return 0;
}
```
