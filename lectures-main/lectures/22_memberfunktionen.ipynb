# <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Memberfunktionen</strong>
- In C++ kann eine Struktur sowohl Datenelemente als auch Funktionen (sogenannte Memberfunktionen) enthalten
- Memberfunktionen bieten eine Möglichkeit, das Verhalten im Zusammenhang mit den Daten der `struct` zu kapseln
- Memberfunktionen sind innerhalb der `struct` definiert und können mit den Datenelementen der `struct` arbeiten
- Sie werden ähnlich wie reguläre Funktionen deklariert, werden aber innerhalb der `struct`-Deklaration angegeben
- Die Memberfunktionen in einer C++ `struct` haben eine starke Ähnlichkeit mit den Methoden einer Klasse in Python und werden ebenfalls über den `.` Operator angesprochen

````c++
#include <iostream> // std::cout, std::endl

// Define a struct called Point
struct Point {
    int x;
    int y;

    // Member function to set the coordinates of the Point
    void setCoordinates(int x_val, int y_val) {
        x = x_val;
        y = y_val;
    }

    // Member function to return the sum of the coordinates
    int sum() {
        return x + y;
    }

    // Member function to print the coordinates of the Point
    void printCoordinates() {
        std::cout << "x = " << x << ", y = " << y;
    }
};

int main() {
    Point myPoint = {1, 2};
    myPoint.printCoordinates();
    std::cout << std::endl;

    myPoint.setCoordinates(3, 2);
    std::cout << "sum of x and y when ";
    myPoint.printCoordinates();
    std::cout << " is " << myPoint.sum() << std::endl;
    return 0;
}
````

## <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Verwendung von Memberfunktionen</strong>
- Memberfunktionen werden auf `struct`-Objekten mit dem Punktoperator (`.`) aufgerufen, ähnlich wie beim Zugriff auf Datenelemente
- Standardmäßig können Memberfunktionen auf `struct`-Elemente zugreifen und diese verändern

### <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Python-Implementierung der Point-Klasse</strong>
- Eine Python-Implementierung des `struct Point` kann mit einer Klasse realisiert werden:

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def set_coordinates(self, x_val, y_val):
        self.x = x_val
        self.y = y_val

    def sum(self):
        return self.x + self.y

    def print_coordinates(self):
        print("x =", self.x, ", y =", self.y, end='')


my_point = Point(1, 2)
my_point.print_coordinates()
print()

my_point.set_coordinates(3, 2)
print("sum of x and y when ", end='')
my_point.print_coordinates()
print(" is", my_point.sum())
```

## <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Header-Dateien</strong>
- In C++ ist es üblich, oftmals verwendete Strukturen in Header-Dateien zu definieren und für andere Dateien sichtbar zu machen
- Die eigentliche Funktionalität wird in C++ Sourcedateien vorgenommen, welche die Header-Datei inkludieren
- Strukturdefinition in `point.hpp`:

````c++
// File point.hpp
#pragma once        // Prevent multiple inclusions
#include <iostream> // std::cout

// Define a struct called Point
struct Point {
    int x;
    int y;

    // Member function to set the coordinates of the point
    void setCoordinates(int x_val, int y_val);

    // Member function to return the sum of the coordinates
    int sum();

    // Member function to print the coordinates of the Point
    void printCoordinates();
};
````

- Um Memberfunktionen außerhalb der Struktur zu definieren, wird der Strukturname mit `::` vorangestellt
- Funktionalität in `point.cpp`:

````c++
// File point.cpp
#include "point.hpp"

// Member function to set the coordinates of the Point
void Point::setCoordinates(int x_val, int y_val) {
    x = x_val;
    y = y_val;
}

// Member function to return the sum of the coordinates
int Point::sum() {
    return x + y;
}

// Member function to print the coordinates of the Point
void Point::printCoordinates() {
    std::cout << "x = " << x << ", y = " << y;
}
````

- `main.cpp` inkludiert `point.hpp` und verwendet die `Point`-Struktur:

````c++
// File main.cpp
#include "point.hpp"

int main() {
    Point myPoint = {1, 2};
    myPoint.printCoordinates();
    std::cout << std::endl;

    myPoint.setCoordinates(3, 2);
    std::cout << "sum of x and y when ";
    myPoint.printCoordinates();
    std::cout << " is " << myPoint.sum() << std::endl;
    return 0;
}
````

- Die Trennung von Deklaration (`.hpp`) und Definition (`.cpp`) ist nicht verpflichtend, verbessert aber die Lesbarkeit und Kompilierungszeiten, besonders bei großen Projekten mit vielen Quelldateien
- Das obige Beispiel kann im Ordner `point_example/` kompiliert und ausgeführt werden:
```bash
g++ -std=c++17 main.cpp point.cpp -o point
./point
```
