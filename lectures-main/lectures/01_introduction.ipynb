## <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Motivation Programmieren zu Lernen</strong>

- Die Digitalisierung unserer Gesellschaft schreitet immer weiter voran
- Computer und Computerchips sind heutzutage aus unserem Alltag nicht mehr wegzudenken

**Einsatz von Computerprogrammen zur Lösung technischer Fragestellungen (ein kleiner Auszug)** 
- Die Wahl der Programmiersprache hängt von der Fragestellung/dem Problem ab
- Messdatenverarbeitung (z.B. im Rahmen einer Bachelorarbeit oder für einen Laborbericht, aber auch im Berufsleben)
  - Auswertung von Messdaten
  - Darstellung von Messdaten
- Lösen von komplexen mathematischen Gleichungssystemen
  - Simulation von Halbleiterbauelementen
  - Modellierung von Mobilfunkkanälen in der Nachrichtentechnik
  - Optimierung von Regelungsaufgaben
- Weitere moderne Anwendungsgebiete
  - Entwicklungen von Systemen im Bereich künstlicher Intelligenz (maschinelles Lernen)
  - Autonomes Fahren (Objekterkennung, Sensordatenverwaltung, ...)
- Beispiele aus dem Alltag
  - Steuerung von Geräten (Waschmaschine, Smart-Watch, Handy, ...)
  - Buchhaltung

**Warum systemnahes Programmieren?**
- Viele dieser Anwendungen erfordern effiziente, ressourcenschonende und zuverlässige Software.
- In technischen Systemen sind Ressourcen wie Speicher, Rechenleistung oder Energie oft begrenzt.
- Systemnahes Programmieren bedeutet, Hardwareressourcen bewusst zu kontrollieren und das Laufzeitverhalten gezielt zu beeinflussen.
- C und C++ ermöglichen genau diese Kontrolle und bilden die Grundlage vieler technischer Systeme.

# <strong style="color: rgb(0, 102, 153); opacity: 1.0;">&nbsp;&nbsp;&nbsp;&nbsp; Fortran (engl. *for*mular *tran*slator)</strong>


- Ursprache aller algorithmisch orientierten wissenschaftlichen Programmiersprachen
- Entwickelt 1954 – 1956 von IBM für wissenschaftliche Berechnungen
- Sehr einfach und leicht lernbar
- Sehr effiziente Zielcodes

![Alt text](images/fortran_example.png)

**Innovationen (Auszug)**
- Mathematische Notation für arithmetische Ausdrücke
- Symbolische Namen für Variablen und Unterprogramme
- Felder
- Schleifen mit Indizes
- IF Anweisungen

*Vor Fortran erfolgte die Programmierung in Maschinencode oder Assembler, wobei Speicheradressen und Hardwaredetails vom Programmierer explizit verwaltet werden mussten.*

Heute noch verbreitet
  - Fortran 90
  - HPF (engl. high performance fortran)
  - Fortran 2003
****

## <strong style="color: rgb(0, 102, 153); opacity: 1.0;">&nbsp;&nbsp;&nbsp;&nbsp; COBOL (engl. *c*ommon *b*usiness *o*riented *l*anguage)</strong>


- Entwickelt Ende der 1950er Jahre
- Beauftragt vom U.S. Verteidigungsministerium
- Erstellung von Programmen für betriebswirtschaftliche Anwendungen
- Konzeption für Geschäftsanwendungen
- Hardwareunabhängige Sprache
- Handhabung großer Datenmengen

- Anlehnung an die natürliche Sprache

# <strong style="color: rgb(0, 102, 153); opacity: 1.0;">&nbsp;&nbsp;&nbsp;&nbsp; LISP (engl. *lis*t *p*rocessing)</strong>

- Entwickelt Ende der 1950er Jahre am MIT unter John McCarthy
- Ursprünglich gedacht für Fortran-Unterprogramme zur Durchführung symbolischer Berechnungen
- Verwendet die Liste als universelle Datenstruktur, wodurch Programmcode und Daten die gleiche Form besitzen
- Führte bahnbrechende Konzepte wie die automatische Speicherbereinigung (*engl. Garbage Collection*) und Rekursion als Standard ein
- Ermöglicht durch seine Flexibilität, dass Programme ihren eigenen Code zur Laufzeit modifizieren oder erweitern können
- Wurde über Jahrzehnte zur dominierenden Sprache im Bereich der Künstlichen Intelligenz und symbolischen Informationsverarbeitung

![Alt text](images/lisp_example1.png)

**Innovationen (Auszug)**
- S-Ausdrücke
  - Kleinste Einheit:  ATOM  A
  - Ausdrücke: CELL (A . B) oder (A . (B . C)) oder (A . B . C)
  - Erstes Element enthält den Namen der Funktion
- Rekursive Funktionen
  
  ![Alt text](images/lisp_example2.png)
- Funktionen höherer Ordnungen
- Dynamische Typisierung (Typprüfung findet zur Laufzeit des Programms statt)
- Dynamische Speicherverwaltung (garbage collection)
****

# <strong style="color: rgb(0, 102, 153); opacity: 1.0;">&nbsp;&nbsp;&nbsp;&nbsp; 1964: BASIC (engl. *b*eginner‘s *a*ll-purpose *s*ymbolic *i*nstruction *c*ode)</strong>

- Wurde entwickelt, um (Elektrotechnik-)Studenten den Einstieg in die Programmierung gegenüber ALGOL und Fortran zu erleichtern
- Bietet eine besonders einfache Syntax mit klaren Fehlermeldungen für Programmieranfänger
- Fokus auf Benutzerfreundlichkeit mit klaren Fehlermeldungen und eine intuitive, englischnahe Syntax (z. B. `PRINT`, `INPUT`, `GOTO`)
- Ermöglichte durch Hardwareunabhängigkeit und Interpreter-Technik die spätere Heimcomputer-Revolution
- Führte das Prinzip des interaktiven Time-Sharing ein, was direktes Feedback am Terminal erlaubte

<img align="left" height="75" src="images/logo_bcpl.png">
<br>

# <strong style="color: rgb(0, 102, 153); opacity: 1.0;">&nbsp;&nbsp;&nbsp;&nbsp; 1967: BCPL (engl. *b*asic *c*ombinded *p*rogramming *l*anguage)</strong>

- Erkennt nur das „Wort“ als Datentyp; Variablen sind lediglich Verweise auf Speicherzellen (typlose Sprache)
- Definition von L-Value (Speicheradresse) und R-Value (Inhalt), was die Basis für moderne Compiler-Logik legte
- Der Compiler erzeugt O-Code, der auf einer virtuellen Maschine läuft und BCPL extrem portabel machte
- Ersetzte die Zeilennummern von BASIC durch echte Blockstrukturen und geschachtelte Logik

<img align="left" width="150" height="150" src="images/logo_b.png">
<br>

# <strong style="color: rgb(0, 102, 153); opacity: 1.0;">&nbsp;&nbsp;&nbsp;&nbsp; 1969: B</strong>

- Starke Anlehnung an BCPL
- Entwickelt von Ken Thompson und Dennis Ritchie
- Erste Sprache mit `++` / `--` Operatoren
- Konzipiert für die Entwicklung des frühen UNIX

<img align="left" width="150" height="150" src="images/logo_pascal.png">
<br>

# <strong style="color: rgb(0, 102, 153); opacity: 1.0;">&nbsp;&nbsp;&nbsp;&nbsp; 1970: PASCAL/Turbo-PASCAL</strong>


- Universelle Programmiersprache für mathematische und kommerzielle Aufgaben
- PASCAL Compiler gibt es für alle Systeme
- Stark typisierte Sprache, die durch klare Strukturen fehlerresistenten Code fördert

## <strong style="color: rgb(0, 102, 153); opacity: 1.0;">&nbsp;&nbsp;&nbsp;&nbsp; 1972: C</strong>

- Entwickelt von Dennis Ritchie als direkte Weiterentwicklung der Sprache B
- Verbessert von Brian W. Kernighan in den Jahren 1973/1974
- Ermöglichte die Portierung des Betriebssystems UNIX von Assembler auf eine Hochsprache
- **Prozedurale Hochsprache**
  - Einfacher Syntax, d.h. wenig Schlüsselwörter
  - Umfangreiche Semantik, d.h. viele Kombinationsmöglichkeiten
- Möglichkeit maschinennaher Programmierung
- Unterstützung von strukturierter, modularer Programmierung
- Auf fast allen Plattformen verfügbar
- Anwendungen
  - Kernels von Betriebssystemen: Windows, Linux, macOS, Android, iOS
  - Datenbanken: Oracle, MySQL,...
  - Embedded Systems
****

<img align="left" width="150" height="150" src="images/logo_c++.png">
<br>

# <strong style="color: rgb(0, 102, 153); opacity: 1.0;">&nbsp;&nbsp;&nbsp;&nbsp; 1979: C++</strong>

- Entwickelt von Bjarne Stroustrup bei den Bell Labs
- Anforderungen
  - Effiziente Programmorganisation (Klassen, Hierarchien, ...)
  - Schnelle prozedurale Programme
  - Möglichkeit separat kompilierbare Einheiten in ein Programm zusammenzufassen
  - Portable Implementierungen (nur minimale Limitierung an ein Betriebssystem)
- Im Jahr 1979 startete die Entwicklung von C mit Klassen (*engl. C with classes*)
- Erste Implementierung 1980 enthielt
  - Klassen
  - Abgeleitete Klassen
  - Zugriffskontrolle (`public`/`private`)
  - Konstruktor und Destruktor
  - Aufruf- sowie Rückgabefunktion
    - Werden implizit vor jedem Funktionsaufruf/jeder Rückgabe aufgerufen
    - Kann zur Synchronisierung von Prozessen verwendet werden
  - `friend` Klassen (Zugriff auf `private` und `protected` Elemente)
- Erweiterungen 1982
  - Inline Funktionen
  - Default Argumente
  - Überladen von Operatoren
- Erster C++ Standard wurde 1998 als ISO/IEC 14882:1998 festgelegt
- C++ ist eine sehr dynamische Sprache: C++03, C++07, C++11, C++17, **C++20**, C++23
  - Abwärtskompatibilität nicht garantiert

  - Dynamische Sprache, welche ständig weiterentwickelt wird
  - [TIOBE Programming Community Index Feb. 2026](https://www.tiobe.com/tiobe-index/)

![Alt text](images/tiobe.png)

- **Warum C++ im und nicht nur Python?**
  - C++ wird zu Maschinencode kompiliert und ermöglicht sehr effiziente Programme mit geringem Laufzeit-Overhead.
  - Speicherverwaltung, Datenlayout und Objektlebensdauer können gezielt kontrolliert werden, was für performanzkritische und systemnahe Anwendungen entscheidend ist.
  - C++ wird in Betriebssystemen, Embedded Systems, Hochleistungsrechnern, Datenbanken und Echtzeitsystemen eingesetzt.
  - Viele rechenintensive Python-Bibliotheken (z.B. NumPy, SciPy, PyTorch, TensorFlow, OpenCV) sind intern in C oder C++ implementiert.
  - Python dient häufig als komfortable Schnittstelle, während die eigentliche Rechenarbeit durch optimierten C/C++-Code erfolgt.
  - Das Verständnis von C++ vermittelt grundlegende Kenntnisse über Speicher, Performance und Systemarchitektur, die für das Verständnis moderner Softwaresysteme essenziell sind.

- **C++ hat einen enormen Funktionsumfang**
  - Wir werden diesen bei weitem nicht erschöpfend behandeln können

- **Was ist das Lernziel der Lehrveranstaltung?**
  - Kennenlernen von Werkzeugen der Softwareentwicklung (GIT, Grafische Programmierumgebung)
  - Grundlagen des **Programmierens in C**
    - Funktionen, Schleifen, Selektionen
    - Variablen, Arrays
    - Zeiger, dynamische Speicherverwaltung
  - Kennenlernen von Konzepten des **objektorientieren Programmierens**
    - Klassen (Konstruktor/Destruktor, Zugriffskontrolle, Vererbung, Polymorphismus,...)
    - Überladen von Funktionen und Operatoren
    - Templates
  - Ausarbeiten einfacher Codebeispiele zum Erlernen der Konzepte und der Syntax

- **Welche Themen werden nicht behandelt?**
  - Ausnahmebehandlung (engl. *exception handling*)
  - Multithreading und Synchronization von Prozessen
  - u.v.m.
