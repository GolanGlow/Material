# <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Was ist `this`?</strong>
- `this` ist ein Zeiger (Pointer, Verweis) 
- `this` wird automatisch vom Compiler für jede Struktur oder Klasse generiert
- Der `this`-Zeiger steht den (nicht statischen) Memberfunktionen einer Struktur oder Klasse zur Verfügung
- Er verweist auf die aktuelle Instanz/das aktuelle Objekt 
- Über ihn kann auf die Member der Struktur zugegriffen werden
- Zeiger werden wir noch genauer kennenlernen
    - Zu beachten ist an dieser Stelle nur, dass hierbei auf die Member mit `->` zugegriffen werden muss, nicht mit `.`

`this` kann verwendet werden, um das aktuelle Objekt an eine Funktion weiterzureichen oder als Rückgabewert zu übergeben.\
Innerhalb einer Memberfunktion kann der `this`-Zeiger verwendet werden, um auf die Datenelemente des aktuellen Objekts zuzugreifen.
Dies ist besonders wichtig, wenn es einen Namenskonflikt zwischen einem Parameter und einem Datenelement gibt:

```c++
#include <iostream> // std::cout, std::endl

// Define a struct called Point
struct Point {
    int x;
    int y;

    // Member function to set the coordinates of the Point
    void setCoordinates(int x, int y) {
        this->x = x;  // which x is which? better to use this->x = x;
        this->y = y;  // which y is which? better to use this->y = y;
        // Here this-> is necessary to distinguish between the parameter and the struct member
    }

    // Member function to return the sum of the coordinates
    int sum() const {
        return x + y; // Here x and y are well-defined. this-> may be used.
    }

    // Member function to print the coordinates of the Point
    void printCoordinates() const {
        std::cout << "x = " << x << ", y = " << y;
    }

};

int main() {
    Point myPoint;
    myPoint.setCoordinates(10, 20);
    myPoint.printCoordinates();
    
    return 0;
}
```

- Es ist vergleichbar mit `self` in Python, muss aber nicht explizit als Parameter definiert werden:

```c++
class MyClass:
    # Member function to set the value
    def set_value(self, value):
        self.value = value

    # Member function to print the value
    def print_value(self):
        print("Value:", self.value)

# Create an object of MyClass and print it
obj = MyClass()
obj.set_value(360)
obj.print_value()
```

- Allerdings, darf `this` nicht explizit in der Funktionssignatur enthalten sein:

```c++
#include <iostream>

// Define a struct called MyClass
struct MyClass {
    int value;

    // Member function to set the value
    void set_value(int value) {
        // Use 'this->value' to resolve name conflict with parameter
        this->value = value;
    }

    // Member function to change the value using a distinct parameter name
    void change_value(int new_value) {
        // 'this->' is optional here; no name conflict
        value = new_value;
    }

    // Member function to print the value
    void print_value() const {
        // 'this->' is optional here; used for clarity
        std::cout << "Current value: " << this->value << std::endl;
    }
};

int main() {
    MyClass obj;

    std::cout << "=== Setting value using set_value() ===" << std::endl;
    obj.set_value(360);
    obj.print_value();

    std::cout << "\n=== Changing value using change_value() ===" << std::endl;
    obj.change_value(120);
    obj.print_value();

    return 0;
}
```

Selbstverständlich, ist `this` auch in Memberdefinitionen verfügbar, welche außerhalb der eigentlichen Strukturdefinition erfolgen.

```c++
#include <iostream>

// Define a struct called MyClass
struct MyClass {
    int value;

    // Declare member functions
    void setValue(int value);
    void printValue() const;
};

// Define setValue outside the struct
void MyClass::setValue(int value) {
    // Use 'this' to resolve naming conflict between parameter and member
    this->value = value;
}

// Define printValue outside the struct
void MyClass::printValue() const {
    // 'this->' is optional here; used for clarity
    std::cout << "Current value: " << this->value << std::endl;
}

int main() {
    MyClass obj;

    std::cout << "=== Using setValue() ===" << std::endl;
    obj.setValue(360);

    std::cout << "=== Using printValue() ===" << std::endl;
    obj.printValue();

    return 0;
}
```
