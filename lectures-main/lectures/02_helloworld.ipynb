 # <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Wie werden C/C++ Programme ausgeführt</strong>

- Es gibt zwei Arten wie der Quellcode von der jeweiligen Programmiersprache behandelt wird:<br><br>
- **Skriptsprache (Python, ...)**
  - Der Programmcode wird von einem Interpreter direkt ausgeführt
    - Bei jeder Programmausführung muss der Interpreter neu aufgerufen werden
  - Zuerst wird der Code in einen Bytecode übersetzt
    - In Python wird dabei aus der *.py* Datei eine *.pyc* Datei erzeugt
    - Überprüfung der Syntax bei der Übersetzung
    - Es erfolgt aber keine Überprüfung der Datentypen, etc.
  - Die Anweisungen werden dann der Reihe nach, unter Einhaltung von bestimmten Regeln, ausgeführt

<img src="images/interpreter_python.png" style="width:90%;">

- **Compilersprache (C/C++, ...)**
  - Der Quellcode wird kompiliert und dabei eine ausführbare Datei erstellt
    - Danach kann die ausführbare Datei beliebig oft (ohne Compiler) aufgerufen werden
  - Überprüfen der Syntax, Datentypen, etc. erfolgt beim Kompiliervorgang
  - Programmcode kann in diesem Fall zum Beispiel in Bezug auf Laufzeit optimiert werden

 # <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Das *Hello World* Programm</strong>


- Rufen wir uns unser *Hello World* Programm aus der LVA Wissenschaftliches Programmieren in Erinnerung
  - Die Programmiersprache war Python

 ```py
 print('Hello World!')
 ```

- Um C/C++ Programme via Jupyter Notebooks auszuführen, muss der entsprechende Kernel installiert sein
  - Wir verwenden den **jupyter-cpp-kernel**
  - Dieser kann durch folgenden Aufruf unter Linux installiert werden: `pip3 install jupyter-cpp-kernel`
  - Es gibt Kernel für Windows aber auch MacOS

- Der größte Unterschied zu Python ist nun zu Beginn, dass es in C/C++ ein Hauptprogramm gibt
  - Dieses wird durch die Funktion `main` gekennzeichnet
  - Es muss in jedem Programm genau eine solche Funktion geben

````c++
int main() {
    return 0;
}
````

- Um nun eine Zeichenkette auszugeben muss die Bibliothek `iostream` eingebunden werden
  - Es handelt sich dabei um eine Header-Datei, welche Funktionen für die Eingabe/Ausgabe in C++ bereit stellt
  - Das Einbinden einer Header-Datei erfolgt mit der Präprozessoranweisung `#include <iostream>`

````c++
#include <iostream>

int main() {
    return 0;
}
````

- Ausgabe eines Textes erfolgt dann mittels `std::cout` Befehl 

````c++
#include <iostream>

int main() {
  std::cout << "Hello World!";
  return 0;
}
````

- Die Angabe von `std` ist erforderlich, um anzugeben, dass cout zum Namensraum (engl. namespace) `std` (Standardbibliothek) gehört
- Bei `cout` handelt es sich um ein Objekt, das eine Instanz einer Klasse namens `ostream` ist
- `<<` ist der Einfügeoperator, der auch mehrmals in einer Ausgabezeile verwendet werden kann

- Operatoren, wie z.B. `<<`, sind
  - Funktionen und
  - werden auf ähnliche Weise wie herkömmliche (im Vergleich zu C) Funktionen implementiert!

````c++
#include <iostream>

int main() {
  std::cout << "Hello" << " " << "World!" << "\n";
  std::cout << "Hello World!";

  return 0;
}
````

- In Jupyter Notebooks kann der Code auch ohne explizite `main` Funktion aufgerufen werden
  - Das erledigt der Jupyter-Kernel automatisch für uns
