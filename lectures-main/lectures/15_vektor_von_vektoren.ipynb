# <strong style="color: rgb(0, 102, 153); opacity: 1.0;">2D-Felder</strong>
- In Python haben wir das NumPy `ndarray` Objekt verwendet, um ein Feld mit einer beliebigen Anzahl von Dimensionen darzustellen
- Ein 2D-Feld wurde mit `myArray = np.array([[1, 2, 3], [4, 5, 6]], np.int32)` definiert
    - Hier sind `[1, 2, 3]` und `[4, 5, 6]` Listen, welche die Zeilen des 2D-Feldes darstellen
    - `np.int32` gibt an, dass das Feld 32-Bit-Integer-Werte speichern soll
- Der Zugriff auf die Elemente erfolgt über eckige Klammern
    - Der Zugriff auf das Element mit dem Index `(0, 1)` erfolgt über `myArray[0][1]`
    - In diesem Beispiel liefert `print(myArray[0][1])` den Wert `2`
- Die C++-Standardbibliothek bietet keine direkte Implementierung eines mehrdimensionalen Feldes
    - Stattdessen kann ein Vektor von Vektoren verwendet werden
    - Es gibt noch andere Möglichkeiten, ein 2D-Feld in C++ zu erstellen, aber hier konzentrieren wir uns auf den Vektor von Vektoren

# <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Vektor von Vektoren</strong>
- Ein Vektor kann aus `double`- oder `int`-Elementen bestehen, aber auch aus Elementen eines beliebigen anderen Datentyps, z. B. eines Vektors!
- Ein Vektor von Vektoren, auch als 2D-Vektor bezeichnet, ist ein Vektor, dessen Elemente selbst Vektoren sind
- Jeder innere Vektor stellt eine Zeile oder Spalte in einer matrixartigen Struktur dar
- Dies ermöglicht die Erstellung flexibler und in der Größe veränderbarer 2D-Arrays

## <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Deklaration und Initialisierung</strong>
- Ein Vektor von Vektoren des Typs `dataType` wird mit der Anweisung `std::vector<std::vector<dataType>>` deklariert
- Elemente werden dem Vektor mit der Methode `push_back(std::vector type)` hinzugefügt
- Kleines Beispiel für die Implementierung eines Vektors von Vektoren zur Erstellung einer `3x3`-Matrix:

```c++
#include <iostream> // std::cout, std::endl
#include <vector>   // std::vector

int main() {
    // Define a 2D matrix using std::vector<std::vector<double>>
    std::vector<std::vector<double>> matrix;

    // Populate the matrix with values
    matrix.push_back({1.1, 2.2, 3.3});
```

## <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Eigenschaften eines Vektors von Vektoren</strong>
- Die Größe eines Vektors aus Vektoren kann dynamisch geändert werden, sowohl hinsichtlich der Anzahl der Zeilen als auch der Länge jeder Zeile
- Ein Vektor von Vektoren in C++ muss nicht für jede Zeile die gleiche Länge haben
- Jeder innere Vektor wird in einem zusammenhängenden Speicher abgelegt, was die Cache-Lokalität und die Zugriffseffizienz verbessert
- Der Zugriff auf die Elemente erfolgt über eckige Klammern mit der Formulierung `myVectorOfVectors[i][j]`, wobei `i` den Zeilenindex (übergeordneter Vektor) und `j` den Spaltenindex (untergeordneter Vektor) bezeichnet
- Einfaches Beispiel mit `std::vector<std::vector<int>>`:

<img align="left" width="700" src=images/vector_of_vectors.png>

```c++
#include <iostream>  // std::cout, std::endl
#include <vector>    // std::vector

int main() {
    // Declaration and initialization of a vector of vectors
    std::vector<std::vector<int>> myVec;
    
    // Adding elements to the vector of vectors
    myVec.push_back({1, 2, 3, 4, 5, 6, 7});
    myVec.push_back({8, 9, 10, 11, 12, 13, 14});
    myVec.push_back({15, 16, 17, 18, 19});
    myVec.push_back({20, 21});
    myVec.push_back({22, 23, 24});

    // Accessing elements
    std::cout << "Element at row 1, column 2: " << myVec[0][1] << std::endl;
    
    return 0;
}
```

## <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Iteration durch Elemente</strong>
Die Iteration durch einen Vektor von Vektoren kann mit verschachtelten Schleifen erfolgen:

```c++
#include <iostream>  // std::cout, std::endl
#include <vector>    // std::vector

int main() {
    // Declaration and initialization of a vector of vectors
    std::vector<std::vector<int>> myVec;
    
    // Adding elements to the vector of vectors
    myVec.push_back({1, 2, 3, 4, 5, 6, 7});
    myVec.push_back({8, 9, 10, 11, 12, 13, 14});
    myVec.push_back({15, 16, 17, 18, 19});
    myVec.push_back({20, 21});
    myVec.push_back({22, 23, 24});

    // Accessing elements
    std::cout << "Element at row 1, column 2: " << myVec[0][1] << std::endl;
#if 1
    std::cout << "****** Variant 1 ******" << std::endl;
    for (int i = 0; i < myVec.size(); ++i) {
        for (int j = 0; j < myVec[i].size(); ++j) {
            std::cout << myVec[i][j] << " ";
        }
        std::cout << std::endl;
    }
#else    
    std::cout << "****** Variant 2 ******" << std::endl; 
    for (auto row : myVec) {
        for (auto element : row) {
            std::cout << element << " ";
        }
        std::cout << std::endl;
    }
#endif
    return 0;
}
```

# <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Beispiel</strong>

```c++
#include <iostream>  // std::cout, std::endl
#include <vector>    // std::vector

void print(std::vector<std::vector<double>> grid) {
    // Iterating through rows of grid
    for (auto row : grid) {
        // Iterating through elements in each row
        for (auto element : row) {
            std::cout << element << " ";
        }
        std::cout << std::endl;
    }
}

void add_column(std::vector<std::vector<double>> &grid, std::vector<double> column) {
    for (int i = 0; i < grid.size(); ++i) {
        grid[i].push_back(column[i]);
    }
}

void remove_column(std::vector<std::vector<double>> &grid) {
    for (int i = 0; i < grid.size(); ++i) {
        grid[i].pop_back();
    }
}

int main() {
    // Declaration and initialization of a vector of vectors
    std::vector<std::vector<double>> grid;

    // Adding elements to the vector of vectors
    grid.push_back({1.1, 2.2, 3.3});
    grid.push_back({4.4, 5.5, 6.6});
    grid.push_back({7.7, 8.8, 9.9});
    std::cout << "Initial grid:" << std::endl;
    print(grid);

    // Accessing elements
    std::cout << "Element at row 1, column 2: " << grid[0][1] << std::endl;
    std::cout << "Element at row 2, column 3: " << grid.at(1).at(2) << std::endl;

    // Modifying elements
    grid[1][1] = 10.0;
    std::cout << "Modified value at index (1,1)" << std::endl;
    grid.at(2).at(0) = 22.2;
    std::cout << "Modified value at index (2,0)" << std::endl;
    print(grid);

    // Adding a row
    grid.push_back({11.1, 12.2, 13.3});
    std::cout << "Added a row:" << std::endl;
    print(grid);

    // Removing a row
    grid.pop_back();
    std::cout << "Removed a row:" << std::endl;
    print(grid);

    // Adding a column (adding an element to each row)
    add_column(grid, {14.4, 15.5, 16.6});
    std::cout << "Added a column:" << std::endl;
    print(grid);

    // Removing a column (removing the last element from each row)
    remove_column(grid);
    std::cout << "Removed a column:" << std::endl;
    print(grid);

    return 0;
}
```
