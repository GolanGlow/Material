# <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Was sind Zeiger?</strong>

- Bisher haben wir direkt auf den Inhalt einer Variablen durch Angabe des Variablennamens zugegriffen

- Sie kennen bereits Referenzen (`&`) als Funktionsargumente oder Rückgabewerte
  - Eine Referenz ist eine Art "Alias" für eine bestehende Variable und wird intern ebenfalls über Speicheradressen realisiert

- Im Unterschied dazu ist
  - ein Zeiger (*engl. pointer*) eine Variable, welche die Speicheradresse eines Objekts speichert und explizit auf das Objekt verweist
  
- Die Syntax für die Deklaration eines Zeigers lautet `type *name`:
  - `type`: der Typ des Objekts, auf das der Zeiger zeigt
  - `name`: der Name des Zeigerobjekts

- Ein Zeiger kann auf `nullptr` gesetzt werden, um anzuzeigen, dass er auf kein gültiges Objekt zeigt.  
  - `nullptr` ist typsicherer als `NULL` und sollte in modernem C++ immer verwendet werden.  
  - Zeiger sollten immer initialisiert werden.
  
- Mit dem Adressoperator `&` erhält man die Speicheradresse, unter der ein Objekt abgelegt ist
  - Man erhält somit den Zeiger auf dieses Objekt
  - Hinweis: `&` hat hier eine andere Bedeutung als bei der Deklaration einer Referenz (`int& ref = x`) 
    - dort kennzeichnet es den Referenztyp, hier liest es die Adresse einer Variable aus

- Um das Objekt zu erhalten, auf welches der Zeiger verweist, kann der Dereferenzierungsoperator `*` verwendet werden
  - Mit `*ptr` kann man auf den Inhalt der Variablen zugreifen, auf die der Zeiger zeigt
  - Hinweis: `*` hat hier eine andere Bedeutung als bei der Deklaration eines Zeigers (`int* ptr`)
    - dort kennzeichnet es den Zeigertyp, hier liest es den Wert an der gespeicherten Adresse aus

**Hinweis: Referenzen und Zeiger sind verwandt, aber nicht identisch. Referenzen sind einfacher zu verwenden und sicherer, da sie immer auf ein gültiges Objekt zeigen müssen. Zeiger hingegen bieten mehr Flexibilität (z. B. zur Arbeit mit Arrays oder dynamischem Speicher), erfordern aber auch mehr Vorsicht.**

````c++
#include <iostream>  // std::cout, std::endl

int x = 5;        // Declare and initialize a variable
int* x_ptr = nullptr;  // Declare a pointer and initialize it to nullptr

x_ptr = &x;  // Let x_ptr point to x

std::cout << "x: "     << x     << std::endl;
std::cout << "*x_ptr: " << *x_ptr << std::endl;

x = 8;  // Modify x directly
std::cout << "After x = 8:" << std::endl;
std::cout << "*x_ptr: " << *x_ptr << std::endl;

*x_ptr = 10;  // Modify x through the pointer
std::cout << "After *x_ptr = 10:" << std::endl;
std::cout << "x: "     << x     << std::endl;
std::cout << "*x_ptr: " << *x_ptr << std::endl;
````

Das folgende Beispiel verdeutlicht den Unterschied zwischen drei Konzepten, die häufig verwechselt werden:

- `&x_ptr`: die Adresse des Zeigers selbst (auch ein Zeiger ist eine Variable und hat eine eigene Adresse im Speicher)
- `x_ptr`: der Wert des Zeigers, d. h. die Adresse der Variable, auf die er zeigt
- `*x_ptr`: der dereferenzierte Wert, d. h. der Inhalt der Variable, auf die der Zeiger zeigt

````c++
#include <iostream>  // std::cout, std::endl

int x = 8;
int* x_ptr = nullptr;

std::cout << "Address of x:      " << &x     << std::endl;
std::cout << "Address of x_ptr:  " << &x_ptr << std::endl;

std::cout << "x_ptr (before assignment): " << x_ptr << std::endl;

x_ptr = &x;

std::cout << "After assigning x_ptr = &x:" << std::endl;
std::cout << "Address of x_ptr:            " << &x_ptr << std::endl;
std::cout << "x_ptr (holds address of x):  " << x_ptr  << std::endl;
std::cout << "Dereferenced x_ptr (*x_ptr): " << *x_ptr << std::endl;
````

# <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Arbeiten mit Zeiger</strong>

#### **Deklaration von Zeigern:**

````c++
int    *int_ptr    = nullptr; // pointer to a variable of type int
double *double_ptr = nullptr; // pointer to a variable of type double
char   *char_ptr   = nullptr; // pointer to a variable of type char
````

#### **Initialisierung von Zeigern:**

````c++
int  x = 5;
int *x_ptr = &x;  // x_ptr points to the address of x
````

#### **Dereferenzierung von Zeigern:**

````c++
#include <iostream>  // std::cout, std::endl

int x = 5;
int *x_ptr = &x;
std::cout << *x_ptr << std::endl; // prints the value of x via the pointer
````

#### **Ändern des Werts, auf den ein Zeiger zeigt:**

````c++
#include <iostream>  // std::cout, std::endl

int x = 5;
int *x_ptr = &x;
*x_ptr = 10;                 // modifies the value of x via the pointer
std::cout << x << std::endl; // prints 10
````

#### **Zeiger auf Objekte und Zugriff auf ihre Member:**
- Bisher haben wir auf Member eines Objekts mit dem Punktoperator `.` zugegriffen
- Wenn das Objekt jedoch über einen Zeiger angesprochen wird, verwenden wir stattdessen den Pfeiloperator `->`
  - Der Operator `->` ist eine Kurzform für `(*ptr).member`
  - Diese Langform funktioniert ebenfalls, erfordert jedoch Klammern, da der `.`-Operator eine höhere Priorität als `*` hat

````c++
#include <iostream>
#include <string>

class Person {
public:
    std::string name;
    int age;
};

int main() {
    Person person;                   // Create a Person object
    Person* person_ptr = &person;    // Pointer to the Person object

    person_ptr->name = "Alice";      // Access member using -> (preferred)
    person_ptr->age = 25;

    // Equivalent but more verbose and less readable
    (*person_ptr).age = 25;

    std::cout << person.name << " is " << person.age << " years old." << std::endl;

    return 0;
}
````

#### **`const` mit Zeigern:**
Die `const`-Schlüsselwörter können bei Zeigern auf verschiedene Weise verwendet werden, um entweder den Zeiger selbst oder das, worauf er zeigt, vor Änderungen zu schützen

| **Deklaration**         | **Bedeutung**                                           |
|-------------------------|----------------------------------------------------------|
| `const int* ptr`        | Der Inhalt, auf den gezeigt wird, ist konstant           |
| `int* const ptr`        | Der Zeiger selbst ist konstant                           |
| `const int* const ptr`  | Beides ist konstant (Inhalt und Zeiger)                  |

````c++
#include <iostream>  // std::cout, std::endl

int a = 5;
int b = 10;

const int* ptr1 = &a;
// *ptr1 = 6;         // Error: can't modify value through ptr1
ptr1 = &b;            // OK: can point to a different int

int* const ptr2 = &a;
*ptr2 = 6;            // OK: can modify value
// ptr2 = &b;         // Error: can't change where ptr2 points

const int* const ptr3 = &a;
// *ptr3 = 7;         // Error: can't modify value
// ptr3 = &b;         // Error: can't change pointer

std::cout << "a = " << a << ", b = " << b << std::endl;
std::cout << "*ptr1 = " << *ptr1 << std::endl;
std::cout << "*ptr2 = " << *ptr2 << std::endl;
std::cout << "*ptr3 = " << *ptr3 << std::endl;
````

#### **Häufige Fehler bei Zeigern**
- Uninitialisierte Zeiger verwenden: Zeiger sollten immer initialisiert werden, z. B. mit `nullptr`
- Dereferenzieren eines `nullptr`: Führt zu undefiniertem Verhalten (häufig ein Programmabsturz)
- Mehrfaches Freigeben (`delete`) desselben Zeigers: Nach dem Löschen (`delete`) sollte der Zeiger auf `nullptr` gesetzt werden *(dynamische Speicherverwaltung wird in einer späteren Vorlesung behandelt)*
- Zeiger auf lokale Variablen zurückgeben: Lokale Variablen existieren nur während der Ausführung der Funktion
    - der Zeiger wird zum sogenannten dangling pointer (hängenden Zeiger)
- Verwechslung von `*ptr` und `&var`: Das eine ist Dereferenzierung (Wert lesen), das andere ist Adressierung (Adresse lesen)

````c++
#include <iostream>  // std::cout, std::endl

int* getPtr() {
    int x = 5;
    return &x;  // x is a local variable — it is destroyed when getPtr() returns!
}

int main() {
    int* ptr = getPtr();  // ptr now points to memory that no longer belongs to us

    // Dereferencing ptr is undefined behavior: the memory may have been
    // overwritten or reused by the time we get here
    std::cout << "Value at ptr: " << *ptr << std::endl;

    return 0;
}
````

## <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Geschichte und Unterschiede: `NULL` vs. `nullptr` in C++</strong>

### <strong style="color: rgb(0, 102, 153); opacity: 1.0;">`NULL` in C / frühes C++</strong>
- In der Programmiersprache C (und frühem C++) wurde `NULL` verwendet, um einen leeren Zeiger darzustellen
- `NULL` ist ein Makro, das üblicherweise so definiert ist:
  ```cpp
  #define NULL 0
  ```
- In C++ kann `0` sowohl ein `int` als auch ein Zeiger sein. Das führt zu Mehrdeutigkeiten bei Funktionsüberladungen:
  ```cpp
  void func(int);
  void func(char*);

  func(NULL); // Ambiguität: ist das ein int oder ein Zeiger?
  ```

### <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Einführung von `nullptr` in C++11</strong>
- Mit dem Standard C++11 wurde `nullptr` eingeführt
- `nullptr` ist vom Typ `std::nullptr_t` und nur in Zeiger konvertierbar, nicht in Ganzzahlen
- Dadurch werden Mehrdeutigkeiten vermieden:
  ```cpp
  func(nullptr); // eindeutig: ruft die Version mit Zeiger auf
  ```

### <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Vergleich: `NULL` vs. `nullptr`</strong>
| Eigenschaft          | `NULL`           | `nullptr`            |
|----------------------|------------------|-----------------------|
| Typsicherheit        | Nein (Makro)     | Ja (`std::nullptr_t`) |
| Mehrdeutigkeit       | Ja               | Nein                  |
| Einführungszeitpunkt | C, C++98         | C++11                 |
| Konvertierbar zu     | `int`, `pointer` | nur `pointer`         |
