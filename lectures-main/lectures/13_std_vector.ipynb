# <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Die Standardbibliothek in C/C++</strong>

- Die Standardbibliothek in C++ ist eine Sammlung von Funktionen und Klassen, die im C++-Standard definiert sind
- Sie stellt eine Vielzahl nützlicher Funktionen bereit, die in unterschiedlichen Anwendungen eingesetzt werden können
- Wichtige Bestandteile der Standardbibliothek:
  - `iostream`: Funktionen für Ein- und Ausgabe von Daten (`std::cout`, `std::cin`, ...)
  - `string`: Funktionen und Klassen zur Verarbeitung von Zeichenketten
  - `vector`: Dynamische Arrays, die das Erstellen und Verwalten von Arrays erleichtern
  - `tuple`: Tupel zur Speicherung mehrerer Werte unterschiedlicher Typen
  - `algorithm`: Funktionen zur Datenverarbeitung, z. B. Sortieren, Suchen und Manipulieren von Containern
  - `cmath`: Mathematische Funktionen (z.B. `std::sqrt`, `std::pow`)
  - `cstdlib`: Funktionen für Zufallszahlen, Speicherverwaltung und grundlegende C-Funktionen

Zugriff auf die Standardbibliothek:
- Die Bibliotheken werden mit der entsprechenden #include-Anweisung in den Code eingebunden

```cpp
#include <vector>
#include <algorithm>
```
- Die Standardbibliothek besteht aus Header-Dateien, die zur Kompilierungszeit eingebunden werden
  - Im Gegensatz dazu gibt es Bibliotheken, die vorkompilierten Code enthalten und zur Laufzeit hinzugelinkt werden müssen (z. B. dynamische oder statische Bibliotheken)

# <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Die C++ Klasse `std::vector`</strong>

- Die Klasse `std::vector` stellt eine dynamische Array-Implementierung dar
  - Sie zählt zu den am häufigsten verwendeten Klassen der Standardbibliothek von C++

### **Was ist der `std::vector` und wie kann dieser erzeugt werden?**
- Ein `std::vector` ist ein Container, der eine Sammlung von gleichen Elementen (im Gegensatz zu einer Liste in Python) in einer bestimmten Reihenfolge speichert und verwaltet
- Ein `std::vector` kann dynamisch in der Größe verändert werden
  - Das macht den `std::vector` sehr flexibel und nützlich in vielen verschiedenen Anwendungen
- Es handelt sich beim `std::vector` um eine Templateklasse
  - Der Elementtyp wird in spitzen Klammern `<...>` angegeben

```cpp
#include <vector>
std::vector<elementTyp> var;
```

- Hier ein einfaches Beispiel zur Erzeugung eines Vektors

````cpp
#include <iostream>  
#include <vector>  
  
int main() {  
    std::vector<int> v;  // initialization of a vector with the ability to store integer values
    std::vector<double> d;  // initialization of a vector with the ability to store double values
  
    std::cout << "size of vector v = " << v.size() << std::endl;  
    std::cout << "size of vector d = " << d.size() << std::endl;  
    
    return 0;  
}  
````

**In C++ kann ein `std::vector` auch mit einer definierten Länge initialisiert werden**

Initialisieren mit einer bestimmten Anzahl von Standardwerten
  - Die einfachste Möglichkeit besteht darin, den Vektor mit einer bestimmten Anzahl von Standardwerten zu initialisieren
  - Die Syntax dafür ist:

```cpp
std::vector<typ> name(length);  
```

- Hier ist `typ` der Datentyp der Elemente im Vektor und 
- `name` der Variablenname des Vektors
- Der Parameter `length` ist die Anzahl der Elemente, die der Vektor enthalten soll

````cpp
#include <iostream>
#include <vector>

int main() {
    // Initialize vector with 10 elements, all set to 0 (default for int)
    std::vector<int> v(10); 

    std::cout << "Size of vector: " << v.size() << std::endl;

    std::cout << "Vector elements: ";
    // Print all elements using range-based for loop
    for (int num : v) {
        std::cout << num << " ";
    }
    return 0;
}
````

Initialisieren mit einem bestimmten Wert
  - Eine weitere Möglichkeit ist, einen `std::vector` zu erstellen, bei dem alle Elemente mit demselben Wert initialisiert werden
  - Alle Elemente erhalten direkt beim Erstellen denselben Initialwert
  - Die Syntax dafür lautet:
```cpp
std::vector<Typ> name(length, value);  
```
- Die Parameter:
  - `length` - Anzahl der Elemente im Vektor
  - `value` - Wert, mit dem alle Elemente initialisiert werden

````cpp
#include <iostream>
#include <vector>

int main() {
    // Erstellt einen Vektor mit 5 Elementen, alle mit Wert -21
    std::vector<int> numbers(5, -21);

    std::cout << "Vector elements: ";
    // Print all elements using range-based for loop
    for (int num : numbers) {
        std::cout << num << ", ";
    }
    return 0;
}
````

- Es ist auch möglich den Vektor mit unterschiedlichen Werten zu initialisieren

````cpp
#include <iostream>  
#include <vector>  
  
int main() {  
    std::vector<int> v = {1, 2, 3, 4, 5};
  
    std::cout << "size of vector = " << v.size() << std::endl;  
    
    std::cout << "Vector elements: ";
    // Print all elements using range-based for loop
    for (int num : v) {
        std::cout << num << ", ";
    }
    return 0;
}  
````

### **Wie kann man Elemente in einen Vektor einfügen?**

- Die einfachste Möglichkeit, Elemente in einen Vektor einzufügen, besteht darin, die Funktion `push_back()` zu verwenden 
  - Die Größe des Vektors wird hierbei automatisch vergrößert
- Die Syntax für `push_back()` lautet

```cpp
varname.push_back(newElement);  
```

- `varname` ist der Name des Vektors und `newElement` das Element, welches hinzugefügt werden soll

````cpp
#include <iostream>  
#include <vector>  
  
int main() {  
    std::vector<int> v;  // initialization of a vector of integers
  
    v.push_back(1);  // add new elements to vector
    v.push_back(2);  
    v.push_back(3);   
  
    std::cout << "size of vector = " << v.size() << std::endl;  
    
    std::cout << "Vector elements: ";
    // Print all elements using range-based for loop
    for (int num : v) {
        std::cout << num << ", ";
    }
    return 0;  
}  
````

- Sehr praktisch ist, dass sich der Vektor beim Hinzufügen automatisch vergrößert
  - Der Programmierer muss sich um die Speicherzuweisung und Speicherfreigabe nicht kümmern
  - Die gespeicherten Daten werden automatisch am Ende des Gültigkeitsblocks freigegeben


Eine weitere Möglichkeit Elemente einzufügen besteht darin, den Vektor durch Aufruf der Funktion `insert` zu erweitern
- Die Syntax für insert() lautet
```cpp
varname.insert(varname.begin() + index, newElement);  
```
- Die Parameter bedeuten folgendes
  - `varname` ist der Name des Vektors
  - `index` ist jener Index, an welchem das Element eingefügt werden soll
    - Bereits bestehende Elemente werden nach hinten gerückt
  - `newElement` ist das Element, welches eingefügt werden soll

````cpp
#include <iostream>  
#include <vector>  
  
int main() {  
    std::vector<int> v;  // initialization of a vector with integers
  
    v.insert(v.end(), 1);  // add new element at the end of a vector
    v.insert(v.end(), 2);
    v.insert(v.end(), 3);
    
    std::cout << "size of vector = " << v.size() << std::endl;  
    
    std::cout << "Vector elements: ";
    // Print all elements using range-based for loop
    for (int num : v) {
        std::cout << num << ", ";
    }
    return 0;  
}  
````

- Die Funktion `end` liefert dabei einen sogenannten Iterator auf den Vektor
  - Ein Iterator ist ein Objekt, das verwendet wird, um auf Elemente in einem Container zuzugreifen und diese zu bearbeiten
  - Ein Iterator verhält sich wie ein Zeiger auf ein Element im Container und kann verwendet werden, um dieses Element zu lesen oder zu schreiben
    - Es gibt verschiedene Arten von Iteratoren, darunter begin(), end(), rbegin(), rend(), cbegin(), cend(), crbegin() und crend().
    - Beispiele für Iteratoren
      - `begin()`: Gibt einen Iterator zurück, der auf das erste Element im Vektor zeigt
      - `end()`: Gibt einen Iterator zurück, der auf das Element nach dem letzten Element im Vektor zeigt
      - Es gibt noch weitere Iteratoren, hier sei aber auf die CPP Referenz verwiesen: https://en.cppreference.com/w/cpp/container/vector
  - Iteratoren können verwendet werden, um die Elemente im Vektor zu durchlaufen und diese zu bearbeiten.

````cpp
#include <iostream>  
#include <vector>  
  
int main() {  
    std::vector<int> v {1, 2, 3, 4, 5};  

    for (auto it = v.begin(); it != v.end(); it++) {  
        std::cout << *it << " ";  
    }  
    
    return 0;  
}  
````

### **Wie kann man auf die einzelnen Elemente eines Vektors zugreifen?**

Die einfachste Möglichkeit ist mit Hilfe von Indizes
- Dazu kann die eckige Klammer-Notation `[index]` mit dem Index des Elements `index` verwendet werden

````cpp
#include <iostream>  
#include <vector>  
  
int main() {  
    std::vector<int> v {1, 2, 3};  
    int x = v[1];
    
    std::cout << v[1] << std::endl;  
    
    return 0;  
}  
````

Eine andere Möglichkeit, auf ein Element im Vektor zuzugreifen, besteht darin, die `at()`-Funktion mit dem Index des Elements zu verwenden

````cpp
#include <iostream>  
#include <vector>  
  
int main() {  
    std::vector<int> v {1, 2, 3};  
    int x = v.at(1);
    
    std::cout << v.at(1) << std::endl;  
    
    return 0;  
}  
````

- **Hinweis**
  - Beim Zugriff mit `[...]` gibt es keine Absicherung gegen Speicherzugriffe außerhalb der reservierten Größe
    - Das kann zu einem Fehler in der Ausführung des Programms führen und stellt eine Sicherheitslücke dar
  - Abgesicherter Zugriff findet mittels `at()`
    - Hier wird eine Fehlermeldung ausgegeben, sollte der Index größer als die Länge des Vektor sein

### Vergleich: `std::vector` vs. C-Array  

|                   | `std::vector` | C-Array (`int arr[10]`) |
|-------------------|--------------|------------------|
| **Größe**         | Dynamisch     | Fest |
| **Speichersicherheit** | Sicher (`.at()`, mit Bounds-Checking) | Unsicher (keine Überprüfung auf gültige Indizes) |
| **Resize-Funktionalität** | `push_back()`, `resize()` | Nicht möglich |
| **Iteration**     | Iteratoren (`begin()`, `end()`) | Manuelles Indexieren (`arr[i]`) |

**Range basierte for Schleifen**

Mittels Range basierten `for`-Schleifen kann man über die Elemente eines Vektors iterieren

````cpp
#include <iostream>  
#include <vector>  
  
int main() {  
    std::vector<int> v {1, 2, 3};  
    
    for(auto x : v) {
        std::cout << x << std::endl;  
    }
    
    return 0;  
}  
````

**Eine Weitere Möglichkeit ist noch der Zugriff über einen Iterator**

Dies ist der analoge Ausdruck zur Range basierten `for` Schleife

````cpp
#include <iostream>  
#include <vector>  
  
int main() {  
    std::vector<int> v {1, 2, 3};  
    
    for (auto it = v.begin(); it != v.end(); it++) {  
        std::cout << *it << " ";  
    }
    
    return 0;  
}  
````

### **Wie kann man einzelne Elemente eines Vektors verändern?**

Die einfachste Möglichkeit besteht darin, die eckige Klammer-Notation `[..]` mit dem Index des Elements zu verwenden
- Die Syntax dazu ist
```cpp
std::vector<int> v {1, 2, 3};  
v[1] = 4;
```
- Hier wird das Element an Index 1 im Vektor `v` auf den Wert `4` gesetzt.

Eine andere Möglichkeit besteht darin, die `at()`-Funktion mit dem Index des Elements zu verwenden
```cpp
std::vector<int> v {1, 2, 3};  
v.at(1) = 4;
```
- Hier wird das Element an Index `1` im Vektor `v` auf den Wert `4` gesetzt
- Beachten Sie, dass `at()` eine Ausnahme auslöst, wenn der Index ungültig ist bzw. auch `[...]` einen Fehler auslösen kann

Der Vollständigkeit halber sei hier erwähnt, dass man auch auf die Elemente eines Vektors mithilfe von Iteratoren zugreifen und diese verändern kann
```cpp
std::vector<int> v {1, 2, 3};  
auto it = v.begin();  
*it = 4;
```
- In diesem Beispiel wird ein Iterator `it` erstellt 
- Dieser wird auf das erste Element des Vektors `v` gesetzt
- Der Wert des Elements, auf das der Iterator zugreift, wird dann auf `4` gesetzt
  - Da der Iterator ein Zeiger ist muss dieser dereferenziert werden um den Inhalt zu manipulieren
  - Das geschieht mit dem `*` Operator
  - Details hierzu folgen in späteren Vorlesungen

### **Wie kann man einzelne Elemente eines Vektors löschen?**

Eine Möglichkeit, ein Element im Vektor zu löschen, besteht darin, die `erase`-Funktion zu verwenden
- Dabei wird als Argument ein Iterator auf das zu löschende Element an die Funktion `erase` übergeben
```cpp
std::vector<int> v {1, 2, 3};  
v.erase(v.begin() + 1);
```
- Hier wird die `erase`-Funktion verwendet, um das Element mit Index `1` aus dem Vektor `v` zu entfernen

- Ein Beispiel, welches den Einsatz der Funktion `erase` demonstriert

````cpp
#include <iostream>
#include <vector>

int main() {
  std::vector<int> myvector;

  // set some values (from 1 to 10)
  for (int i=1; i<=10; i++) {
    myvector.push_back(i);
  }

  // erase the 6th element
  myvector.erase(myvector.begin()+5);

  // erase the first 3 elements:
  myvector.erase(myvector.begin(), myvector.begin()+3);

  std::cout << "myvector contains:";
  for (unsigned i=0; i<myvector.size(); ++i) {
    std::cout << ' ' << myvector[i];
  }
  std::cout << '\n';

  return 0;
}
````


Mehrere Elemente können auch auf einmal aus einem Vektor gelöscht werden, indem die Funktion `erase` mit einem Iteratorbereich aufgerufen wird
```cpp
std::vector<int> v {1, 2, 3, 4, 5};  
auto it1 = v.begin() + 1;  
auto it2 = v.begin() + 3;  
v.erase(it1, it2);
```
- Hier werden zwei Iteratoren `it1` und `it2` erstellt
- Diese verweisen auf das zweite bzw. vierte Element im Vektor `v`
- Die `erase`-Funktion wird anschließend verwendet, um alle Elemente im Bereich von `it1` bis ausschließlich `it2` aus dem Vektor zu entfernen


Beachten Sie, dass man auch alle Elemente eines Vektors auf einmal löschen kann
  - Dazu kann die Funktion `clear` verwendet werden
```cpp
std::vector<int> v {1, 2, 3};  
v.clear();
```

# <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Ein zusammenfassendes Beispiel für `std::vector`</strong>

````cpp
#include <iostream>  
#include <vector>  
  
int main() {  
    std::vector<int> v;  // initialization of a vector with integers
  
    v.push_back(1);  // append new elements
    v.push_back(2);  
    v.push_back(3);  
    v.push_back(4);  
    v.push_back(5);  
  
    // print size of vector
    std::cout << "size of vector = " << v.size() << std::endl;  
  
    // print elements of vector
    std::cout << "elements of vector: ";  
    for (int i = 0; i < v.size(); i++) {  
        std::cout << v[i] << " ";  
    }  
    std::cout << std::endl;  
  
    // modify element at index 2
    v[2] = 7;  
  
    // print elements of vector
    std::cout << "elements of vector: ";
    for (int i = 0; i < v.size(); i++) {  
        std::cout << v[i] << " ";  
    }  
    std::cout << std::endl;  
  
    // clear vector
    v.clear();
    std::cout << "size of vector = " << v.size() << std::endl;  

    return 0;  
}  
````
