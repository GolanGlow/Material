 # <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Anforderungen an Variablen und Konstanten</strong>


**Name**
- Variablen und Konstanten müssen mit eindeutigen Bezeichnern versehen werden.  

**Ort (Speicherplatz)**
- Sie müssen gespeichert werden und haben somit einen Ort
- Alle Variablen und Konstanten haben ihre eigene Speicheradresse

**Datentyp**
- Bestimmt, welche Daten die Variablen speichern können
- Zu einem Datentyp gehören auch die auf ihm verwendeten Operatoren und Methoden

**Wert**
- Eine Variable hat zu jeder Zeit einen Wert, auch wenn dieser ungültig ist

**Gültigkeitszeitraum**
- Eine Variable muss vor Gebrauch angefordert bzw. erschaffen (definiert/deklariert) werden
- Sie kann solange verwendet werden, bis sie zerstört wird, zum Beispiel durch Beendigung des Programms

**Sichtbarkeitsbereich**
- Existierende Variablen können in manchen Programmteilen unsichtbar sein

- Der Datentyp einer Variablen wird bei der Zuweisung eines Wertes festgelegt

 # <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Welche Arten von Variablen gibt es?</strong>

**Definition von Variablen**
 - Datentyp, Name und Speicherklasse festlegen

```cpp
long value;
double result;
long diff, sum;
```

**Globale Variablen**
- Können von überall im Programm angesprochen werden
- Sollten aber nur im absoluten Notfall verwendet werden, da sie die Wartbarkeit des Codes verringern

**Lokale Variablen**
- Werden innerhalb einer Funktion (oder eines Blocks) deklariert, in welcher sie angesprochen werden
- Müssen in C (C89) am Anfang einer Funktion definiert werden
- Lokale Variablen sind nur innerhalb der Funktion oder ihres Blocks sichtbar


````c++
// Example: Global vs. Local Variables
#include <iostream>

int globalvar = 1; // Global variable

int main() { 
    int localvar = 2; // Local variable

    std::cout << "localvar = " << localvar << std::endl;
    std::cout << "globalvar = " << globalvar << std::endl;

    return 0;
}
````

 # <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Namensgebung von Variablen</strong>

- Folge von Buchstaben und Ziffern (der ANSI-Standard erkennt eine Länge von 31 Zeichen an)
- Müssen mit Buchstaben oder mit einem Unterstrich „_“ beginnen
- Umlaute sind nicht erlaubt
- Unterscheidung von Groß- und Kleinschreibung (case sensitive): `name` != `Name` != `NAME`

**Empfehlungen**
- Begrenzung der Länge des Namens auf deutlich unter 31 Zeichen
- Verwendung aussagekräftiger Namen für die Variablen
- Beginn eines Variablennamens mit Kleinbuchstaben
- Trennung von „Wörtern“ durch gemischte Groß- und Kleinschreibung (z.B. camelCase) oder Unterstriche (z.B. snake_case)
- Verwendung ausschließlich von Großbuchstaben für symbolische Konstanten (Präprozessorkonstanten)
  - Zum Beispiel `#define LENGTH 100` oder `#define PI 3.14`
****

 # <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Initialisierung von Variablen</strong>

- Es wird empfohlen Variablen bei deren Definition zu initialisieren

````c++
long value = 12;
double result = 12.45;
````

- Nicht initialisierte Variablen haben einen „undefinierten“ Wert!
- Ein Beispiel zur uninitialisierten Verwendung von Variablen
  - Der Wert der Variablen kann zufälligerweise und compilerabhängig beim Anlegen auch 0 sein
  - Es gibt keine Fehlermeldung, dass eine Variable uninitialisiert verwendet wurde
  - Eventuell kann der Compiler eine Warnung ausgeben, dass Variablen uninitialisiert verwendet wurden

````c++
#include <iostream>

int main() {
  double value;
  double result;
  
  result = 10.0 * value; // Using uninitialized value
  
  std::cout << "Result = " << result << std::endl;

  return 0;
}
````

**Zuweisung an eine Variable**
- Bei einer Zuweisung wird einer Variablen in C/C++ ein neuer Wert zugewiesen
  - Der Wert auf der rechten Seite des Gleichheitszeichens wird in die Variable auf der linken Seite kopiert
- In Python unterscheiden wir zwischen veränderbaren (***mutable***) und nicht veränderbaren (***immutable***) Datentypen

 # <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Konstanten</strong>

- Der Wert einer Konstanten kann nicht verändert werden
- Konstanten können in speziellen Speicherbereichen (ROM) abgelegt werden
- Sie müssen initialisiert werden
- Für deren Namen gelten die selben Bestimmungen wie Namen von Variablen
- Sie werden durch Verwenden des Schlüsselwortes `const` definiert

````c++
#include <iostream>

const double PI = 3.14159265358979323846; // Correct usage

int main() {
    PI = 3.14; // ERROR: Cannot modify a const variable

    std::cout << "pi = " << PI << std::endl;

    return 0;
}
````

 # <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Numerische Datentypen</strong>

**Was ist ein Datentyp?**
- Eine Menge von Daten inklusive aller Operatoren und Funktionen, die auf dieser Menge definiert sind

**Mathematische Datentypen (z.B. die Menge aller ganzen Zahlen)**
- Enthalten unendlich viele Elemente

**Reale Datentypen**
- Aufgrund der begrenzten Speicherkapazität müssen die mathematischen Datentypen begrenzt werden

### **Ganze Zahlen**

- Datentyp `int` (Abkürzung des Wortes integer)
- Qualifizierer
  - Anzahl der verwendeten Bits: `short`/`long` (architekturabhängig!)
```cpp
short int value; // 16 Bit
long int value;  // 32 Bit
```
  - Positive und negative (signed) oder nur positive Werte (unsigned)


<img align="left" width="1000" src="images/ganze_zahlen.png">

- **Grundlegende Eigenschaften von Zahlensystemen**  
  - Ein Zahlensystem mit der Basis `n` hat `n` Symbole zur Darstellung von Zahlen.  
  - Mit `m` Stellen lassen sich `n^m` Zahlen darstellen (`0 ... n^m - 1`).  

- **Beispiel: Dezimalsystem**  
  - Das Dezimalsystem (`10er-Zahlensystem`) hat 10 Symbole zur Darstellung von Zahlen (`0-9`).  
  - Mit 3 Stellen lassen sich `10^3 = 1000` Zahlen darstellen (`0 ... 999`).  
  
**Reichen die vorhandenen Stellen zur Darstellung der Zahl nicht aus, so kann die Zahl nicht dargestellt werden**

<img align="left" width="1000" src="images/ganze_zahlen_speicherverbrauch.png">


- **Operatoren für ganze Zahlen**


<img align="left" width="300" src="images/ganze_zahlen_operatoren.png">


- Das Ergebnis der Ganzzahlendivision / ist wieder eine ganze Zahl.
- Der Modulo-Operator % liefert den Rest der Ganzzahlendivision.

````c++
#include <iostream>

int main() {
    long num1 = 10;
    long num2 = 3;

    long div = num1 / num2; // Integer division (result = 3)
    long mod = num1 % num2; // Modulus (result = 1)
    double div_float = static_cast<double>(num1) / num2;

    std::cout << "Integer Division: " << div << std::endl;
    std::cout << "Modulus: " << mod << std::endl;
    std::cout << "Floating-Point Division: " << div_float << std::endl;

    return 0;
}
````

### **Gleitpunkt-Zahlensysteme**

 # <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Wahrheitswerte/Boolesche Zahlen</strong>

- In C++ (seit C++98) gibt es einen eigenen Datentyp für Wahrheitswerte: `bool`.  
- Es gibt zwei Literale für den Datentyp `bool`:  
  - `true` (intern als `1` dargestellt) und `false` (intern als `0` dargestellt).  
  - Diese werden auch als Wahrheitsliterale bezeichnet.  

````c++
#include <iostream>

int main() {
    bool status = true;  // Stored as 1
    std::cout << "Status (true) = " << status << std::endl;

    status = false;  // Stored as 0
    std::cout << "Status (false) = " << status << std::endl;

    return 0;
}
````

 # <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Zeichen</strong>

- Zeichen werden in Form von sogenannten ASCII-Codes gespeichert.  
- Datentyp: `char`  
- Speicherverbrauch: 8 Bit (1 Byte)  
- Wertebereich:  
  - `char`: `-128 ... 127` (Vorzeichenbehaftet, *signed char*)  
  - `unsigned char`: `0 ... 255` (Vorzeichenlos)  

````c++
#include <iostream>

int main() {
  char c = 'T'; // ok
  //char c = 'Test'; // truncates, compiler dependent (multicharacter literal)!
  //char c = "Test"; // compile error

  std::cout << c << std::endl; 

  return 0;
}
````

 # <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Ein Beispiel zum Arbeiten mit Variablen und Konstanten</strong>

````c++
#include <iostream>

// Definition of a mathematical constant
const double PI = 3.14159265358979323846;

int main() {
    // Initialize the radius
    double radius = 10.0;

    // Calculate areas
    double areaHalf = (PI * radius * radius) / 2.0;   // Area of half a circle
    double areaQuarter = areaHalf / 2.0;              // Area of a quarter circle

    // Output results
    std::cout << "Area of half of a circle: " << areaHalf << std::endl;
    std::cout << "Area of a quarter of a circle: " << areaQuarter << std::endl;

    return 0;
}
````

- Im Vergleich dazu das Beispiel in Python3
```python

radius = 10 # initialization
areaHalf = radius * radius * pi / 2 # do calculation
areaQuarter = areaHalf / 2 # do calculation

print("Area of half of a circle = ", areaHalf)
print("Area of a quarter of a circle = ", areaQuarter)
```
