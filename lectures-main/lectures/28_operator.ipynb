## Operator Overloading

In C++ ist es möglich, diverse Operatoren selbst zu definieren. Im Detail bedeutet dies, beispielsweise für eine eigene 
Klasse die mathematischen Operationen `+`,`-`,`*`,`/` selbst zu definieren, sofern dies sinnvoll erscheint. Dies ist 
nicht nur auf eine "mathematische" Addition beschränkt, wie schon bei Stringklassen ersichtlich, 
bei denen ein `+` ein Aneinanderfügen der Teilstrings bewirkt. Auch sind nicht immer alle Operatoren sinnvoll 
erweiterbar. Nicht veränderbar ist die Hierarchie der Operatoren. `(` hat immer höhere Priorität als `*` und dies 
höhere Priorität als `+`. 

Um in C++ die Operation `op1 @ op2` neu zu definieren, wird der eigentlich Operator `@` überladen:

```C++
return_type operator@(parameter_list)
``` 

definiert den Operator `@` neu, der die Parameter der `parameter_list` übergeben bekommt und den `return_type` zurückliefert.

Wie ist dies beispielsweise bei der Operation `op1 + op2` zu verstehen?
* Intern wird für diese Operation eine Funktion aufgerufen.
* Hierbei ist jedenfalls ein `operator+` zu überladen.
* Der Operator bekommt zwei Argumente, `op1` und `op2`(wobei eines links vom Operator steht und eines rechts), und
* liefert auch ein Ergebnis zurück.
* Somit ist folgende Funktion zu implementieren:

```C++
result_type operator+(op1_type op1, op2_type op2);
```

Wobei noch zu überlegen ist, ob Parameter `const` oder per Referenz übergeben werden.

Anstelle von `@` sind folgende Ausdrücke gültig:

    <tr><td>Art der Operatoren</td><td></td><td>Beispiel</td></tr>

* Die `parameter_list` kann hierbei, je nach Operation, bis zu 2 Argumente besitzen. 
* Wird ein Operator einer Klasse überladen, so ist implizit `*this` der erste Parameter. 
* Die Operatoren der intrinsischen Datentypen (wie `int`, `double`, `char`) können nicht überladen oder verändert werden.
* Es können keine zusätzlichen Operatoren definiert werden

### Operatordefinition außerhalb der Klasse

Beispielsweise kann die Addition der `Vec2d` Klasse außerhalb der Klasse überladen werden:

````c++
class Vec2d
{
public:
    double x,y;
    Vec2d(const double &x_=0, const double &y_=0) : x(x_), y(y_) {}
};

Vec2d operator+(const Vec2d &op1, const Vec2d &op2) {
    Vec2d result;
    result.x = op1.x + op2.x;
    result.y = op1.y + op2.y;
    return result;
}

int main()
{
    Vec2d a(2,2), b(3,3), c;
    
    c  =  a + b;
    // |  \---/
    // |   operator+
    // operator=, is created implicitely
    
    std::cout << a.x << " " << a.y << std::endl;
    std::cout << b.x << " " << b.y << std::endl;
    std::cout << c.x << " " << c.y << std::endl;
}
````

Wie ist dieser Ausdruck zu lesen?
* Der `+`-Operator wird außerhalb einer Klasse definiert. 
* Es handelt sich um den Operator mit 2 Parametern (Operanden), somit `op1 + op2`
* Die beiden Operanden sind vom Typ `const Vec2d&` und `const Vec2d&` (denkbar wären auch andere Typen)
* Die Parameter werden als `const` übergeben, da sie nicht verändert werden. Ohne dieser `const` Definition dürften keine `const` Variablen übergeben werden.
* Rückgabewert ist ebenfalls vom Typ `Vec2d`
* Im Funktionsrumpf werden die Komponenten der Parameter addiert und eine Kopie des Ergebnisses zurückgeliefert.

In diesem Beispiel ist ebenfalls zu beachten, dass innerhalb von `operator+` auf die Member von `Vec2d` zugegriffen wird (`op1.x`, `result.x`,...). Dies ist nur möglich, wenn die verwendeten Member `public` sind oder die Funktion `operator+` als `friend` in `Vec2d` definiert ist. Andernfalls müsste `Vec2d` explizite Setter- und Getterfunktionen bereitstellen.

### Operatordefinition innerhalb der Klasse

Analog könnte diese Funktionalität auch innerhalb der Klassendefinition deklariert werden (entweder oder):

````c++
class Vec2d
{
public:
    double x,y;
    Vec2d(const double &x_=0, const double &y_=0) : x(x_), y(y_) {}

    Vec2d operator+(const Vec2d &op2) const;
};

Vec2d Vec2d::operator+(const Vec2d &op2) const {
    Vec2d result;
    result.x = this->x + op2.x;
    result.y = this->y + op2.y;
    return result;
}

int main()
{
    Vec2d a(2,2), b(3,3), c;
    
    c  =  a + b;
    // |  \---/
    // |   operator+
    // operator=, is created implicitely
    
    std::cout << a.x << " " << a.y << std::endl;
    std::cout << b.x << " " << b.y << std::endl;
    std::cout << c.x << " " << c.y << std::endl;
}
````

* Hierbei ist `*this` implizit der erste Parameter, `op2` der zweite. Es wird also dieselbe Funktionalität wie vorher definiert.
* Die Funktion selbst ist `const`, da sie keine Member der Klasse verändert. Ohne die `const` Definition dürfte `+` nicht auf eine `const` Variable angewendet werden.
* Ist die Klassendefinition nicht veränderbar (beispielsweise Bibliotheken), so sind nur Operatoren außerhalb der Klasse möglich

Einige Operatoren benötigen nur einen Parameter. Dies ist beispielsweise das Negieren eines logischen Ausdrucks `!expression`, das Negieren eines Bitmusters `~expression` oder die Vorzeichenumkehr `-expression`.

```c++
Vec2d Vec2d::operator-() const {          // *this is implicitely the first argument
    return Vec2d(-this->x, -this->y);
}
```

### Die Operatoren << und >>

Der Linksshift- und Rechtsshift-Operator bewirkt bei Basistypen ein Verschieben der Bitfolge. `x << i` bzw. `x >> i` verschieben die Bitfolge um `i` Stellen nach links bzw. rechts.

Die Hauptanwendung dieser Operatoren in C++ ist, in Verbindung mit Datenströmen (I/O Streams, Stringstreams und dergleichen), ein Verschieben von Daten aus Variablen in einen Stream bzw. aus einem Stream in Variablen.

Beispielsweise:

````c++
double x = 17;
std::cout << "Die Variable x ist " << x << std::endl;  // print x on cout
std::cin >> x;                                         // read from cin to x

std::cout << "Nun ist die Variable x gleich " << x << std::endl;
````

Um gegebenenfalls für eigene Klassen ebenfalls eine Ein- und Ausgabe zu implementieren, können diese beiden 
Operatoren überladen werden.

* Beide Funktionen erhalten als ersten Parameter den Stream 
* Als weiterer Parameter ist die Variable, die ausgegeben werden soll (vorzugsweise `const`), bzw. die Variable, die eingelesen werden soll (als Referenz) zu übergeben.
* Rückgabewert ist wieder der Stream (vorzugsweise als Referenz), um die Operatoren in Folge aufrufen zu können (`std::cout << x << y << z`)

Beispielsweise:

````c++
class Vec2d
{
    friend std::ostream &operator<<(std::ostream &o, const Vec2d &vec);
    friend std::istream &operator>>(std::istream &i, Vec2d &vec);

    double x,y;
public:
    Vec2d(double x_ = 0, double y_ = 0) : x(x_), y(y_) {}
};

std::ostream &operator<<(std::ostream &o, const Vec2d &vec) {
    o << "(" << vec.x << "," << vec.y << ")";
    return o;
}
std::istream &operator>>(std::istream &i, Vec2d &vec) {
    i >> vec.x >> vec.y;
    return i;
}

int main() 
{
    Vec2d v1,v2;
    
    std::cin >> v1 >> v2;
    std::cout << "v1 ist " << v1 << " und v2 ist " << v2 << std::endl;
}
````

Hierbei ist zu beachten, dass 
* diese Operatoren als freie Funktion definiert sind, da `std::ostream` in der Standardbibliothek nicht verändert werden kann.
* Hierzu müssen allerdings die Komponenten `vec.x` und `vec.y` ansprechbar sein. 
* Diese Komponenten müssen demzufolge `public:` verfügbar sein bzw. `Vec2d` muss diese Operatoren als `friend` definieren.

### Die Operatoren [] und ()

#### `return_type class_name::operator[](one_integer_type)`

Mit `[]` werden in C und C++ Elemente eines Feldes angesprochen. Mit diesem Operator kann der Aufruf `[]` ebenfalls bei neu definierten Klassen ermöglicht werden (intern wird davon beispielsweise bei der Klasse `std::vector` Gebrauch gemacht).
* Die Definition darf nur innerhalb der Klasse erfolgen und 
* Der einzige Parameter darf nur ein ganzzahliger Typ sein. 

Verdeutlicht wird dies am Beispiel eines in einer Klasse eingebetteten Feldes:

````c++
class MyArray
{
    std::vector<int> elements;
public:
    MyArray(int n=10) : elements(n) {}
    
    int &operator[](const int i) {
        return elements[i];
    }
    int operator[](const int i) const {
        return elements[i];
    }
};

int main()
{
    MyArray my(10);
    
    my[0] = 17;
    my[1] = my[0];
    
    std::cout << "my[0]=" << my[0] << std::endl;
    std::cout << "my[1]=" << my[1] << std::endl;
}
````

Hierbei sind zwei unterschiedliche Funktionen möglich, mit und ohne `const`.
* Der Compiler ruft automatisch die richtige Funktion auf.
* Die non-const Version auf der linken Seite der Zuweisung, diese muss ja veränderbar sein.
* Die const Version auf der rechten Seite der Zuweisung.

#### `return_type class_name::operator()(parameter_list)`

Ähnlich verhält es sich mit dem `()`-Operator, welcher ebenfalls nur innerhalb einer Klasse definiert werden kann, allerdings mit beliebigen Parametern (auch keine) definiert werden kann.

````c++
class MyClass
{
public:
     void operator()() {}                               // define operator() for a function which takes no parameters
     void operator()(const double x, const double y) {} // define operator() for a function which takes two doubles
};

int main()
{
    MyClass x;    // we define an Object of MyClass
    
    // now we call operator()
    x();          // looks like a Function x(), internally operator() of the object x is called
    x(2.0,3.0);
}
````

* Dieser Operator sieht wie ein Funktionsaufruf aus
* Ist aber eine Funktion, die auf das Objekt wirkt
* Das Objekt enthält Variablen und kann somit Zustände speichern

````c++
class Adder 
{
    double sum = 0;
public:
    double operator()(const double x) {
        sum += x;
        return sum;
    }
    double get_sum() {
        return sum;
    }
};

int main() {
    Adder adder;
    
    std::cout << adder.get_sum() << std::endl;
    adder(5);
    std::cout << adder.get_sum() << std::endl;
    
    std::cout << adder(2) << std::endl;
    std::cout << adder(3) << std::endl;    
}
````

### Die Operatoren `++` und `--`

In C++ können auch die Inkrement- und Dekrementoperatoren überladen werden, wenn dies sinnvoll erscheint. Zu beachten ist hierbei dass es jeweils eine Prefix- und Postfixform gibt:

* Prefix `++i`  erhöht die Variable sofort und gibt auch den neuen Wert zurück, für diese Operation ist `operator++()` verantwortlich
* Prefix `--i`  erniedrigt die Variable sofort und gibt auch den neuen Wert zurück, für diese Operation ist `operator--()` verantwortlich
* Postfix `i++` gibt i zurück und erhöht die Variable sozusagen danach, hierfür ist `operator++(int)` verantwortlich.
* Postfix `i--` gibt i zurück und erniedrigt die Variable sozusagen danach, hierfür ist `operator--(int)` verantwortlich.

Da die Prefix Notation zumeist performanter ist, wird vorzugsweise diese verwendet.

Hierzu ein Beispiel mit der schon bekannten `Vec2d`-Klasse

````c++
struct Vec2d
{
    double x,y;
    Vec2d(const double x_, const double y_) : x(x_), y(y_) {}
    Vec2d operator++() {
        ++x; ++y;          // this function increments both, x and y
        return *this;      // and delivers the new value
    }
    Vec2d operator++(int) {
        Vec2d tmp(*this);  // we have to store the actual Vec2d
        ++x; ++y;          // increment x and y
        return tmp;        // but return the old Vec2d, less performant as we need a temporary Vec2d
    }
};
std::ostream &operator<<(std::ostream &o, const Vec2d &v) {
    o << v.x << "," << v.y;
    return o;
}

int main()
{
    Vec2d v1(1,2);
    
    std::cout << v1 << ", ok" << std::endl;
    v1++;    // we increment v1
    std::cout << v1 << ", clear v1 ist incremented" << std::endl;
    
    std::cout << v1++ << ", we return old v1 and increment afterwards" << std::endl;
    std::cout << v1 << ", but now v1 is the incremented value" << std::endl;
    
    std::cout << ++v1 << ", now we increment and return the new value" << std::endl;
    std::cout << v1 << ", clear this is the new value" << std::endl;  
}
````
