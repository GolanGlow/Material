# <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Anwendung von Strukturen</strong>
Da Strukturen einen Datentyp definieren, 
- können diese in einem Array/Vector und 
- als Daten innerhalb anderer Strukturen verwendet werden

## <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Beispiel</strong>
Ein Beispiel für die Verwendung der Struktur `Address` innerhalb der Struktur `Person`:

```c++
#include <iostream> // std::cout, std::endl
#include <string>   // std::string

// Define a struct representing an address:
struct Address {
    std::string country;
    std::string city;
    int postal_code;
    std::string street;
    int house_number;
};

// Define a struct representing a person: (name, age, height, is_student, address)
struct Person {
    std::string name;
    int age;
    double height;
    bool is_student;
    Address myAddress;
};

int main() {
    // Declare and initialize a person
    // Use nested aggregate initialization
    Person myPerson = {"Alice", 30, 1.65, true, {"Austria", "Vienna", 1040, "Karlsgasse", 25}};

    std::string study;
    if (myPerson.is_student) {
        study = "student";
    } else {
        study = "non-student";
    }

    // Access elements of "Address" using the dot operator
    std::cout << myPerson.name << " is a "
              << myPerson.age << "-year old "
              << study << " who is "
              << myPerson.height << "m tall and lives at "
              << myPerson.myAddress.street << " " 
              << myPerson.myAddress.house_number << ", "
              << myPerson.myAddress.postal_code << " "
              << myPerson.myAddress.city << ", "
              << myPerson.myAddress.country;
    
    return 0;
}
```

## <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Beispiel</strong>
Ein Beispiel für die Verwendung eines `std::vector` von `struct`-Objekten
- hier ein Vektor von `Person`-Objekten:

```c++
#include <iostream> // std::cout, std::endl
#include <string>   // std::string
#include <vector>   // std::vector

// Define a struct representing a person:
struct Person {
    std::string name;
    int age;
    bool is_student;
};

int main() {
    // Declare and initialize a vector of Person structs
    std::vector<Person> people = {
        {"Alice", 30, true},
        {"Bob",   25, false},
        {"Carol", 22, true}
    };

    // Iterate over the vector using a range-based for loop
    for (const auto& person : people) {
        std::string study = person.is_student ? "student" : "non-student";
        std::cout << person.name << " is a "
                  << person.age << "-year old "
                  << study << std::endl;
    }

    return 0;
}
```

## <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Beispiel</strong>
Unten ein Beispiel für die Verwendung eines Vektors der `struct` `Grade` innerhalb der Struktur `Person`

```c++
#include <iostream> // std::cout, std::endl
#include <string>   // std::string
#include <vector>   // std::vector

// Define a struct representing a course grade:
struct Grade {
    std::string course;
    int grade;
};

// Define a struct representing a person: (name, age, height, is_student, grades)
struct Person {
    std::string name;
    int age;
    double height;
    bool is_student;
    std::vector<Grade> grades;
};

int main() {
    // Declare and initialize a person
    // Use aggregate initialization
    Person myPerson = {"Alice", 30, 1.65, true};

    // Add grade for "Math" by first creating a Grade object
    Grade myMathGrade = {"Math", 1};
    myPerson.grades.push_back(myMathGrade);

    // Add grade for "History" without creating a Grade object
    myPerson.grades.push_back({"History", 2});

    // Print grades
    std::cout << "Here are the grades for " << myPerson.name << ":" << std::endl;
    for (int i = 0; i < myPerson.grades.size(); ++i) {
        std::cout << myPerson.grades[i].course << ": " << myPerson.grades[i].grade << std::endl;
    }

    return 0;
}
```
