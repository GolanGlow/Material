# <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Was ist Vererbung?</strong>
- Oftmals soll eine neue Klasse Attribute einer bestehenden Klasse erhalten
- Vererbung ermöglicht die Erstellung einer neuen Klasse basierend auf einer bestehenden Klasse
- Das ist insbesondere dann von Interesse, wenn Verschiedene Datentypen einen gemeinsamen Kern besitzen
    - Der Kern ist eine so genannte Basisklasse (*engl. base class*)
    - Eine Basisklasse kann auch **abstrakt** sein (*engl. abstract base class*)
- Eine abstrakte Basisklasse ist eine bestimmte Basisklasse, die ausdrücklich nur zur Verwendung als Basisklasse vorgesehen ist
    - Sie ist nicht dazu bestimmt, instanziiert zu werden
    - Sie kann nicht verwendet werden, um Objekte zu erzeugen
    - Sie enthält mindestens eine rein virtuelle Funktion (abstrakte Memberfunktion oder **Interface**)

<img align="left" width="700" src=images/inheritance.png>

```c++
#include <iostream> // std::cout, std::endl

class Vehicle {
protected:
    int    wheels;
    double weight;
public:
    Vehicle(int wheels_, double weight_) : wheels(wheels_), weight(weight_) {}
    int    get_wheels() const { return wheels; }
    double get_weight() const { return weight; }
    double wheel_load() const { return weight / wheels; }
    const char* name() const { return "Vehicle"; }
    void print() const;
};

int main() {
    Vehicle Car(4, 800.0);
    Vehicle Motorcycle(2, 350.0);
    Vehicle Truck(18, 10000.0);
    Car.print();
    Motorcycle.print();
    Truck.print();

    return 0;
}

void Vehicle::print() const {
    std::cout << name() << " has wheel load of " << wheel_load() << std::endl;
}
```

- Im obigen Code sind die Variablen `wheels` und `weight` als `protected` gekennzeichnet
- `protected` spezifiziert den Zugriff auf die Variable analog zu `private`, mit der Erweiterung, dass `protected` Variablen und Funktionen in abgeleiteten Klassen verfügbar sind
- Die Memberzugriffssteuerung entscheidet, welche Elemente der Basisklasse in die abgeleitete Klasse übernommen werden

## <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Einfache Vererbung</strong>
- Bei der einfachen Vererbung haben Klassen nur eine Basisklasse
    - Obwohl Klassen mehrere indirekte Basisklassen haben können

### <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Memberzugriffssteuerung</strong>

- Die Basisklasse `Baseclass` gibt ihre Eigenschaften an die neu definierte abgeleitete Klasse `DerivedClass` weiter
- Zu den bisher in der Basisklasse definierten Eigenschaften können zusätzliche Membervariablen und -funktionen speziell für die abgeleitete Klasse im Deklarationsblock definiert werden
- Der optionale `access_specifier` gibt die Art der Vererbung an und kann hierbei `public`, `protected` oder `private` annehmen
    - Default ist `private`
    - Abhängig von dem `access_specifier` werden nur bestimmte Funktionalitäten an die abgeleitete Klasse weitergegeben

**Private Member** der Basisklasse werden nicht weitervererbt und sind in der abgeleiteten Klasse nicht sichtbar!

**Public** -- `public` bzw. `protected` Member der Basisklasse werden ebenfalls `public` Member bzw. `protected` Member der abgeleiteten Klasse

**Protected** -- `public` bzw. `protected` Member der Basisklasse werden `protected` Member der abgeleiteten Klasse

**Private** -- `public` bzw. `protected` Member der Basisklasse werden `private` Member der abgeleiteten Klasse

<img align="left" width="450" src=images/access.png>

```c++
#include <iostream> // std::cout, std::endl

class Vehicle {
protected:
    int    wheels;
    double weight;
public:
    Vehicle(int wheels_, double weight_) : wheels(wheels_), weight(weight_) {}
    int    get_wheels() const { return wheels; }
    double get_weight() const { return weight; }
    double wheel_load() const { return weight / wheels; }
    const char* name() const { return "Vehicle"; }
    void print() const;
};

class Car : public Vehicle {   // public inheritance
  int seats;

public:
  Car(int wheels_, double weight_, int passengers = 4) : Vehicle(wheels_, weight_), seats(passengers) {}

  int get_seats() const { return seats; }
  const char* name () const { return "Car"; }
  void print() const;
};


int main()
{
  Vehicle Automobile(4, 800.0);
  Vehicle Motorcycle(2, 350.0);
  Vehicle Truck(18, 25000.0);
  
  Car Audi(4, 850.0, 5);
  std::cout << "The Audi has " << Audi.get_wheels() << " wheels!" << std::endl;
  Audi.print();
  
  return 0;
}

void Vehicle::print() const {
  std::cout << name() << " has a wheel load of " << wheel_load() << std::endl;
}

void Car::print() const {
  std::cout << name() << " has a wheel load of " << wheel_load() << std::endl;
}
```

#### <strong style="color: rgb(0, 102, 153); opacity: 1.0;">`protected` und `private` Vererbung</strong>

- Was würde im obigen Beispiel passieren, wenn wir `protected` oder `private` statt `public` vererben würden?
    - `class Car : protected Vehicle`, bzw.
    - `class Car : private Vehicle`

- Wenn die Vererbung `protected` (`private`) ist, hat die Klasse `Car` geschützten (keinen) Zugriff auf `get_wheels()`, was bedeutet, dass ein `Car`-Objekt keinen Zugriff darauf hat
    - `Audi.get_wheels()` ist nicht erlaubt!

#### <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Zugriff auf Funktionen der Basisklasse</strong>

- Wir können jedoch die Funktion `get_wheels()` in der abgeleiteten Klasse neu implementieren, um auf geschützte Variablen der Basisklasse zuzugreifen
- In der abgeleiteten Klasse kann man mittels `baseClassName::functionName()` eine Funktion aus der Basisklasse aufrufen
    - Wir können `get_wheels()` in `Car` implementieren, das `private` von `Vehicle` erbt
    - In `Car::get_wheels()` können wir dann `Vehicle::get_wheels()` aufrufen

## <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Arten der Vererbung</strong>
- Neben der Einfachvererbung gibt es mehrere Arten der Vererbung:

    - Hierarchische Vererbung (*engl. hierarchical inheritance*)
    - Mehrfachvererbung (*engl. multiple inheritance*)
    - Hybride Vererbung (*engl. hybrid inheritance*)

### <strong style="color: rgb(0, 102, 153); opacity: 1.0;">Mehrfachvererbung</strong>

- Mehrfachvererbung ermöglicht es einer abgeleiteten Klasse, von mehreren Basisklassen zu erben

```c++
#include <iostream> // std::cout, std::endl

class X {
protected:
  int x;
public:
  X(int x_ = 0) : x(x_) {}
  int get_x() const { 
    return x; 
  }
};

class Y {
protected:
  int y;
public:
  Y(int y_ = 0) : y(y_) {}
  int get_y() const { 
    return y; 
  }
};

class Sum : public X, public Y {
private:
  int sum;
public:
  Sum(int x_ = 0, int y_ = 0) : X(x_), Y(y_) {}  // delegate to base class constructors
  void compute() { 
    sum = get_x() + get_y(); 
  }

  void print() const {
    std::cout << get_x() << " + " << get_y() << " = " << sum << std::endl;
  }
};

int main() {
    Sum mySum(5,10);
    mySum.compute();
    mySum.print();
   
    return 0;
}
```

- Was passiert, wenn beide Basisklassen eine Funktion mit demselben Namen und derselben Signatur definieren?
    - Im obigen Beispiel: Was wäre, wenn `get_x()` und `get_y()` beide `get_value()` heißen würden?
    - Welche Funktion `get_value()` würde von `compute()` oder `print()` aufgerufen werde? Die aus Klasse `X` oder die aus Klasse `Y`?
    - Der Compiler kann das nicht entscheiden und gibt einen Fehler aus:
        - `error: reference to 'get_value' is ambiguous`
- Die Auflösung erfolgt durch explizite Angabe der Basisklasse mit dem Scope-Resolution-Operator `::`:
    - `X::get_value()` ruft eindeutig die Funktion aus `X` auf
    - `Y::get_value()` ruft eindeutig die Funktion aus `Y` auf

```c++
#include <iostream> // std::cout, std::endl

class X {
protected:
  int x;
public:
  X(int x_ = 0) : x(x_) {}

  int get_value() const {
    return x;
  }
};

class Y {
protected:
  int y;
public:
  Y(int y_ = 0) : y(y_) {}

  int get_value() const {
    return y;
  }
};

class Sum : public X, public Y {
private:
  int sum;
public:
  Sum(int x_ = 0, int y_ = 0) { x = x_; y = y_; }

  void compute() {
    sum = get_value() + get_value();   
    // which get_value() will be called?
  }

  void print() const {
    std::cout << get_value() << " + " << get_value() << " = " << sum;  
    // which get_value() will be called?
  }
};

int main() {
    Sum mySum(5,10);
    mySum.compute();
    mySum.print();
   
    return 0;
}
```

Die folgende korrigierte Version verwendet `X::get_value()` und `Y::get_value()`, um dem Compiler eindeutig mitzuteilen, welche Funktion gemeint ist:

```c++
#include <iostream> // std::cout, std::endl

class X {
protected:
  int x;
public:
  X(int x_ = 0) : x(x_) {}

  int get_value() const {
    return x;
  }
};

class Y {
protected:
  int y;
public:
  Y(int y_ = 0) : y(y_) {}

  int get_value() const {
    return y;
  }
};

class Sum : public X, public Y {
private:
  int sum;
public:
  Sum(int x_ = 0, int y_ = 0) : X(x_), Y(y_) {}

  void compute() {
    sum = X::get_value() + Y::get_value();  // scope resolves ambiguity
  }

  void print() const {
    std::cout << X::get_value() << " + " << Y::get_value() << " = " << sum << std::endl;
  }
};

int main() {
    Sum mySum(5, 10);
    mySum.compute();
    mySum.print();  // 5 + 10 = 15
   
    return 0;
}
```
